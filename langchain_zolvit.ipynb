```python
import os
import json
from PyPDF2 import PdfReader

# Folder path containing PDFs
pdf_folder = "/content/sample_data/Untitled Folder"

# Output JSON file path
output_json = "/content/sample_data/output.json"

# Function to process and extract text from each PDF
def extract_text_from_pdf(pdf_path):
    reader = PdfReader(pdf_path)
    pages_data = []

    for page_num in range(len(reader.pages)):
        page = reader.pages[page_num]
        page_data = {
            "page_number": page_num + 1,
            "content": page.extract_text()
        }
        pages_data.append(page_data)
    return pages_data

# Process all PDFs in the folder
all_pdfs_data = {}
for pdf_file in os.listdir(pdf_folder):
    if pdf_file.endswith(".pdf"):
        pdf_path = os.path.join(pdf_folder, pdf_file)
        print(f"Processing {pdf_file}...")
        pdf_data = extract_text_from_pdf(pdf_path)
        all_pdfs_data[pdf_file] = pdf_data

# Save the extracted data to a JSON file
with open(output_json, "w") as json_file:
    json.dump(all_pdfs_data, json_file, indent=4)

print(f"Extraction complete. Data saved to {output_json}")
```

```python
import re
import json
import pandas as pd

# Example JSON data from PDF
data = {
    "12Jul_Invoice_627_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 627\nDate: 12/07/2024\nPlace of Supply: 33-Tamil NaduBill To:\nVARASAI COMMERCIALS\n1St Floor 22 ANNA PILLAI STREET, \nPARRYS, Georgetown,\nCHENNAI-600 001.\nContact No.: 9884927736\nGSTIN Number: 33AALFV6889P1ZZ\nState: 33-Tamil NaduTransporta\u0000on Details:\nVehicle Number: TN 29 BR 6160\nDelivery Date: 13/07/24\nDelivery Loca\u0000on: Chennai\nShip To:\n1St Floor 22 ANNA PILLAI STREET, \n PARRYS, Georgetown,\n CHENNAI-600 001.\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit Amount\n1 Cane Jaggery Ball 1701 110 Bag \u20b9 1,350.00 \u20b9 1,48,500.00\n2 Cane Jaggery Ball 1701 120 Bag \u20b9 1,360.00 \u20b9 1,63,200.00\n3 Cane Jaggery Powder 1701 30 Bag \u20b9 1,765.00 \u20b9 52,950.00\nTotal 260 \u20b9 3,64,650.00\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nThree Lakh Sixty Four Thousand Six Hundred Fi\u0000y Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 3,64,650.00\nTotal \u20b9 3,64,650.00"
        }
    ],
    "INV-118_Rashu.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-118 Invoice Date: 30 Jan 2024 Due Date: 30 Jan 2024\nCustomer Details:\nRashu\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Vitamin b12 test  350.00 1350.000.00 (0%)350.00\nTaxable Amount \u20b9350.00\nTotal \u20b9350.00\nTotal Items / Qty : 1 / 1.000 Total amount (in words): \u00a0INR Three Hundred And Fifty Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "IN2-PID1327250_signed_compressed.pdf": [
        {
            "page_number": 1,
            "content": "1Sub Total \n(Tax Inclusive)23,892.00\nCGST9 (9%) 1,822.27\nSGST9 (9%) 1,822.27\nTotal \u20b923,892.00\nPayment Made (-) 23,892.00\nBalance Due \u20b90.00\nAuthorized SignatureTotal In Words\nIndian Rupee Twenty-Three Thousand Eight Hundred Ninety-Two \nOnly\nNotes\nThanks for your business.\n100 Cubes Home Interiors\nKarnataka \nIndia\nGSTIN 29CITPK3346L2ZG\n8660089169\nahad.khan@100cubes.in\nTAX INVOICE\n  # : IN2-PID1327250\n  Invoice Date : 06/07/2024\n  Terms : Due on Receipt\n Due Date : 06/07/2024 Place Of Supply : Karnataka (29)\nBill To\nMr. Soumyajit Paul\n# Item & DescriptionHSN\n/SAC Qty RateCGST SGST\nAmount % Amt % Amt\n1 Custom work - Home Interiors 998391 1.00 23,892.00 9% 1,822.27 9% 1,822.27 23,892.00\n"
        }
    ],
    "19Jul_Invoice_632_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 632\nDate: 19/07/2024\nPlace of Supply: 33-Tamil NaduBill To:\nVARASAI COMMERCIALS\n1St Floor 22 ANNA PILLAI STREET, \nPARRYS, Georgetown,\nCHENNAI-600 001.\nContact No.: 9884927736\nGSTIN Number: 33AALFV6889P1ZZ\nState: 33-Tamil NaduTransporta\u0000on Details:\nDelivery Date: 20/07/24\nDelivery Loca\u0000on: Chennai\nShip To:\n1St Floor 22 ANNA PILLAI STREET, \n PARRYS, Georgetown,\n CHENNAI-600 001.\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit Amount\n1 Cane Jaggery Ball 1701 38 Bag \u20b9 2,555.00 \u20b9 97,090.00\nTotal 38 \u20b9 97,090.00\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nNinety Seven Thousand Ninety Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 97,090.00\nTotal \u20b9 97,090.00"
        }
    ],
    "INV-112_Gauri.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-112 Invoice Date: 29 Jan 2024 Due Date: 29 Jan 2024\nCustomer Details:\nGauri\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Niosalic 6 Ointment  125.22\n147.32 (-15%)1 TUB125.2215.03 (12%)140.25\n2Halox ES Ointment 129.02\n151.79 (-15%)1 TUB129.0215.48 (12%)144.50\n3Beltas Tablet - 20 mg  109.21\n124.11 (-12%)3327.6439.32 (12%)366.96\n4Solasafe sunscreen gel spf 50  480.68\n600.85 (-20%)1480.6886.52 (18%)567.20\nTaxable Amount \u20b91,062.56\nCGST\u00a06.0% \u20b934.91\nSGST\u00a06.0% \u20b934.91\nCGST\u00a09.0% \u20b943.26\nSGST\u00a09.0% \u20b943.26\nRound Off 0.09\nTotal\u20b91,219.00\nTotal Discount \u20b9242.09\nTotal Items / Qty : 4 / 6.000 Total amount (in words): \u00a0INR One Thousand, Two Hundred And Nineteen Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "INV-141_Kasturi Kalwar.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-141 Invoice Date: 06 Mar 2024 Due Date: 06 Mar 2024\nCustomer Details:\nKasturi Kalwar\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Acne-UV Gel - spf 50 (50 gm)  620.34\n775.42 (-20%)1 PAC620.34111.66 (18%)732.00\n2Bioderma Sebium Hydra Moisturiser - \n40 ml  865.68\n961.86 (-10%)1865.68155.82 (18%)1,021.50\nTaxable Amount \u20b91,486.02\nCGST\u00a09.0% \u20b9133.74\nSGST\u00a09.0% \u20b9133.74\nRound Off 0.50\nTotal\u20b91,754.00\nTotal Discount \u20b9296.50\nTotal Items / Qty : 2 / 2.000 Total amount (in words): \u00a0INR One Thousand, Seven Hundred And Fifty-Four Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "OD14423-sales_invoice-Pankaj_Sharma_compressed.pdf": [
        {
            "page_number": 1,
            "content": "Opulent Duende\nFourth Floor, Ft.405, Eeco Valley-west block, Madhavapuri Hills,Ameenapur, Near GSM Mall,\nChanda Nagar, Hyderabad, Rangareddy, Hyderabad, Telangana, 500050\nMobile: 9885001313 GSTIN:36BARPS5773E1Z2\nEmail:opulentduende@gmail.com\nInvoice No. :OD14423 Invoice Date :21/08/2024\nBILL TO\nPankaj Sharma\nMobile:9999071129\nSERVICES SAC TAX AMOUN T\nQUEENPRO BOT CREDITS\nLicensing services for the right to use computer software and databases.997331 4,576.27\n(18%)30,000\nSUB TOTAL \u20b9 4,576.27 \u20b9 30,000\nBANK DE TAILS\nName: Opulent Duende\nIFSC Code:UTIB0000538\nAccount No:924020007176709\nBank: Axis Bank,CHANDANAGAR\nPAYMEN T QR CODE\nUPI ID:\nopulentduende@axisbank\nTERMS AND CONDI TIONS\nFees once collected is non-refundable.TAXABLE AMOUNT \u20b9 25,423.73\nCGST @9% \u20b9 2,288.14\nSGST @9% \u20b9 2,288.14\nTOTAL AMOUN T \u20b9 30,000\nReceived Amount \u20b9 30,000\nBalance \u20b9 0\nTotal A mount (in wor ds)\nThirty Thousand RupeesTAX INVOICEORIGINAL FOR RECIPIENT"
        }
    ],
    "INV-802.pdf": [
        {
            "page_number": 1,
            "content": "1\n POWERED BY\nSub Total 9,750.00\nCGST9 (9%) 877.50\nSGST9 (9%) 877.50\nTotal \u20b911,505.00\nBalance Due \u20b911,505.00\nAuthorized SignatureTotal In Words\nIndian Rupee Eleven Thousand Five Hundred Five Only\nNotes\nThanks for your business.\nTerms & Conditions\nGoods Once sold cannot be taken back.\nSRI BHAVANI RUBBER & PLASTICS\nNo, 1, Thilagar Nagar 1st Street, Ashok Sylicate Campus,\nKaladipet, Thiruvottiyur,\nChennai Tamil Nadu 600019\nIndia\nGSTIN 33AFIPM5717R1ZV\nsribhavanirubber.plastics@gmail.com TAX INVOICE\n  # : INV-802\n  Invoice Date : 01/07/2024\n  Terms : Net 15\n Due Date : 16/07/2024 Place Of Supply : Tamil Nadu (33)\nBill To Ship To\nGANESH ENGINEERING WORKS\nChennai\n600019 Tamil Nadu\nGSTIN 33AGDPJ1042J1ZVChennai\n600019 Tamil Nadu\nGSTIN 33AGDPJ1042J1ZV\n# Item & DescriptionHSN\n/SAC Qty RateCGST SGST\nAmount % Amt % Amt\n1 PUWHEEL0001\nPU Wheel - Size: 250 mm * 70 \nmm Thick3921139\n05.00\nNos1,950.00 9% 877.50 9% 877.50 9,750.00"
        }
    ],
    "22Jul_Invoice_634_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 634\nDate: 22/07/2024\nPlace of Supply: 29-KarnatakaBill To:\nSri Mahalakshmi Commercials\nNO.72/2 2ND MAIN ROAD NEW THARAGUPET\nContact No.: 08026705904, 08048903677\nGSTIN Number: 29ADHFS9241H1ZM\nState: 29-KarnatakaTransporta\u0000on Details:\nDelivery Date: 22/07/24\nDelivery Loca\u0000on: Bangalore\nShip To:\nNO.72/2 2ND MAIN ROAD NEW THARAGUPET\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit GST Amount\n1 HTO Jaggery Powder 1701 20 Box \u20b9 590.00 \u20b9 590.00 (5%) \u20b9 12,390.00\nTotal 20 \u20b9 590.00 \u20b9 12,390.00\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nTwelve Thousand Three Hundred Ninety Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 11,800.00\nIGST@5% \u20b9 590.00\nTotal \u20b9 12,390.00"
        }
    ],
    "IN2-100CMBTS01_signed_compressed.pdf": [
        {
            "page_number": 1,
            "content": "1Sub Total \n(Tax Inclusive)19,179.00\nCGST9 (9%) 1,462.81\nSGST9 (9%) 1,462.81\nTotal \u20b919,179.00\nPayment Made (-) 19,179.00\nBalance Due \u20b90.00\nAuthorized SignatureTotal In Words\nIndian Rupee Nineteen Thousand One Hundred Seventy-Nine Only\nNotes\nThanks for your business.\n100 Cubes Home Interiors\nKarnataka \nIndia\nGSTIN 29CITPK3346L2ZG\n9606472571\npatanabdulahadkhan@gmail.com\nTAX INVOICE\n  # : IN2-100CMBTS01\n  Invoice Date : 20/07/2024\n  Terms : Due on Receipt\n Due Date : 20/07/2024 Place Of Supply : Karnataka (29)\nBill To\nMr. Rajesh YS\n# Item & DescriptionHSN\n/SAC Qty RateCGST SGST\nAmount % Amt % Amt\n1 Site measurement stage - \nHome Interiors998391 1.00 19,179.00 9% 1,462.81 9% 1,462.81 19,179.00\n"
        }
    ],
    "FY2407111008609.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n649\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-11\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n19450288PDYDAB218F\n2024-\n07-10\n2024-07-\n11\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n2\n32488243YSL9F1BCC8\n2024-\n07-10\n2024-07-\n11\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n3\n5821442ADPC10351E\n2024-\n07-11\n2024-07-\n11\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n4\n10722768SKJ491C49F\n2024-\n07-10\n2024-07-\n11\n17580784\nWudnKrafts\nSheesham\n(Rosewood)\nCooking\nUtensils\tNo\nVCWKHKCB-\n160\n44190010@12\n1\n499.00\n45.00\n274.45\n134.77\n134.77\n6.00\n8.09\n6.00\n8.09\n150.94\n5\n14615332JOW5F5B282\n2024-\n07-10\n2024-07-\n11\n17580741\nWudnKrafts\n100\tpcs\t12\nColor\nWooden\nDominos\nConstru\nVCWKTYWO-\n117\n95036010@12\n1\n699.00\n40.00\n419.4\n205.96\n205.96\n6.00\n12.36\n6.00\n12.36\n230.68\n6\n32499096UOPA22D6E9\n2024-\n07-11\n2024-07-\n11\n17580762\nWudn\nKrafts\nWooden\nStick\tBaby\nRattles\nwith\t12\nGhun\nVCWKTYWO-\n138\n95036010@12\n1\n199.00\n0.00\n199\n97.72\n97.72\n6.00\n5.86\n6.00\n5.86\n109.45\n7\n9077594VED818D1C4\n2024-\n07-10\n2024-07-\n11\n17580772\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n148\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n7\n3493.00\n966.73\n966.73\n44.82\n44.82\n1056.37\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n5.00\n2\n376.66\n2.50\n9.42\n2.50\n9.42\n18.83\n12.00\n5\n590.07\n6.00\n35.40\n6.00\n35.40\n70.81\nTotal\n7\n966.73\n44.82\n44.82\n89.64\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n44190010\n1\n134.77\n8.09\n8.09\n150.94\n950300\n2\n376.66\n9.42\n9.42\n395.49\n95036010\n4\n455.30\n27.32\n27.32\n509.94\nGrand\tTotal\n7\n966.73\n44.82\n44.82\n1056.37\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n7\n966.73\n44.82\n44.82\n1056.37"
        }
    ],
    "30Jul_Invoice_645_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 645\nDate: 30/07/2024\nPlace of Supply: 33-Tamil NaduBill To:\nGrace Mart Private Limited\nNo.21/4B2, Noombal Road,\nPuliambedu,\nChennai - 600077\nContact No.: 9095334777\nGSTIN Number: 33AAKCG0744Q1Z9\nState: 33-Tamil NaduTransporta\u0000on Details:\nVehicle Number: TN 29 BR 6160\nDelivery Date: 31/07/24\nDelivery Loca\u0000on: Chennai\nShip To:\nNo.21/4B2, Noombal Road,\n Puliambedu,\n Chennai - 600077\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit Amount\n1 Cane Jaggery Ball 1701 100 Bag \u20b9 1,590.00 \u20b9 1,59,000.00\n2 Cane Jaggery Powder 1701 80 Bag \u20b9 1,450.00 \u20b9 1,16,000.00\nTotal 180 \u20b9 2,75,000.00\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nTwo Lakh Seventy Five Thousand Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 2,75,000.00\nTotal \u20b9 2,75,000.00"
        }
    ],
    "FY2407291027782.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n667\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-29\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n10026216RMLDB42ABD\n2024-\n07-29\n2024-07-\n29\n17580769\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n145\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n2\n18593784QSZ6D67BAC\n2024-\n07-29\n2024-07-\n29\n17580780\nWudnKrafts\n3D\tWooden\nCapital\nAlphabet,\nSmall\nAlpha\nVCWKTYWO-\n156\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n3\n32765039WLZB3A19D5\n2024-\n07-28\n2024-07-\n29\n17580778\nWudnKrafts\nWooden\t3D\nCapital\nAlphabets\nand\nNumber\nVCWKTYWO-\n154\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n4\n31895843PIAE4C3E15\n2024-\n07-28\n2024-07-\n29\n17580784\nWudnKrafts\nSheesham\n(Rosewood)\nCooking\nUtensils\tNo\nVCWKHKCB-\n160\n44190010@12\n1\n499.00\n45.00\n274.45\n134.77\n134.77\n6.00\n8.09\n6.00\n8.09\n150.94\n5\n22399237GFBA4AD4EB\n2024-\n07-28\n2024-07-\n29\n17580769\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n145\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n6\n19139880ZYH9EAD513\n2024-\n07-28\n2024-07-\n29\n17580784\nWudnKrafts\nSheesham\n(Rosewood)\nCooking\nUtensils\tNo\nVCWKHKCB-\n160\n44190010@12\n1\n499.00\n45.00\n274.45\n134.77\n134.77\n6.00\n8.09\n6.00\n8.09\n150.94\n7\n31989716KGJ5718247\n2024-\n07-29\n2024-07-\n29\n17580772\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n148\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n8\n25009217HWC69BA213\n2024-\n07-28\n2024-07-\n29\n17580772\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n148\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n9\n2318935EVE6B2FC62\n2024-\n07-29\n2024-07-\n29\n17580728\nWudnKrafts\nWooden\nCalculation\nShelf\nLearning\nand\tE\nVCWKTYWO-\n104\n95036010@12\n1\n799.00\n35.00\n519.35\n255.04\n255.04\n6.00\n15.30\n6.00\n15.30\n285.64\n9\n4191.00\n1249.50\n1249.50\n74.97\n74.97\n1399.44\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n12.00\n9\n1249.5\n6.00\n74.97\n6.00\n74.97\n149.94\nTotal\n9\n1249.50\n74.97\n74.97\n149.94\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n44190010\n2\n269.54\n16.17\n16.17\n301.88\n95036010\n7\n979.96\n58.80\n58.80\n1097.56\nGrand\tTotal\n9\n1249.50\n74.97\n74.97\n1399.44\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n9\n1249.50\n74.97\n74.97\n1399.44"
        }
    ],
    "INV-117_Naman.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-117 Invoice Date: 01 Feb 2024 Due Date: 29 Jan 2024\nCustomer Details:\nNaman\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Kera M 5% Solution 492.86\n616.07 (-20%)1 BTL492.8659.14 (12%)552.00\n2Arachitol Nano (60k) 4*5ml  299.58\n340.43 (-12%)3 BTL898.73107.85 (12%)1,006.58\n3Neurobion Forte - 30 tablets   30.58\n34.75 (-12%)3 STRP91.7316.51 (18%)108.24\nTaxable Amount \u20b91,483.32\nCGST\u00a06.0% \u20b983.50\nSGST\u00a06.0% \u20b983.50\nCGST\u00a09.0% \u20b98.26\nSGST\u00a09.0% \u20b98.26\nRound Off 0.18\nTotal\u20b91,667.00\nTotal Discount \u20b9290.02\nTotal Items / Qty : 3 / 7.000 Total amount (in words): \u00a0INR One Thousand, Six Hundred And Sixty-Seven Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "FY2407091006228.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n647\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-09\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n24987247EEL54EAFE9\n2024-\n07-08\n2024-07-\n09\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n2\n9471271KFPFBFD2FC\n2024-\n07-08\n2024-07-\n09\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n2\n799.00\n55.00\n359.55\n188.33\n376.66\n2.50\n9.42\n2.50\n9.42\n395.49\n3\n17033681CVO2FC4EE8\n2024-\n07-08\n2024-07-\n09\n17580773\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n149\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n4\n12230404KNY33450B2\n2024-\n07-08\n2024-07-\n09\n17580773\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n149\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n5\n21084650LXVEC3B773\n2024-\n07-08\n2024-07-\n09\n17580770\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n146\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n6\n11978895FSI43F8DB3\n2024-\n07-08\n2024-07-\n09\n17580778\nWudnKrafts\nWooden\t3D\nCapital\nAlphabets\nand\nNumber\nVCWKTYWO-\n154\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n7\n21427873SIUB878B1B\n2024-\n07-08\n2024-07-\n09\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n8\n2793.00\n790.51\n978.84\n45.55\n45.55\n1069.93\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n12.00\n6\n602.18\n6.00\n36.13\n6.00\n36.13\n72.26\n5.00\n2\n376.66\n2.50\n9.42\n2.50\n9.42\n18.83\nTotal\n8\n978.84\n45.55\n45.55\n91.09\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n950300\n2\n376.66\n9.42\n9.42\n395.49\n95036010\n6\n602.18\n36.13\n36.13\n674.44\nGrand\tTotal\n8\n978.84\n45.55\n45.55\n1069.93\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n8\n978.84\n45.55\n45.55\n1069.93"
        }
    ],
    "INV-142_Urmila Jangam.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-142 Invoice Date: 07 Mar 2024 Due Date: 07 Mar 2024\nCustomer Details:\nUrmila Jangam\nPh: 9552724364\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Kera XL New Hair Growth Serum  874.58\n1,093.22 (-20%)1 BTL874.58157.42 (18%)1,032.00\nTaxable Amount \u20b9874.58\nCGST\u00a09.0% \u20b978.71\nSGST\u00a09.0% \u20b978.71\nTotal\u20b91,032.00\nTotal Discount \u20b9258.00\nTotal Items / Qty : 1 / 1.000 Total amount (in words): \u00a0INR One Thousand And Thirty-Two Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "CITYMART_SUPERMARKET_TWO_Sales_Invoice_AGNI1004.pdf": [
        {
            "page_number": 1,
            "content": "AGNI IN TEGRA TED F ARM\nNear ganapathy temple,Dhoni, P alakkad, K erala, 678009\nMobile: \u00a0 9004693241 GSTIN: \u00a0 32AOFPA0580C1ZT PAN Number: \u00a0AOFPA0580C\nEmail: \u00a0Ankhulvs@gmail.com\nInvoice No. :AGNI1004 Invoice Date :27/05/2024\nBILL TO\nCITYMART SUPERMARKET TWO\n14/1437, H.M.T OWER, Kalmandapam, Chandr anagar ,\nPalakkad, P alakkad, 678007\nMobile: 8281813003\nGSTIN: 32AA QFC4284N1ZY\nPAN Number: AAQFC4284N\nState: KeralaSHIP TO\nCITYMART SUPERMARKET TWO\n14/1437, H.M.T OWER, Kalmandapam, Chandr anagar ,\nPalakkad, P alakkad, 678007\nITEMS HSN QTY. RATE TAX AMOUN T\nMASALA TAPIOCA CHIPS 19059030 16 K GS 98.21 188.57\n(12%)1,760\nSALTED TAPIOCA CHIPS 19059030 12 K GS 98.21 141.43\n(12%)1,320\nSUB TOTAL 28 \u20b9 330 \u20b9 3,080\nTERMS AND CONDI TIONS\n1. Goods once sold will not be tak en back or ex changed\n2. All disputes ar e subject t o KERAL A jurisdiction onlyTAXABLE AMOUN T \u20b9 2,750\nTOTAL AMOUN T \u20b9 3,080\nReceiv ed Amount \u20b9 0\nTotal A mount (in wor ds)\nThree Thousand Eighty RupeesTAX INVOICEORIGINAL Harvesting Hope Since 2020"
        }
    ],
    "INV-103_Jaiprakash Kumawat.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-103 Invoice Date: 27 Jan 2024 Due Date: 27 Jan 2024\nCustomer Details:\nJaiprakash Kumawat\nPh: 8175992426\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Arachitol Nano (60k) 4*5ml  299.58\n340.43 (-12%)3 BTL898.73107.85 (12%)1,006.58\n2Neurobion Forte - 30 tablets   29.53\n34.75 (-15%)3 STRP88.6015.95 (18%)104.55\nTaxable Amount \u20b9987.33\nCGST\u00a06.0% \u20b953.92\nSGST\u00a06.0% \u20b953.92\nCGST\u00a09.0% \u20b97.97\nSGST\u00a09.0% \u20b97.97\nRound Off -0.13\nTotal\u20b91,111.00\nTotal Discount \u20b9155.71\nTotal Items / Qty : 2 / 6.000 Total amount (in words): \u00a0INR One Thousand, One Hundred And Eleven Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "INV-127_Avik Mallick.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-127 Invoice Date: 23 Feb 2024 Due Date: 23 Feb 2024\nCustomer Details:\nAvik Mallick\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Vitamin b12 test  397.50\n750.00 (-47%)1397.500.00 (0%)397.50\n2HBA1C Test  349.27\n659.00 (-47%)1349.270.00 (0%)349.27\n3Last bill pending  197.00 1197.000.00 (0%)197.00\nTaxable Amount \u20b9943.77\nRound Off 0.23\nTotal \u20b9944.00\nTotal Discount \u20b9662.23\nTotal Items / Qty : 3 / 3.000 Total amount (in words): \u00a0INR Nine Hundred And Forty-Four Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "10Jul_Invoice_623_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 623\nDate: 10/07/2024\nPlace of Supply: 29-KarnatakaBill To:\nSri Mahalakshmi Commercials\nNO.72/2 2ND MAIN ROAD NEW THARAGUPET\nContact No.: 08026705904, 08048903677\nGSTIN Number: 29ADHFS9241H1ZM\nState: 29-KarnatakaTransporta\u0000on Details:\nDelivery Date: 10/07/24\nDelivery Loca\u0000on: Bangalore\nShip To:\nNO.72/2 2ND MAIN ROAD NEW THARAGUPET\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit GST Amount\n1 HTO Jaggery Powder 1701 15 Box \u20b9 590.00 \u20b9 442.50 (5%) \u20b9 9,292.50\nTotal 15 \u20b9 442.50 \u20b9 9,292.50\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nNine Thousand Two Hundred Ninety Three Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 8,850.00\nIGST@5% \u20b9 442.50\nRound o\ufb00 \u20b9 0.50\nTotal \u20b9 9,293.00"
        }
    ],
    "INV-806.pdf": [
        {
            "page_number": 1,
            "content": "1\n POWERED BY\nSub Total 15,500.00\nCGST9 (9%) 1,395.00\nSGST9 (9%) 1,395.00\nTotal \u20b918,290.00\nBalance Due \u20b918,290.00\nAuthorized SignatureTotal In Words\nIndian Rupee Eighteen Thousand Two Hundred Ninety Only\nNotes\nThanks for your business.\nTerms & Conditions\nGoods Once sold cannot be taken back.\nSRI BHAVANI RUBBER & PLASTICS\nNo, 1, Thilagar Nagar 1st Street, Ashok Sylicate Campus,\nKaladipet, Thiruvottiyur,\nChennai Tamil Nadu 600019\nIndia\nGSTIN 33AFIPM5717R1ZV\nsribhavanirubber.plastics@gmail.com TAX INVOICE\n  # : INV-806\n  Invoice Date : 17/07/2024\n  Terms : Due on Receipt\n Due Date : 17/07/2024 Place Of Supply : Tamil Nadu (33)\nBill To Ship To\nExcel Engineering Works\nNew no:1003/2,Old no:326,\nT.H.Road,Old Washermanpet,\nChennai\n600021 Tamil Nadu\nIndia\nGSTIN 33AFPPR8494F1ZSNew no:1003/2,Old no:326,\nT.H.Road,Old Washermanpet,\nChennai\n600021 Tamil Nadu\nIndia\nGSTIN 33AFPPR8494F1ZS\n# Item & DescriptionHSN\n/SAC Qty RateCGST SGST\nAmount % Amt % Amt\n1 BUSHM0015\nBrass Metal Bush, Size: 68 mm \nID * 89 mm OD * 114 mm \nheight 4016939\n02.00\nNos2,500.00 9% 450.00 9% 450.00 5,000.00\n2 BUSHM0014\nBrass Metal Bush, Size: 56 mm \nID * 75 mm OD * 135 mm \nheight 4016939\n06.00\nNos1,750.00 9% 945.00 9% 945.00 10,500.00"
        }
    ],
    "INV-135_Mohith Saragur.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-135 Invoice Date: 01 Mar 2024 Due Date: 01 Mar 2024\nCustomer Details:\nMohith Saragur\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Tab flucon 400mg  22.58\n28.23 (-20%)3 STRP67.7512.19 (18%)79.94\n2Lupizol ZS Shampoo 100 ml  369.91\n420.36 (-12%)1 BTL369.9144.39 (12%)414.30\n3Anaboom AD Lotion - 50 ml  253.56\n288.14 (-12%)1 BTL253.5645.64 (18%)299.20\nTaxable Amount \u20b9691.22\nCGST\u00a06.0% \u20b922.19\nSGST\u00a06.0% \u20b922.19\nCGST\u00a09.0% \u20b928.92\nSGST\u00a09.0% \u20b928.92\nRound Off -0.45\nTotal \u20b9793.00\nTotal Discount \u20b9117.28\nTotal Items / Qty : 3 / 5.000 Total amount (in words): \u00a0INR Seven Hundred And Ninety-Three Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "FY2407231021362.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n661\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-23\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n21316560ESS687BC92\n2024-\n07-22\n2024-07-\n23\n10992045\nZivilion\nPlastic\nGeometric\nShape\nSorting\nBoard\t-\tM\nVCZITYGS-\n16\n950300@12\n1\n699.00\n50.00\n349.5\n171.63\n171.63\n6.00\n10.30\n6.00\n10.30\n192.23\n2\n18950176YSL5B000C1\n2024-\n07-22\n2024-07-\n23\n17580780\nWudnKrafts\n3D\tWooden\nCapital\nAlphabet,\nSmall\nAlpha\nVCWKTYWO-\n156\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n3\n31653178ZMH503412B\n2024-\n07-22\n2024-07-\n23\n17580784\nWudnKrafts\nSheesham\n(Rosewood)\nCooking\nUtensils\tNo\nVCWKHKCB-\n160\n44190010@12\n1\n499.00\n45.00\n274.45\n134.77\n134.77\n6.00\n8.09\n6.00\n8.09\n150.94\n4\n30087877FDHEE7BB7E\n2024-\n07-22\n2024-07-\n23\n17580737\nWudnKrafts\nWooden\nCalendar\nClock\nPuzzle\nWooden\nClo\nVCWKTYWO-\n113\n95036010@12\n1\n699.00\n40.00\n419.4\n205.96\n205.96\n6.00\n12.36\n6.00\n12.36\n230.68\n5\n20352613WLTB74BD35\n2024-\n07-23\n2024-07-\n23\n17580769\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n145\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n5\n2895.00\n811.30\n811.30\n48.68\n48.68\n908.66\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n12.00\n5\n811.3\n6.00\n48.68\n6.00\n48.68\n97.36\nTotal\n5\n811.30\n48.68\n48.68\n97.36\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n44190010\n1\n134.77\n8.09\n8.09\n150.94\n950300\n1\n171.63\n10.30\n10.30\n192.23\n95036010\n3\n504.90\n30.29\n30.29\n565.49\nGrand\tTotal\n5\n811.30\n48.68\n48.68\n908.66\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n5\n811.30\n48.68\n48.68\n908.66"
        }
    ],
    "INV-807.pdf": [
        {
            "page_number": 1,
            "content": "1\n POWERED BY\nSub Total 30,122.50\nCGST9 (9%) 2,711.03\nSGST9 (9%) 2,711.03\nTotal \u20b935,544.56\nBalance Due \u20b935,544.56\nAuthorized SignatureTotal In Words\nIndian Rupee Thirty-Five Thousand Five Hundred Forty-Four and \nFifty-Six Paise Only\nNotes\nThanks for your business.\nTerms & Conditions\nGoods Once sold cannot be taken back.\nSRI BHAVANI RUBBER & PLASTICS\nNo, 1, Thilagar Nagar 1st Street, Ashok Sylicate Campus,\nKaladipet, Thiruvottiyur,\nChennai Tamil Nadu 600019\nIndia\nGSTIN 33AFIPM5717R1ZV\nsribhavanirubber.plastics@gmail.com TAX INVOICE\n  # : INV-807\n  Invoice Date : 21/07/2024\n  Terms : Due on Receipt\n Due Date : 21/07/2024 Place Of Supply : Tamil Nadu (33)\nBill To Ship To\nRUBBERAGE INDUSTRIES\nFirst Floor, No 41,\nSembudoss Street,\nChennai\n600001 Tamil Nadu\nIndia\nGSTIN 33AAAFR0628M1ZBFirst Floor, No 41,\nSembudoss Street,\nChennai\n600001 Tamil Nadu\nIndia\nGSTIN 33AAAFR0628M1ZB\n# Item & DescriptionHSN\n/SAC Qty RateCGST SGST\nAmount % Amt % Amt\n1 DIENO13383\nDie No-13383, Size: 40mm * \n55mm * 7mm Thick Inner \nMetal Ring8480490\n020.00\nNos25.00 9% 45.00 9% 45.00 500.00\n2 DIENO13383ND2\nDie No-13383 New Bottom \nSingle Lip Plate, Size: 40mm * \n55mm * 7mm Thick 8480490\n01.00\nNos200.00 9% 18.00 9% 18.00 200.00\n3 SSFITTINGM00005\nSS Fitting Diaphragm 190 mm \nDia 8480490\n030.00\nNos850.00 9% 2,295.00 9% 2,295.00 25,500.00\n4 ORINGD0032\n'O' Ring, Size: 14.3 mm *  2.4 C\n/S8480490\n01.00\nNos200.00 9% 18.00 9% 18.00 200.00\n5 PINBUSH0002\n4.2 Pin Bush * 73mm Length8480490\n0675.00\nNos3.50 9% 212.63 9% 212.63 2,362.50\n6 DIENO7426\nDie No-7426, Size: 25 * 47 * 78480490\n010.00\nNos30.00 9% 27.00 9% 27.00 300.00\n7 DIENO14214\nDie No-14214, Size: 37 * 62 * 8 \nInner Metal Ring8480490\n08.00\nNos30.00 9% 21.60 9% 21.60 240.00\n8 DIENO14215\nDie No-14215, Size: 35 * 62 * \n10 Inner Metal Ring8480490\n04.00\nNos30.00 9% 10.80 9% 10.80 120.00\n9 DIENO11702\nDie No-11702, Size: 15 * 35 * 7 8480490\n010.00\nNos20.00 9% 18.00 9% 18.00 200.00\n10 DIENO14356\nDie No-14356, Size: 25 * 37 * 5 \nOuter Metal Ring8480490\n020.00\nNos25.00 9% 45.00 9% 45.00 500.00"
        }
    ],
    "FY2407131010876.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n651\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-13\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n4999408JWTE06A081\n2024-\n07-12\n2024-07-\n13\n10992045\nZivilion\nPlastic\nGeometric\nShape\nSorting\nBoard\t-\tM\nVCZITYGS-\n16\n950300@12\n1\n699.00\n50.00\n349.5\n171.63\n171.63\n6.00\n10.30\n6.00\n10.30\n192.23\n2\n25208143RZU8E04607\n2024-\n07-12\n2024-07-\n13\n17580778\nWudnKrafts\nWooden\t3D\nCapital\nAlphabets\nand\nNumber\nVCWKTYWO-\n154\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n3\n31832106KEN22241C7\n2024-\n07-13\n2024-07-\n13\n17580780\nWudnKrafts\n3D\tWooden\nCapital\nAlphabet,\nSmall\nAlpha\nVCWKTYWO-\n156\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n4\n32528483GBU93DCC6F\n2024-\n07-13\n2024-07-\n13\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n5\n8485944GWZ8971FF9\n2024-\n07-12\n2024-07-\n13\n17580773\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n149\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n6\n19626806MXIE330D75\n2024-\n07-13\n2024-07-\n13\n17580728\nWudnKrafts\nWooden\nCalculation\nShelf\nLearning\nand\tE\nVCWKTYWO-\n104\n95036010@12\n1\n799.00\n35.00\n519.35\n255.04\n255.04\n6.00\n15.30\n6.00\n15.30\n285.64\n6\n4094.00\n1149.36\n1149.36\n62.37\n62.37\n1274.10\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n12.00\n5\n961.03\n6.00\n57.66\n6.00\n57.66\n115.32\n5.00\n1\n188.33\n2.50\n4.71\n2.50\n4.71\n9.42\nTotal\n6\n1149.36\n62.37\n62.37\n124.74\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n950300\n2\n359.96\n15.01\n15.01\n389.97\n95036010\n4\n789.40\n47.36\n47.36\n884.13\nGrand\tTotal\n6\n1149.36\n62.37\n62.37\n1274.10\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n6\n1149.36\n62.37\n62.37\n1274.10"
        }
    ],
    "OD14428-sales_invoice-Ravinder_Reddy_Yedulla_compressed.pdf": [
        {
            "page_number": 1,
            "content": "Opulent Duende\nFourth Floor, Ft.405, Eeco Valley-west block, Madhavapuri Hills,Ameenapur, Near GSM Mall,\nChanda Nagar, Hyderabad, Rangareddy, Hyderabad, Telangana, 500050\nMobile: 9885001313 GSTIN:36BARPS5773E1Z2\nEmail:opulentduende@gmail.com\nInvoice No. :OD14428 Invoice Date :28/08/2024\nBILL TO\nRavinder Reddy Yedulla\nMobile:9949515251\nSERVICES SAC TAX AMOUN T\nQUEENPRO BOT CREDITS\nLicensing services for the right to use computer software and databases.997331 1,525.42\n(18%)10,000\nSUB TOTAL \u20b9 1,525.42 \u20b9 10,000\nBANK DE TAILS\nName: Opulent Duende\nIFSC Code:UTIB0000538\nAccount No:924020007176709\nBank: Axis Bank,CHANDANAGAR\nPAYMEN T QR CODE\nUPI ID:\nopulentduende@axisbank\nTERMS AND CONDI TIONS\nFees once collected is non-refundable.TAXABLE AMOUNT \u20b9 8,474.58\nCGST @9% \u20b9 762.71\nSGST @9% \u20b9 762.71\nTOTAL AMOUN T \u20b9 10,000\nReceived Amount \u20b9 10,000\nBalance \u20b9 0\nTotal A mount (in wor ds)\nTen Thousand RupeesTAX INVOICEORIGINAL FOR RECIPIENT"
        }
    ],
    "FY2407281026479.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n666\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-28\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n32750037HRN8DF2F79\n2024-\n07-27\n2024-07-\n28\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n2\n23146411PMC7E2400D\n2024-\n07-27\n2024-07-\n28\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n3\n17053391FREACA3D24\n2024-\n07-28\n2024-07-\n28\n17580742\nWudnKrafts\n120\tPcs\t12\nColor\nWooden\nDomino\nBlocks\tS\nVCWKTYWO-\n118\n95036010@12\n1\n749.00\n40.00\n449.4\n220.69\n220.69\n6.00\n13.24\n6.00\n13.24\n247.17\n4\n9954074IYY68B78E5\n2024-\n07-27\n2024-07-\n28\n17580773\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n149\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n5\n7570148LIZF9C91CF\n2024-\n07-27\n2024-07-\n28\n17580772\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n148\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n6\n7045599CZLC751B42\n2024-\n07-27\n2024-07-\n28\n17580737\nWudnKrafts\nWooden\nCalendar\nClock\nPuzzle\nWooden\nClo\nVCWKTYWO-\n113\n95036010@12\n1\n699.00\n40.00\n419.4\n205.96\n205.96\n6.00\n12.36\n6.00\n12.36\n230.68\n7\n17637941RHQCB80836\n2024-\n07-27\n2024-07-\n28\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n8\n32748704KFH85A2273\n2024-\n07-27\n2024-07-\n28\n17580778\nWudnKrafts\nWooden\t3D\nCapital\nAlphabets\nand\nNumber\nVCWKTYWO-\n154\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n9\n14976729DIM49BBD7D\n2024-\n07-28\n2024-07-\n28\n17580778\nWudnKrafts\nWooden\t3D\nCapital\nAlphabets\nand\nNumber\nVCWKTYWO-\n154\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n10\n19208452UVAC4DFD5E\n2024-\n07-28\n2024-07-\n28\n10992045\nZivilion\nPlastic\nGeometric\nShape\nSorting\nBoard\t-\tM\nVCZITYGS-\n16\n950300@12\n1\n699.00\n50.00\n349.5\n171.63\n171.63\n6.00\n10.30\n6.00\n10.30\n192.23\n10\n6540.00\n1761.15\n1761.15\n85.89\n85.89\n1932.94\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n5.00\n3\n564.99\n2.50\n14.12\n2.50\n14.12\n28.25\n12.00\n7\n1196.16\n6.00\n71.77\n6.00\n71.77\n143.54\nTotal\n10\n1761.15\n85.89\n85.89\n171.79\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n950300\n4\n736.62\n24.42\n24.42\n785.47\n95036010\n6\n1024.53\n61.47\n61.47\n1147.47\nGrand\tTotal\n10\n1761.15\n85.89\n85.89\n1932.94\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n10\n1761.15\n85.89\n85.89\n1932.94"
        }
    ],
    "INV-805.pdf": [
        {
            "page_number": 1,
            "content": "1\n POWERED BY\nSub Total 79,200.00\nCGST9 (9%) 7,128.00\nSGST9 (9%) 7,128.00\nTotal \u20b993,456.00\nBalance Due \u20b993,456.00\nAuthorized SignatureTotal In Words\nIndian Rupee Ninety-Three Thousand Four Hundred Fifty-Six Only\nNotes\nThanks for your business.\nTerms & Conditions\nGoods Once sold cannot be taken back.\nSRI BHAVANI RUBBER & PLASTICS\nNo, 1, Thilagar Nagar 1st Street, Ashok Sylicate Campus,\nKaladipet, Thiruvottiyur,\nChennai Tamil Nadu 600019\nIndia\nGSTIN 33AFIPM5717R1ZV\nsribhavanirubber.plastics@gmail.com TAX INVOICE\n  # : INV-805\n  Invoice Date : 17/07/2024\n  Terms : Net 30\n  Due Date : 16/08/2024\n P.O.# : 226274 Place Of Supply : Tamil Nadu (33)\nBill To Ship To\nSRF LIMITED\nK-1, Sipcot Industrial Area,\nGummudipondi\n601201 Tamilnadu\nIndia\nGSTIN 33AAACS0206P1ZKK-1, Sipcot Industrial Area,\nGummudipondi\n601201 Tamilnadu\nIndia\nGSTIN 33AAACS0206P1ZK\n# Item & DescriptionHSN\n/SAC Qty RateCGST SGST\nAmount % Amt % Amt\n1 TEXST00006\n\r\nSPONGE PAD FOR TO4\nSize: ID 110 mm * OD 305 mm \n* 25 mm Thick8482101\n11,100.00\nNos72.00 9% 7,128.00 9% 7,128.00 79,200.00"
        }
    ],
    "IN3-PID2529864 (1)_compressed.pdf": [
        {
            "page_number": 1,
            "content": "1Sub Total \n(Tax Inclusive)1,43,569.00\nCGST9 (9%) 10,950.18\nSGST9 (9%) 10,950.18\nTotal \u20b91,43,569.00\nPayment Made (-) 1,43,569.00\nBalance Due \u20b90.00\nAuthorized SignatureTotal In Words\nIndian Rupee One Lakh Forty-Three Thousand Five Hundred Sixty-\nNine Only\nNotes\nThanks for your Patronage.\n100 Cubes Home Interiors\nKarnataka \nIndia\nGSTIN 29CITPK3346L2ZG\n9606472571\npatanabdulahadkhan@gmail.com\nTAX INVOICE\n  # : IN3-PID2529864\n  Invoice Date : 21/07/2024\n  Terms : Due on Receipt\n Due Date : 21/07/2024 Place Of Supply : Karnataka (29)\nBill To\nMr. Raja Shankar\n# Item & DescriptionHSN\n/SAC Qty RateCGST SGST\nAmount % Amt % Amt\n1 100% Stage payment - Custom \nworks998391 1.00 1,43,569.0\n09% 10,950.18 9% 10,950.18 1,43,569.00"
        }
    ],
    "OD14421-sales_invoice-Sachin_Shilwant_compressed.pdf": [
        {
            "page_number": 1,
            "content": "Opulent Duende\nFourth Floor, Ft.405, Eeco Valley-west block, Madhavapuri Hills,Ameenapur, Near GSM Mall,\nChanda Nagar, Hyderabad, Rangareddy, Hyderabad, Telangana, 500050\nMobile: 9885001313 GSTIN:36BARPS5773E1Z2\nEmail:opulentduende@gmail.com\nInvoice No. :OD14421 Invoice Date :21/08/2024\nBILL TO\nSachin Shilwant\nMobile:9581855196\nSERVICES SAC TAX AMOUN T\nQUEENPRO BOT CREDITS\nLicensing services for the right to use computer software and databases.997331 3,050.85\n(18%)20,000\nSUB TOTAL \u20b9 3,050.85 \u20b9 20,000\nBANK DE TAILS\nName: Opulent Duende\nIFSC Code:UTIB0000538\nAccount No:924020007176709\nBank: Axis Bank,CHANDANAGAR\nPAYMEN T QR CODE\nUPI ID:\nopulentduende@axisbank\nTERMS AND CONDI TIONS\nFees once collected is non-refundable.TAXABLE AMOUNT \u20b9 16,949.15\nCGST @9% \u20b9 1,525.42\nSGST @9% \u20b9 1,525.42\nTOTAL AMOUN T \u20b9 20,000\nReceived Amount \u20b9 20,000\nBalance \u20b9 0\nTotal A mount (in wor ds)\nTwenty Thousand RupeesTAX INVOICEORIGINAL FOR RECIPIENT"
        }
    ],
    "03Jul_Invoice_615_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 615\nDate: 03/07/2024\nPlace of Supply: 29-KarnatakaBill To:\nSri Mahalakshmi Enterprises\nNO.10/41 APMC YARD 6TH MAIN ROAD YESHWANTHPUR\nContact No.: 08026705904, 08048903677\nGSTIN Number: 29AGWPA9333Q1ZE\nState: 29-KarnatakaTransporta\u0000on Details:\nDelivery Date: 03/07/24\nDelivery Loca\u0000on: Yeshwanthpur\nShip To:\nNO.10/41 APMC YARD 6TH MAIN ROAD YESHWANTHPUR\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit GST Amount\n1 HTO Jaggery Powder 1701 30 Box \u20b9 590.00 \u20b9 885.00 (5%) \u20b9 18,585.00\nTotal 30 \u20b9 885.00 \u20b9 18,585.00\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nEighteen Thousand Five Hundred Eighty Five Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 17,700.00\nIGST@5% \u20b9 885.00\nTotal \u20b9 18,585.00"
        }
    ],
    "OD14419-sales_invoice-Joyce_Supriya_Sakani_compressed.pdf": [
        {
            "page_number": 1,
            "content": "Opulent Duende\nFourth Floor, Ft.405, Eeco Valley-west block, Madhavapuri Hills,Ameenapur, Near GSM Mall,\nChanda Nagar, Hyderabad, Rangareddy, Hyderabad, Telangana, 500050\nMobile: 9885001313 GSTIN:36BARPS5773E1Z2\nEmail:opulentduende@gmail.com\nInvoice No. :OD14419 Invoice Date :15/08/2024\nBILL TO\nJoyce Supriya Sakani\nMobile:9885134663\nSERVICES SAC TAX AMOUN T\nQUEENPRO BOT CREDITS\nLicensing services for the right to use computer software and databases.997331 762.71\n(18%)5,000\nSUB TOTAL \u20b9 762.71 \u20b9 5,000\nBANK DE TAILS\nName: Opulent Duende\nIFSC Code:UTIB0000538\nAccount No:924020007176709\nBank: Axis Bank,CHANDANAGAR\nPAYMEN T QR CODE\nUPI ID:\nopulentduende@axisbank\nTERMS AND CONDI TIONS\nFees once collected is non-refundable.TAXABLE AMOUNT \u20b9 4,237.29\nCGST @9% \u20b9 381.36\nSGST @9% \u20b9 381.36\nTOTAL AMOUN T \u20b9 5,000\nReceived Amount \u20b9 5,000\nBalance \u20b9 0\nTotal A mount (in wor ds)\nFive Thousand RupeesTAX INVOICEORIGINAL FOR RECIPIENT"
        }
    ],
    "23Jul_636_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 636\nDate: 23/07/2024\nPlace of Supply: 29-KarnatakaBill To:\nH.N.Anjanayache\u0000y\n466/13 BAZAR ROAD BANGRPET\nGSTIN Number: 29AADFH2874G1Z8\nState: 29-KarnatakaTransporta\u0000on Details:\nVehicle Number: Tn 76 Z 6768\nDelivery Date: 24/07/24\nDelivery Loca\u0000on: Bangarpet\nShip To:\n466/13 BAZAR ROAD BANGARPET\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit GST Amount\n1 HTO Jaggery Powder 1701 20 Box \u20b9 572.00 \u20b9 572.00 (5%) \u20b9 12,012.00\nTotal 20 \u20b9 572.00 \u20b9 12,012.00\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nTwelve Thousand Twelve Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 11,440.00\nIGST@5% \u20b9 572.00\nTotal \u20b9 12,012.00"
        }
    ],
    "17Jul_Invoice_630_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 630\nDate: 17/07/2024\nPlace of Supply: 29-KarnatakaBill To:\nSri Mahalakshmi Enterprises\nNO.10/41 APMC YARD 6TH MAIN ROAD YESHWANTHPUR\nContact No.: 08026705904, 08048903677\nGSTIN Number: 29AGWPA9333Q1ZE\nState: 29-KarnatakaTransporta\u0000on Details:\nDelivery Date: 17/07/24\nDelivery Loca\u0000on: Yeshwanthpur\nShip To:\nNO.10/41 APMC YARD 6TH MAIN ROAD YESHWANTHPUR\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit GST Amount\n1 Cane Jaggery Ball 1701 2 Bag \u20b9 1,390.00 \u20b9 0.00 (0%) \u20b9 2,780.00\n2 HTO Cane Jaggery Powder 1701 25 Bag \u20b9 590.00 \u20b9 737.50 (5%) \u20b9 15,487.50\nTotal 27 \u20b9 737.50 \u20b9 18,267.50\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nEighteen Thousand Two Hundred Sixty Eight Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 17,530.00\nIGST@5% \u20b9 737.50\nRound o\ufb00 \u20b9 0.50\nTotal \u20b9 18,268.00"
        }
    ],
    "FY240702998718.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n640\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-02\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n23983147PTSC2FC379\n2024-\n07-01\n2024-07-\n02\n14546408\nZivilion\nEducational\nShape\nSorting\nCylinder\t-\n12\tS\nVCZITYHT-\n39\n950300@5\n1\n899.00\n55.00\n404.55\n211.91\n211.91\n2.50\n5.30\n2.50\n5.30\n222.51\n2\n5223118KZR66564A2\n2024-\n07-01\n2024-07-\n02\n17580737\nWudnKrafts\nWooden\nCalendar\nClock\nPuzzle\nWooden\nClo\nVCWKTYWO-\n113\n95036010@12\n1\n699.00\n40.00\n419.4\n205.96\n205.96\n6.00\n12.36\n6.00\n12.36\n230.68\n3\n21176008BJUBBBA9AF\n2024-\n07-02\n2024-07-\n02\n17580779\nWudnKrafts\nWooden\t3D\nAlphabet\nand\nNumber\nBoard\tCom\nVCWKTYWO-\n155\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n4\n25537186DKM326AFEC\n2024-\n07-01\n2024-07-\n02\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n5\n24878315KTJEC6C728\n2024-\n07-01\n2024-07-\n02\n17580769\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n145\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n5\n2895.00\n804.91\n804.91\n40.88\n40.88\n886.67\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n5.00\n1\n211.91\n2.50\n5.30\n2.50\n5.30\n10.60\n12.00\n4\n593\n6.00\n35.58\n6.00\n35.58\n71.16\nTotal\n5\n804.91\n40.88\n40.88\n81.76\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n950300\n1\n211.91\n5.30\n5.30\n222.51\n95036010\n4\n593.00\n35.58\n35.58\n664.16\nGrand\tTotal\n5\n804.91\n40.88\n40.88\n886.67\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n5\n804.91\n40.88\n40.88\n886.67"
        }
    ],
    "08Jul_Invoice_621_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 621\nDate: 08/07/2024\nPlace of Supply: 33-Tamil NaduBill To:\nLAKSHMI SAI COMMERCIALS\n22 ANNAPILLAI STREET,\nParrys, Georgetown,\nCHENNAI-600 001\nContact No.: 9884927736\nGSTIN Number: 33AJXPK0990M1ZI\nState: 33-Tamil NaduTransporta\u0000on Details:\nVehicle Number: TN 29 BR 6160\nDelivery Date: 09/07/24\nDelivery Loca\u0000on: Chennai\nShip To:\n22 ANNAPILLAI STREET,\n Parrys, Georgetown,\n CHENNAI-600 001\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit Amount\n1 Cane Jaggery Ball 1701 84 Bag \u20b9 1,500.00 \u20b9 1,26,000.00\n2 Cane Jaggery Ball 1701 10 Bag \u20b9 2,500.00 \u20b9 25,000.00\n3 Cane Jaggery Powder 1701 40 Bag \u20b9 1,450.00 \u20b9 58,000.00\nTotal 134 \u20b9 2,09,000.00\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nTwo Lakh Nine Thousand Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 2,09,000.00\nTotal \u20b9 2,09,000.00"
        }
    ],
    "08Jul_Invoice_619_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 619\nDate: 08/07/2024\nPlace of Supply: 29-KarnatakaBill To:\nSri Mahalakshmi Enterprises\nNO.10/41 APMC YARD 6TH MAIN ROAD YESHWANTHPUR\nContact No.: 08026705904, 08048903677\nGSTIN Number: 29AGWPA9333Q1ZE\nState: 29-KarnatakaTransporta\u0000on Details:\nDelivery Date: 08/07/24\nDelivery Loca\u0000on: Yeshwanthpur\nShip To:\nNO.10/41 APMC YARD 6TH MAIN ROAD YESHWANTHPUR\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit GST Amount\n1 HTO Jaggery Powder 1701 35 Box \u20b9 590.00\u20b9 1,032.50 (5%) \u20b9 21,682.50\nTotal 35 \u20b9 1,032.50 \u20b9 21,682.50\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nTwenty One Thousand Six Hundred Eighty Three Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 20,650.00\nIGST@5% \u20b9 1,032.50\nRound o\ufb00 \u20b9 0.50\nTotal \u20b9 21,683.00"
        }
    ],
    "03Jul_Invoice_616_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 616\nDate: 03/07/2024\nPlace of Supply: 29-KarnatakaBill To:\nSri Mahalakshmi Commercials\nNO.72/2 2ND MAIN ROAD NEW THARAGUPET\nContact No.: 08026705904, 08048903677\nGSTIN Number: 29ADHFS9241H1ZM\nState: 29-KarnatakaTransporta\u0000on Details:\nDelivery Date: 03/07/24\nDelivery Loca\u0000on: Bangaore\nShip To:\nNO.72/2 2ND MAIN ROAD NEW THARAGUPET\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit GST Amount\n1 HTO Jaggery Powder 1701 10 Box \u20b9 590.00 \u20b9 295.00 (5%) \u20b9 6,195.00\nTotal 10 \u20b9 295.00 \u20b9 6,195.00\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nSix Thousand One Hundred Ninety Five Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 5,900.00\nIGST@5% \u20b9 295.00\nTotal \u20b9 6,195.00"
        }
    ],
    "INV-98_Monika.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-98 Invoice Date: 16 Jan 2024 Due Date: 16 Jan 2024\nCustomer Details:\nMonika\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Acne-UV Gel - spf 50 (50 gm)  620.34\n775.42 (-20%)1 PAC620.34111.66 (18%)732.00\n2Keraglo- AD shampoo  237.29\n296.61 (-20%)1 UNT237.2942.71 (18%)280.00\n3Bioderma Pigmentbio C-concentrate  2,363.64\n2,626.27 (-10%)1 PCS2,363.64425.46 (18%)2,789.10\n4Cetaphil gentle cleansing lotion - 250 \nML 458.64\n521.19 (-12%)1458.6482.56 (18%)541.20\n5Anaboom AD Lotion - 50 ml  241.63\n274.58 (-12%)1 BTL241.6343.49 (18%)285.12\nTaxable Amount \u20b93,921.54\nCGST\u00a09.0% \u20b9352.94\nSGST\u00a09.0% \u20b9352.94\nRound Off -0.42\nTotal\u20b94,627.00\nTotal Discount \u20b9675.58\nTotal Items / Qty : 5 / 5.000 Total amount (in words): \u00a0INR Four Thousand, Six Hundred And Twenty-Seven Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "IN2-100C24022601_signed_compressed.pdf": [
        {
            "page_number": 1,
            "content": "1Sub Total \n(Tax Inclusive)15,000.00\nCGST9 (9%) 1,144.07\nSGST9 (9%) 1,144.07\nTotal \u20b915,000.00\nPayment Made (-) 15,000.00\nBalance Due \u20b90.00\nAuthorized SignatureTotal In Words\nIndian Rupee Fifteen Thousand Only\nNotes\nThanks for your business.\n100 Cubes Home Interiors\nKarnataka \nIndia\nGSTIN 29CITPK3346L2ZG\n9606472571\npatanabdulahadkhan@gmail.com\nTAX INVOICE\n  # : IN2-100C24022601\n  Invoice Date : 16/07/2024\n  Terms : Due on Receipt\n Due Date : 16/07/2024 Place Of Supply : Karnataka (29)\nBill To\nMr. Raju Gandhi\n# Item & DescriptionHSN\n/SAC Qty RateCGST SGST\nAmount % Amt % Amt\n1 Site measurement fee - Home \nInteriors998391 1.00 15,000.00 9% 1,144.07 9% 1,144.07 15,000.00\n"
        }
    ],
    "OD14424-sales_invoice-Mukesh_Choudhari_compressed.pdf": [
        {
            "page_number": 1,
            "content": "Opulent Duende\nFourth Floor, Ft.405, Eeco Valley-west block, Madhavapuri Hills,Ameenapur, Near GSM Mall,\nChanda Nagar, Hyderabad, Rangareddy, Hyderabad, Telangana, 500050\nMobile: 9885001313 GSTIN:36BARPS5773E1Z2\nEmail:opulentduende@gmail.com\nInvoice No. :OD14424 Invoice Date :24/08/2024\nBILL TO\nMukesh Choudhari\nMobile:9765988736\nSERVICES SAC TAX AMOUN T\nQUEENPRO BOT CREDITS\nLicensing services for the right to use computer software and databases.997331 762.71\n(18%)5,000\nSUB TOTAL \u20b9 762.71 \u20b9 5,000\nBANK DE TAILS\nName: Opulent Duende\nIFSC Code:UTIB0000538\nAccount No:924020007176709\nBank: Axis Bank,CHANDANAGAR\nPAYMEN T QR CODE\nUPI ID:\nopulentduende@axisbank\nTERMS AND CONDI TIONS\nFees once collected is non-refundable.TAXABLE AMOUNT \u20b9 4,237.29\nCGST @9% \u20b9 381.36\nSGST @9% \u20b9 381.36\nTOTAL AMOUN T \u20b9 5,000\nReceived Amount \u20b9 5,000\nBalance \u20b9 0\nTotal A mount (in wor ds)\nFive Thousand RupeesTAX INVOICEORIGINAL FOR RECIPIENT"
        }
    ],
    "INV-803.pdf": [
        {
            "page_number": 1,
            "content": "1\n POWERED BY\nSub Total 13,530.00\nCGST9 (9%) 1,217.70\nSGST9 (9%) 1,217.70\nTotal \u20b915,965.40\nBalance Due \u20b915,965.40\nAuthorized SignatureTotal In Words\nIndian Rupee Fifteen Thousand Nine Hundred Sixty-Five and Forty \nPaise Only\nNotes\nThanks for your business.\nTerms & Conditions\nGoods Once sold cannot be taken back.\nSRI BHAVANI RUBBER & PLASTICS\nNo, 1, Thilagar Nagar 1st Street, Ashok Sylicate Campus,\nKaladipet, Thiruvottiyur,\nChennai Tamil Nadu 600019\nIndia\nGSTIN 33AFIPM5717R1ZV\nsribhavanirubber.plastics@gmail.com TAX INVOICE\n  # : INV-803\n  Invoice Date : 01/07/2024\n  Terms : Due on Receipt\n  Due Date : 01/07/2024\n P.O.# : 223247 Place Of Supply : Tamil Nadu (33)\nBill To Ship To\nSRF Limited\nManali Industrial Area, \nNedunchezhiyan Street, Manali,\nChennai\n600068 Tamilnadu\nIndia\nGSTIN 33AAACS0206P1ZKManali Industrial Area, \nNedunchezhiyan Street, Manali,\nChennai\n600068 Tamilnadu\nIndia\nGSTIN 33AAACS0206P1ZK\n# Item & DescriptionHSN\n/SAC Qty RateCGST SGST\nAmount % Amt % Amt\n1 82290-000918\nTex Foam Pad with OD 232mm \nID 90mm and thickness 50mm8483101\n096.00\nNos60.00 9% 518.40 9% 518.40 5,760.00\n2 82290-001012\nFoam Pad, Size: 110mm ID * \n230mm OD * 20mm thick8483101\n0222.00\nNos35.00 9% 699.30 9% 699.30 7,770.00"
        }
    ],
    "INV-147_Divya Suhane.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-147 Invoice Date: 29 Mar 2024 Due Date: 29 Mar 2024\nCustomer Details:\nDivya Suhane\nPh: 6261616609\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Solasafe sunscreen gel spf 50  450.64\n600.85 (-25%)1450.6481.11 (18%)531.75\n2Trapic 500 TA - 10  163.09\n185.33 (-12%)3 STRP489.2824.46 (5%)513.74\n3Depiwhite Advanced Cream - 15 ml  495.93\n563.56 (-12%)1 TUB495.9389.27 (18%)585.20\n4Fixderma Shadow SPF 50+ Cream  410.17\n466.10 (-12%)1 TUB410.1773.83 (18%)484.00\n5Sortet F 16 mg - 10 cap  316.80\n360.00 (-12%)6 STRP1,900.800.00 (0%)1,900.80\nTaxable Amount \u20b93,746.82\nCGST\u00a02.5% \u20b912.23\nSGST\u00a02.5% \u20b912.23\nCGST\u00a09.0% \u20b9122.11\nSGST\u00a09.0% \u20b9122.11\nRound Off -0.49\nTotal\u20b94,015.00\nTotal Discount \u20b9652.31\nTotal Items / Qty : 5 / 12.000 Total amount (in words): \u00a0INR Four Thousand And Fifteen Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "FY2407191017209.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n657\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-19\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n21663194DFJ80A6004\n2024-\n07-18\n2024-07-\n19\n17580772\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n148\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n2\n31978811NZH2FA9188\n2024-\n07-18\n2024-07-\n19\n10992045\nZivilion\nPlastic\nGeometric\nShape\nSorting\nBoard\t-\tM\nVCZITYGS-\n16\n950300@12\n1\n699.00\n50.00\n349.5\n171.63\n171.63\n6.00\n10.30\n6.00\n10.30\n192.23\n3\n24940457KTECEC3BEA\n2024-\n07-18\n2024-07-\n19\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n4\n22117229PDLA508486\n2024-\n07-18\n2024-07-\n19\n17580780\nWudnKrafts\n3D\tWooden\nCapital\nAlphabet,\nSmall\nAlpha\nVCWKTYWO-\n156\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n5\n24940457KEU63358AA\n2024-\n07-19\n2024-07-\n19\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n6\n17195740JDSF529684\n2024-\n07-18\n2024-07-\n19\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n7\n11490961XVK98B2D8B\n2024-\n07-18\n2024-07-\n19\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n7\n4893.00\n1223.89\n1223.89\n47.07\n47.07\n1318.02\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n12.00\n3\n470.57\n6.00\n28.23\n6.00\n28.23\n56.47\n5.00\n4\n753.32\n2.50\n18.83\n2.50\n18.83\n37.67\nTotal\n7\n1223.89\n47.07\n47.07\n94.13\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n950300\n5\n924.95\n29.13\n29.13\n983.21\n95036010\n2\n298.94\n17.94\n17.94\n334.81\nGrand\tTotal\n7\n1223.89\n47.07\n47.07\n1318.02\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n7\n1223.89\n47.07\n47.07\n1318.02"
        }
    ],
    "20.pdf": [
        {
            "page_number": 1,
            "content": "Campos Technologies\nF25-26, 1st Floor, Alfran Plaza, Opp. Naturals Ice Cream,\nNr. Don Bosco School, M.G Road, Panaji Goa 403001\nPhone no. : 9145511772\nEmail : campostechgoa@gmail.com\nGSTIN : 30BHLPC5928C2ZS\nState: 30-Goa\nInvoice\nBill To\nUA DIGITAL FUTURES LAB (OPC) Pvt Ltd\nHouse No 673 Bodiem Tivim Bardez Bodiem\nGSTIN : 30AADCU4370M1ZA\nState: 30-GoaInvoice Details\nInvoice No. : CT/2024-25/20\nDate : 11-07-2024\nPlace of supply: 30-Goa\n# Item nameHSN/\nSACMRP Quantity UnitPrice/\nUnitGST Amount\n1HP 1030 x360 Laptop\n(Renewed HP EliteBook 1030\nx360 Laptop, Core i5 8th Gen,\n8 GB RAM, 256 GB SSD, 13.3\ninch Touch Screen, Windows\n10 Pro, 64Bit Ready, Silver)84713020\u20b9\n1,40,480.001 Pcs\u20b9\n36,000.00\u20b9 6,480.00\n(18%)\u20b9\n42,480.00\n\u00a0\n\u00a0\n\u00a0\n\u00a0\nTotal 1\u20b9\n6,480.00\u20b9\n42,480.00\nDescription\nBank Details:\nM/s CAMPOS TECHNOLOGIES\nA/C No. 0346139563,\nKotak Mahindra Bank,\nIFSC:\u00a0KKBK0002495\nInvoice Amount In Words\nForty Two Thousand Four Hundred Eighty Rupees only\nTerms and Conditions\nThank you for your business!Sub Total \u20b9 36,000.00\nSGST@9% \u20b9 3,240.00\nCGST@9% \u20b9 3,240.00\nTotal \u20b9 42,480.00\nReceived \u20b9 0.00\nBalance \u20b9 42,480.00\nPayment mode Credit\nYou Saved \u20b9 98,000.00"
        },
        {
            "page_number": 2,
            "content": "For :Campos Technologies\nAuthorized SignatoryCampos Technologies\nF25-26, 1st Floor, Alfran Plaza, Opp. Naturals Ice Cream,\nNr. Don Bosco School, M.G Road, Panaji Goa 403001\nPhone no. : 9145511772\nEmail : campostechgoa@gmail.com\nGSTIN : 30BHLPC5928C2ZS\nState: 30-Goa\n"
        }
    ],
    "21.pdf": [
        {
            "page_number": 1,
            "content": "Campos Technologies\nF25-26, 1st Floor, Alfran Plaza, Opp. Naturals Ice Cream,\nNr. Don Bosco School, M.G Road, Panaji Goa 403001\nPhone no. : 9145511772\nEmail : campostechgoa@gmail.com\nGSTIN : 30BHLPC5928C2ZS\nState: 30-Goa\nInvoice\nBill To\nUA DIGITAL FUTURES LAB (OPC) Pvt Ltd\nHouse No 673 Bodiem Tivim Bardez Bodiem\nGSTIN : 30AADCU4370M1ZA\nState: 30-GoaInvoice Details\nInvoice No. : CT/2024-25/21\nDate : 15-07-2024\nPlace of supply: 30-Goa\n# Item nameHSN/\nSACMRP Quantity UnitPrice/\nUnitGST Amount\n1Dell 5450\n(Renewed laptop, Dell\nLatitude 5450, Core i5 5th\nGen, 8 GB RAM, 256 GB SD,\nBlack, Backlit Keyboard,\nWindows 10 Pro 64 bit, 14\"\nDisplay Size.)84713020\u20b9\n70,000.001 Pcs\u20b9\n18,000.00\u20b9 3,240.00\n(18%)\u20b9\n21,240.00\n\u00a0\n\u00a0\n\u00a0\n\u00a0\nTotal 1\u20b9\n3,240.00\u20b9\n21,240.00\nInvoice Amount In Words\nTwenty One Thousand Two Hundred Forty Rupees only\nTerms and Conditions\nThank you for your business!Sub Total \u20b9 18,000.00\nSGST@9% \u20b9 1,620.00\nCGST@9% \u20b9 1,620.00\nTotal \u20b9 21,240.00\nReceived \u20b9 0.00\nBalance \u20b9 21,240.00\nPayment mode Credit\nYou Saved \u20b9 48,760.00"
        },
        {
            "page_number": 2,
            "content": "For :Campos Technologies\nAuthorized SignatoryCampos Technologies\nF25-26, 1st Floor, Alfran Plaza, Opp. Naturals Ice Cream,\nNr. Don Bosco School, M.G Road, Panaji Goa 403001\nPhone no. : 9145511772\nEmail : campostechgoa@gmail.com\nGSTIN : 30BHLPC5928C2ZS\nState: 30-Goa\n"
        }
    ],
    "IN3-100C24022601_signed_compressed.pdf": [
        {
            "page_number": 1,
            "content": "1Sub Total \n(Tax Inclusive)2,00,000.00\nCGST9 (9%) 15,254.24\nSGST9 (9%) 15,254.24\nTotal \u20b92,00,000.00\nPayment Made (-) 2,00,000.00\nBalance Due \u20b90.00\nAuthorized SignatureTotal In Words\nIndian Rupee Two Lakh Only\nNotes\nThanks for your business.\n100 Cubes Home Interiors\nKarnataka \nIndia\nGSTIN 29CITPK3346L2ZG\n9606472571\npatanabdulahadkhan@gmail.com\nTAX INVOICE\n  # : IN3-100C24022601\n  Invoice Date : 10/07/2024\n  Terms : Due on Receipt\n Due Date : 10/07/2024 Place Of Supply : Karnataka (29)\nBill To\nMr. Raju Gandhi\n# Item & DescriptionHSN\n/SAC Qty RateCGST SGST\nAmount % Amt % Amt\n1 50% Stage payment- Home\nInteriors998391 1.00 2,00,000.0\n09% 15,254.24 9% 15,254.24 2,00,000.00\n"
        }
    ],
    "RA29-PRA-OD2-JULY24_compressed.pdf": [
        {
            "page_number": 1,
            "content": "Details HSN Code Uom Quantity Rate Amount\n1Drone Videography-4k-60fps- Video Editing Charges- \n(Markups & callouts as provided by client etc)998387 Km 32.5 600       19,500                          \n3,510                            \n23,010.0                      \nAuthorized Signatory\nG Lakshmi Narayana`IGST at 18%\nTotal\nRupees Twenty Three Thousand Ten OnlyTo\nM/s PRA India Private Limited\nRamaswami Temple Marg, Plot no 996, Mauda Kariar Town, Naupada, Odisha- \n766107GST Number:21AAKCP4877D1ZETAX INVOICE\nInvoice Number: RA29/OD-2\nInvoice Date: 1-JULY-2024\nFrom:\nJarvicons Private Ltd\nNo-34, Jhansi Complex, Opposite to Transport Office.\nTirupati-517501GST Number: 37AAFCJ2376N1Z1\nAcc Name: Jarvicons Private Limited\nAcc. No. 630905038559\nBank: ICICI\nIFSC Code: ICIC0006309\nEmail- Gln@jarvicons.io/ Ph no- +91-9848761575/ Web: www.jarvicons.io\n"
        }
    ],
    "INV-129_Divya Suhane.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-129 Invoice Date: 23 Feb 2024 Due Date: 23 Feb 2024\nCustomer Details:\nDivya Suhane\nPh: 6261616609\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1sotret nf 16 mg - 10 capsules  282.86\n321.43 (-12%)3 STRP848.57101.83 (12%)950.40\n2Ekran Aqua Sunscreen Spf 30  268.47\n305.08 (-12%)1 PAC268.4748.33 (18%)316.80\nTaxable Amount \u20b91,117.05\nCGST\u00a06.0% \u20b950.91\nSGST\u00a06.0% \u20b950.91\nCGST\u00a09.0% \u20b924.16\nSGST\u00a09.0% \u20b924.16\nRound Off -0.20\nTotal\u20b91,267.00\nTotal Discount \u20b9172.80\nTotal Items / Qty : 2 / 4.000 Total amount (in words): \u00a0INR One Thousand, Two Hundred And Sixty-Seven Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "INV-114_Vaibhav Bhagat.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-114 Invoice Date: 29 Jan 2024 Due Date: 29 Jan 2024\nCustomer Details:\nVaibhav Bhagat\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Acne-UV Gel - spf 50 (50 gm)  620.34\n775.42 (-20%)1 PAC620.34111.66 (18%)732.00\n2Neurobion Forte - 30 tablets   29.53\n34.75 (-15%)3 STRP88.6015.95 (18%)104.55\n3Arachitol Nano (60k) 4*5ml  299.58\n340.43 (-12%)3 BTL898.73107.85 (12%)1,006.58\nTaxable Amount \u20b91,607.67\nCGST\u00a06.0% \u20b953.92\nSGST\u00a06.0% \u20b953.92\nCGST\u00a09.0% \u20b963.80\nSGST\u00a09.0% \u20b963.80\nRound Off -0.13\nTotal\u20b91,843.00\nTotal Discount \u20b9338.71\nTotal Items / Qty : 3 / 7.000 Total amount (in words): \u00a0INR One Thousand, Eight Hundred And Forty-Three Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "16.pdf": [
        {
            "page_number": 1,
            "content": "Campos Technologies\nF25-26, 1st Floor, Alfran Plaza, Opp. Naturals Ice Cream,\nNr. Don Bosco School, M.G Road, Panaji Goa 403001\nPhone no. : 9145511772\nEmail : campostechgoa@gmail.com\nGSTIN : 30BHLPC5928C2ZS\nState: 30-Goa\nInvoice\nBill To\nSURAJ BUSINESS CENTRE\nSHOP NO 3 VAGLO VISION OPP HEALTH 18TH JUNE\nROAD PANAJI\nGSTIN : 30AEZPM6270R1ZK\nState: 30-GoaInvoice Details\nInvoice No. : CT/2024-25/16\nDate : 08-07-2024\nPlace of supply: 30-Goa\n# Item nameHSN/\nSACMRP Quantity UnitPrice/\nUnitGST Amount\n1EVM Pendrive 8 GB 852351 \u20b9 499.00 23 Pcs \u20b9 210.00\u20b9 869.40\n(18%)\u20b9 5,699.40\n\u00a0\n\u00a0\n\u00a0\n\u00a0\nTotal 23 \u20b9 869.40\u20b9\n5,699.40\nDescription\nBank Details\nM/s CAMPOS TECHNOLOGIES\nA/C No. 0346139563,\nKotak Mahindra Bank,\nIFSC:\u00a0KKBK0002495\nInvoice Amount In Words\nFive Thousand Six Hundred Ninety Nine Rupees only\nTerms and Conditions\nThank you for your business!Sub Total \u20b9 4,830.00\nSGST@9% \u20b9 434.70\nCGST@9% \u20b9 434.70\nRound off - \u20b9 0.40\nTotal \u20b9 5,699.00\nReceived \u20b9 0.00\nBalance \u20b9 5,699.00\nPayment mode Credit\nYou Saved \u20b9 5,777.60"
        },
        {
            "page_number": 2,
            "content": "For :Campos Technologies\nAuthorized SignatoryCampos Technologies\nF25-26, 1st Floor, Alfran Plaza, Opp. Naturals Ice Cream,\nNr. Don Bosco School, M.G Road, Panaji Goa 403001\nPhone no. : 9145511772\nEmail : campostechgoa@gmail.com\nGSTIN : 30BHLPC5928C2ZS\nState: 30-Goa\n"
        }
    ],
    "INV-148_harshit rathore.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-148 Invoice Date: 30 Mar 2024 Due Date: 01 Apr 2024\nCustomer Details:\nharshit rathore\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Zincovit tablet  78.31\n88.98 (-12%)6 STRP469.8384.57 (18%)554.40\n2Sevenseas Cap 303.29\n344.64 (-12%)2 BTL606.5772.79 (12%)679.36\nTaxable Amount \u20b91,076.40\nCGST\u00a06.0% \u20b936.39\nSGST\u00a06.0% \u20b936.39\nCGST\u00a09.0% \u20b942.28\nSGST\u00a09.0% \u20b942.28\nRound Off 0.24\nTotal\u20b91,234.00\nTotal Discount \u20b9168.24\nTotal Items / Qty : 2 / 8.000 Total amount (in words): \u00a0INR One Thousand, Two Hundred And Thirty-Four Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "FY2407301028789.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n668\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-30\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n7800132ZHFDA6B4E3\n2024-\n07-30\n2024-07-\n30\n17580773\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n149\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n2\n16389973JHU5D5ED14\n2024-\n07-29\n2024-07-\n30\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n3\n24640636RGBED87FCD\n2024-\n07-29\n2024-07-\n30\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n4\n6126645VIPC69311D\n2024-\n07-29\n2024-07-\n30\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n5\n32799042LSUB5A2549\n2024-\n07-30\n2024-07-\n30\n17580780\nWudnKrafts\n3D\tWooden\nCapital\nAlphabet,\nSmall\nAlpha\nVCWKTYWO-\n156\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n6\n12069618HMVBDE6845\n2024-\n07-29\n2024-07-\n30\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n6\n3694.00\n952.03\n952.03\n37.35\n37.35\n1026.72\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n12.00\n3\n387.04\n6.00\n23.22\n6.00\n23.22\n46.44\n5.00\n3\n564.99\n2.50\n14.12\n2.50\n14.12\n28.25\nTotal\n6\n952.03\n37.35\n37.35\n74.69\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n950300\n3\n564.99\n14.12\n14.12\n593.24\n95036010\n3\n387.04\n23.22\n23.22\n433.48\nGrand\tTotal\n6\n952.03\n37.35\n37.35\n1026.72\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n6\n952.03\n37.35\n37.35\n1026.72"
        }
    ],
    "17.pdf": [
        {
            "page_number": 1,
            "content": "Campos Technologies\nF25-26, 1st Floor, Alfran Plaza, Opp. Naturals Ice Cream,\nNr. Don Bosco School, M.G Road, Panaji Goa 403001\nPhone no. : 9145511772\nEmail : campostechgoa@gmail.com\nGSTIN : 30BHLPC5928C2ZS\nState: 30-Goa\nInvoice\nBill To\nGenset Electrical Solutions International private Ltd\nGROUND FLOOR SH-2 Hill Crest Heights, Shetye\nVaddo Mapusa\nGSTIN : 30AAJCG2744H1ZV\nState: 30-GoaInvoice Details\nInvoice No. : CT/2024-25/17\nDate : 08-07-2024\nPlace of supply: 30-Goa\n# Item nameHSN/\nSACMRP Quantity UnitPrice/\nUnitGST Amount\n1Dell 7450 Latitude\nLaptop Renewed\n(Core i7 5th Gen Refurbished\nLaptop, 8 GB RAM, 256 GB\nSSD, )84713020\u20b9\n1,09,000.001 Pcs\u20b9\n19,500.00\u20b9 3,510.00\n(18%)\u20b9\n23,010.00\n\u00a0\n\u00a0\n\u00a0\n\u00a0\nTotal 1\u20b9\n3,510.00\u20b9\n23,010.00\nInvoice Amount In Words\nTwenty Three Thousand Ten Rupees only\nTerms and Conditions\nThank you for your business!Sub Total \u20b9 19,500.00\nSGST@9% \u20b9 1,755.00\nCGST@9% \u20b9 1,755.00\nTotal \u20b9 23,010.00\nReceived \u20b9 0.00\nBalance \u20b9 23,010.00\nPayment mode Credit\nYou Saved \u20b9 85,990.00"
        },
        {
            "page_number": 2,
            "content": "For :Campos Technologies\nAuthorized SignatoryCampos Technologies\nF25-26, 1st Floor, Alfran Plaza, Opp. Naturals Ice Cream,\nNr. Don Bosco School, M.G Road, Panaji Goa 403001\nPhone no. : 9145511772\nEmail : campostechgoa@gmail.com\nGSTIN : 30BHLPC5928C2ZS\nState: 30-Goa\n"
        }
    ],
    "INV-145_Indraja Mohite.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-145 Invoice Date: 28 Mar 2024 Due Date: 28 Mar 2024\nCustomer Details:\nIndraja Mohite\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Acutret 20 mg - 10 capsules  206.43\n258.04 (-20%)9 STRP1,857.86222.94 (12%)2,080.80\n2due last bill  60.00160.000.00 (0%)60.00\nTaxable Amount \u20b91,917.86\nCGST\u00a06.0% \u20b9111.47\nSGST\u00a06.0% \u20b9111.47\nRound Off 0.20\nTotal\u20b92,141.00\nTotal Discount \u20b9520.20\nTotal Items / Qty : 2 / 10.000 Total amount (in words): \u00a0INR Two Thousand, One Hundred And Forty-One Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "IN2-100C23090101_signed_compressed.pdf": [
        {
            "page_number": 1,
            "content": "1Sub Total \n(Tax Inclusive)8,00,000.00\nCGST9 (9%) 61,016.95\nSGST9 (9%) 61,016.95\nTotal \u20b98,00,000.00\nPayment Made (-) 8,00,000.00\nBalance Due \u20b90.00\nAuthorized SignatureTotal In Words\nIndian Rupee Eight Lakh Only\nNotes\nThanks for your business.\n100 Cubes Home Interiors\nKarnataka \nIndia\nGSTIN 29CITPK3346L2ZG\n8660089169\nahad.khan@100cubes.in\nTAX INVOICE\n  # : IN2-100C23090101\n  Invoice Date : 01/07/2024\n  Terms : Due on Receipt\n Due Date : 01/07/2024 Place Of Supply : Karnataka (29)\nBill To\nMr. Shashi Kumar A\n# Item & DescriptionHSN\n/SAC Qty RateCGST SGST\nAmount % Amt % Amt\n1 50% Stage payment - Home \nInteriors998391 1.00 8,00,000.0\n09% 61,016.95 9% 61,016.95 8,00,000.00\n"
        }
    ],
    "GST Sales July 24_compressed.pdf": [
        {
            "page_number": 1,
            "content": "9049337337\nxcellentmadha v1@gmail\n.com\nShop No 1, L unkad\nPlaza, Nr BSNL O\ufb03ce,\nViman Nagar , Pune\nXcellent Xerox And Online Services\nGSTIN: 27AJYP J9896Q1Z G\nState: 27-Mahar ashtr aTax Invoice\nBill To\nConvergint India Pvt Ltd\n2nd Floor , Prestige A trium, No.303 and 304\nNo.1, Centr al Str eet, Shiv ajinagar\nContact No.: 7415861269\nGSTIN Number: 29AABCI8139E1ZH\nState: 29-KarnatakaInvoice No.: 216\nDate: 02-07-2024\nPlace of Supply: 29-Karnataka\n#Item name HSN/ SAC Quantity Price/ unit GST Amount\n1A4 Black & White 48211090 929 \u20b9 1.50 \u20b9 250.83 (18.0%) \u20b9 1,644.33\n2A4 Colour 48211090 67 \u20b9 10.00 \u20b9 120.60 (18.0%) \u20b9 790.60\n3A1 Color 48211090 4 \u20b9 150.00 \u20b9 108.00 (18.0%) \u20b9 708.00\n4Ring Box File 481960 2 \u20b9 90.00 \u20b9 32.40 (18.0%) \u20b9 212.40\n5File Saperator 48201010 1 \u20b9 100.00 \u20b9 18.00 (18.0%) \u20b9 118.00\n6Stamp 35069190 1 \u20b9 250.00 \u20b9 45.00 (18.0%) \u20b9 295.00\nTotal 1004 \u20b9 574.83 \u20b9 3,768.33\nPay To:\nBank Name: Y es Bank, Amanor a Pune\nBank Account No.: 058563400001177\nBank IFSC code: YESB0000585\nAccount Holder 's Name: X cellent X erox And Online Ser vices\nInvoice Amount In Words\nThree Thousand Se ven Hundr ed and Sixty Eight Rupees and Thirty Three P aisa\nonly\nTerms And Conditions\nThank y ou for doing business with us.Sub Total \u20b9 3,193.50\nIGST@18.0% \u20b9 574.83\nTotal \u20b9 3,768.33\nReceiv ed \u20b9 3,768.33\nBalance \u20b9 0.00"
        },
        {
            "page_number": 2,
            "content": "9049337337\nxcellentmadha v1@gmail\n.com\nShop No 1, L unkad\nPlaza, Nr BSNL O\ufb03ce,\nViman Nagar , Pune\nXcellent Xerox And Online Services\nGSTIN: 27AJYP J9896Q1Z G\nState: 27-Mahar ashtr aTax Invoice\nBill To\nGalactic Electric Pvt Ltd\nGat No.1143/A Shindewadi P alshi Road\nShirwal\nContact No.: 9145100315\nGSTIN Number: 27AA CCG6871A1ZR\nState: 27-Mahar ashtr aInvoice No.: 217\nDate: 02-07-2024\nPlace of Supply: 27-Mahar ashtr a\n#Item name HSN/ SAC Quantity Price/ unit GST Amount\n112 By 18 Sticker 48211090 110 \u20b9 35.00 \u20b9 693.00 (18.0%) \u20b9 4,543.00\nTotal 110 \u20b9 693.00 \u20b9 4,543.00\nPay To:\nBank Name: Y es Bank, Amanor a Pune\nBank Account No.: 058563400001177\nBank IFSC code: YESB0000585\nAccount Holder 's Name: X cellent X erox And Online Ser vices\nInvoice Amount In Words\nFour Thousand Fiv e Hundr ed and F orty Three Rupees only\nTerms And Conditions\nThank y ou for doing business with us.\nFor: X cellent X erox And Online Ser vices\nAuthorized SignatorySub Total \u20b9 3,850.00\nSGST@9.0% \u20b9 346.50\nCGST@9.0% \u20b9 346.50\nTotal \u20b9 4,543.00\nReceiv ed \u20b9 4,543.00\nBalance \u20b9 0.00"
        },
        {
            "page_number": 3,
            "content": "9049337337\nxcellentmadha v1@gmail\n.com\nShop No 1, L unkad\nPlaza, Nr BSNL O\ufb03ce,\nViman Nagar , Pune\nXcellent Xerox And Online Services\nGSTIN: 27AJYP J9896Q1Z G\nState: 27-Mahar ashtr aTax Invoice\nBill To\nBanyan & Buddha Pvt Ltd\n1361/B/1/3/B, H.No. 1306, Singh Commer cial\nComplex, W agholi, Pune 412207.\nContact No.: 8551964455\nGSTIN Number: 27AAHCD5157H1ZI\nState: 27-Mahar ashtr aInvoice No.: 218\nDate: 05-07-2024\nPlace of Supply: 27-Mahar ashtr a\n#Item name HSN/ SAC QuantityPrice/ unitDiscount GST Amount\n1Chart Paper 48021010 70 \u20b9 10.71 \u20b9 0.00 (0.0%)\u20b9 90.00\n(12.0%)\u20b9 840.00\n2Petty Cash Voucher 48201010 48 \u20b9 22.32 \u20b9 0.00 (0.0%)\u20b9 128.57\n(12.0%)\u20b9 1,200.00\n3Calculator 84701000 1 \u20b9 283.90 \u20b9 0.00 (0.0%)\u20b9 51.10\n(18.0%)\u20b9 335.00\n4FS Register 48201010 40 \u20b9 57.98\u20b9 115.97\n(5.0%)\u20b9 396.61\n(18.0%)\u20b9 2,600.00\n5Pen 96081099 50 \u20b9 4.24 \u20b9 0.00 (0.0%)\u20b9 38.13\n(18.0%)\u20b9 250.00\nTotal 209 \u20b9 115.97 \u20b9 704.41\u20b9 5,225.00\nPay To:\nBank Name: Y es Bank, Amanor a Pune\nBank Account No.: 058563400001177\nBank IFSC code: YESB0000585\nAccount Holder 's Name: X cellent X erox And Online Ser vices\nInvoice Amount In Words\nFive Thousand Two Hundr ed and Twenty Fiv e Rupees only\nTerms And Conditions\nThank y ou for doing business with us.Sub Total \u20b9 4,636.56\nDiscount \u20b9 115.97\nSGST@6.0% \u20b9 109.29\nCGST@6.0% \u20b9 109.29\nSGST@9.0% \u20b9 242.93\nCGST@9.0% \u20b9 242.93\nTotal \u20b9 5,225.00\nReceiv ed \u20b9 0.00\nBalance \u20b9 5,225.00"
        },
        {
            "page_number": 4,
            "content": "9049337337\nxcellentmadha v1@gmail\n.com\nShop No 1, L unkad\nPlaza, Nr BSNL O\ufb03ce,\nViman Nagar , Pune\nXcellent Xerox And Online Services\nGSTIN: 27AJYP J9896Q1Z G\nState: 27-Mahar ashtr aTax Invoice\nBill To\nBanyan & Buddha Pvt Ltd\n1361/B/1/3/B, H.No. 1306, Singh Commer cial\nComplex, W agholi, Pune 412207.\nContact No.: 8551964455\nGSTIN Number: 27AAHCD5157H1ZI\nState: 27-Mahar ashtr aInvoice No.: 219\nDate: 11-07-2024\nPlace of Supply: 27-Mahar ashtr a\n#Item name HSN/ SAC Quantity Price/ unit GST Amount\n1Calculator 84701000 4 \u20b9 283.90 \u20b9 204.41 (18.0%) \u20b9 1,340.01\nTotal 4 \u20b9 204.41 \u20b9 1,340.01\nPay To:\nBank Name: Y es Bank, Amanor a Pune\nBank Account No.: 058563400001177\nBank IFSC code: YESB0000585\nAccount Holder 's Name: X cellent X erox And Online Ser vices\nInvoice Amount In Words\nOne Thousand Three Hundr ed and F orty Rupees and One P aisa only\nTerms And Conditions\nThank y ou for doing business with us.\nFor: X cellent X erox And Online Ser vices\nAuthorized SignatorySub Total \u20b9 1,135.60\nSGST@9.0% \u20b9 102.20\nCGST@9.0% \u20b9 102.20\nTotal \u20b9 1,340.01\nReceiv ed \u20b9 0.00\nBalance \u20b9 1,340.01"
        },
        {
            "page_number": 5,
            "content": "9049337337\nxcellentmadha v1@gmail\n.com\nShop No 1, L unkad\nPlaza, Nr BSNL O\ufb03ce,\nViman Nagar , Pune\nXcellent Xerox And Online Services\nGSTIN: 27AJYP J9896Q1Z G\nState: 27-Mahar ashtr aTax Invoice\nBill To\nEnexion Consulting Pvt Ltd\nSky Lounge Society E-101 Kala yani nagar Lane\nno 7 Pune\nContact No.: 8482893325\nGSTIN Number: 27AAHCE5025D1ZY\nState: 27-Mahar ashtr aInvoice No.: 220\nDate: 16-07-2024\nPlace of Supply: 27-Mahar ashtr a\n#Item name HSN/ SAC Quantity Price/ unit GST Amount\n1ID Card + Holder 3 \u20b9 90.00 \u20b9 48.60 (18.0%) \u20b9 318.60\nTotal 3 \u20b9 48.60 \u20b9 318.60\nPay To:\nBank Name: Y es Bank, Amanor a Pune\nBank Account No.: 058563400001177\nBank IFSC code: YESB0000585\nAccount Holder 's Name: X cellent X erox And Online Ser vices\nInvoice Amount In Words\nThree Hundr ed and Eighteen Rupees and Sixty P aisa only\nTerms And Conditions\nThank y ou for doing business with us.\nFor: X cellent X erox And Online Ser vices\nAuthorized SignatorySub Total \u20b9 270.00\nSGST@9.0% \u20b9 24.30\nCGST@9.0% \u20b9 24.30\nTotal \u20b9 318.60\nReceiv ed \u20b9 0.00\nBalance \u20b9 318.60"
        },
        {
            "page_number": 6,
            "content": "9049337337\nxcellentmadha v1@gmail\n.com\nShop No 1, L unkad\nPlaza, Nr BSNL O\ufb03ce,\nViman Nagar , Pune\nXcellent Xerox And Online Services\nGSTIN: 27AJYP J9896Q1Z G\nState: 27-Mahar ashtr aTax Invoice\nBill To\nBanyan & Buddha Pvt Ltd\n1361/B/1/3/B, H.No. 1306, Singh Commer cial\nComplex, W agholi, Pune 412207.\nContact No.: 8551964455\nGSTIN Number: 27AAHCD5157H1ZI\nState: 27-Mahar ashtr aInvoice No.: 221\nDate: 19-07-2024\nPlace of Supply: 27-Mahar ashtr a\n#Item name HSN/ SAC Quantity Price/ unit GST Amount\n1Chart Paper 48021010 194 \u20b9 10.71 \u20b9 249.43 (12.0%) \u20b9 2,328.00\n2Ring Box File 481960 5 \u20b9 50.85 \u20b9 45.76 (18.0%) \u20b9 300.00\n3Cello Tape 2\" 39199020 6 \u20b9 25.42 \u20b9 27.46 (18.0%) \u20b9 180.00\n4Cello Tape Dispenser 3926 10 \u20b9 127.12 \u20b9 228.81 (18.0%) \u20b9 1,500.00\n5Button Folder 39261099 6 \u20b9 22.32 \u20b9 16.07 (12.0%) \u20b9 150.00\n6Pencil 96092000 6 \u20b9 6.25 \u20b9 4.50 (12.0%) \u20b9 42.00\nTotal 227 \u20b9 572.03 \u20b9 4,500.00\nPay To:\nBank Name: Y es Bank, Amanor a Pune\nBank Account No.: 058563400001177\nBank IFSC code: YESB0000585\nAccount Holder 's Name: X cellent X erox And Online Ser vices\nInvoice Amount In Words\nFour Thousand Fiv e Hundr ed Rupees only\nTerms And Conditions\nThank y ou for doing business with us.Sub Total \u20b9 3,927.97\nSGST@6.0% \u20b9 135.00\nCGST@6.0% \u20b9 135.00\nSGST@9.0% \u20b9 151.02\nCGST@9.0% \u20b9 151.02\nTotal \u20b9 4,500.00\nReceiv ed \u20b9 0.00\nBalance \u20b9 4,500.00"
        },
        {
            "page_number": 7,
            "content": "9049337337\nxcellentmadha v1@gmail\n.com\nShop No 1, L unkad\nPlaza, Nr BSNL O\ufb03ce,\nViman Nagar , Pune\nXcellent Xerox And Online Services\nGSTIN: 27AJYP J9896Q1Z G\nState: 27-Mahar ashtr aTax Invoice\nBill To\nYugendar Patkar\nContact No.:9441425211Invoice No.: 222\nDate: 24-07-2024\n#Item name HSN/ SAC Quantity Price/ unit GST Amount\n1ID Card + Holder 25 \u20b9 90.00 \u20b9 405.00 (18.0%) \u20b9 2,655.00\n2ID card Design 1 \u20b9 100.00 \u20b9 0.00 (0.0%) \u20b9 100.00\nTotal 26 \u20b9 405.00 \u20b9 2,755.00\nPay To:\nBank Name: Y es Bank, Amanor a Pune\nBank Account No.: 058563400001177\nBank IFSC code: YESB0000585\nAccount Holder 's Name: X cellent X erox And Online Ser vices\nInvoice Amount In Words\nTwo Thousand Se ven Hundr ed and Fifty Fiv e Rupees only\nTerms And Conditions\nThank y ou for doing business with us.\nFor: X cellent X erox And Online Ser vices\nAuthorized SignatorySub Total \u20b9 2,350.00\nSGST@9.0% \u20b9 202.50\nCGST@9.0% \u20b9 202.50\nTotal \u20b9 2,755.00\nReceiv ed \u20b9 0.00\nBalance \u20b9 2,755.00"
        },
        {
            "page_number": 8,
            "content": "9049337337\nxcellentmadha v1@gmail\n.com\nShop No 1, L unkad\nPlaza, Nr BSNL O\ufb03ce,\nViman Nagar , Pune\nXcellent Xerox And Online Services\nGSTIN: 27AJYP J9896Q1Z G\nState: 27-Mahar ashtr aTax Invoice\nBill To\nTrak Services Pvt Ltd\nH 165 SEC TOR 63 NOID A\nContact No.: 8888998295\nGSTIN Number: 09AAA CT0911A1Z1\nState: 09-Uttar Pr adeshInvoice No.: 223\nDate: 26-07-2024\nPlace of Supply: 09-Uttar Pr adesh\n#Item name HSN/ SAC Quantity Price/ unit GST Amount\n1A5 Pomplete Both Side 48211090 10 \u20b9 1,000.00\u20b9 1,800.00\n(18.0%)\u20b9 11,800.00\nTotal 10 \u20b9 1,800.00\u20b9 11,800.00\nPay To:\nBank Name: Y es Bank, Amanor a Pune\nBank Account No.: 058563400001177\nBank IFSC code: YESB0000585\nAccount Holder 's Name: X cellent X erox And Online Ser vices\nInvoice Amount In Words\nEleven Thousand Eight Hundr ed Rupees only\nTerms And Conditions\nThank y ou for doing business with us.\nFor: X cellent X erox And Online Ser vices\nAuthorized SignatorySub Total \u20b9 10,000.00\nIGST@18.0% \u20b9 1,800.00\nTotal \u20b9 11,800.00\nReceiv ed \u20b9 0.00\nBalance \u20b9 11,800.00"
        },
        {
            "page_number": 9,
            "content": "9049337337\nxcellentmadha v1@gmail\n.com\nShop No 1, L unkad\nPlaza, Nr BSNL O\ufb03ce,\nViman Nagar , Pune\nXcellent Xerox And Online Services\nGSTIN: 27AJYP J9896Q1Z G\nState: 27-Mahar ashtr aTax Invoice\nBill To\nRamesh Beacon Pvt Ltd\nTHIRD FL OOR DD A FL ATS J1/297 KALKAJI\nNew Delhi\nContact No.: 9975693011\nGSTIN Number: 07AAMCR0350D1Z O\nState: 07-DelhiInvoice No.: 224\nDate: 28-07-2024\nPlace of Supply: 07-Delhi\n#Item name HSN/ SAC Quantity Price/ unit GST Amount\n1ID Card 48211090 10 \u20b9 80.00 \u20b9 144.00 (18.0%) \u20b9 944.00\n2ID Card Holder 39209999 10 \u20b9 10.00 \u20b9 18.00 (18.0%) \u20b9 118.00\n3Yoyo Holder 39209999 10 \u20b9 30.00 \u20b9 54.00 (18.0%) \u20b9 354.00\nTotal 30 \u20b9 216.00 \u20b9 1,416.00\nPay To:\nBank Name: Y es Bank, Amanor a Pune\nBank Account No.: 058563400001177\nBank IFSC code: YESB0000585\nAccount Holder 's Name: X cellent X erox And Online Ser vices\nInvoice Amount In Words\nOne Thousand F our Hundr ed and Sixteen Rupees only\nTerms And Conditions\nThank y ou for doing business with us.\nFor: X cellent X erox And Online Ser vices\nAuthorized SignatorySub Total \u20b9 1,200.00\nIGST@18.0% \u20b9 216.00\nTotal \u20b9 1,416.00\nReceiv ed \u20b9 0.00\nBalance \u20b9 1,416.00"
        },
        {
            "page_number": 10,
            "content": "9049337337\nxcellentmadha v1@gmail\n.com\nShop No 1, L unkad\nPlaza, Nr BSNL O\ufb03ce,\nViman Nagar , Pune\nXcellent Xerox And Online Services\nGSTIN: 27AJYP J9896Q1Z G\nState: 27-Mahar ashtr aTax Invoice\nBill To\nS Form India Pvt Ltd\n4 Mar vel Edge O\ufb03ce No.4030, Wing B Viman\nNagar Road Viman Nagar\nContact No.: 8378844537\nGSTIN Number: 27AARCS2471B1Z A\nState: 27-Mahar ashtr aInvoice No.: 225\nDate: 31-07-2024\nPlace of Supply: 27-Mahar ashtr a\n#Item name HSN/ SAC Quantity Price/ unit GST Amount\n1A1 Color 48211090 405 \u20b9 40.00\u20b9 2,916.00\n(18.0%)\u20b9 19,116.00\nTotal 405 \u20b9 2,916.00\u20b9 19,116.00\nPay To:\nBank Name: Y es Bank, Amanor a Pune\nBank Account No.: 058563400001177\nBank IFSC code: YESB0000585\nAccount Holder 's Name: X cellent X erox And Online Ser vices\nInvoice Amount In Words\nNinteen Thousand One Hundr ed and Sixteen Rupees only\nTerms And Conditions\nThank y ou for doing business with us.\nFor: X cellent X erox And Online Ser vices\nAuthorized SignatorySub Total \u20b9 16,200.00\nSGST@9.0% \u20b9 1,458.00\nCGST@9.0% \u20b9 1,458.00\nTotal \u20b9 19,116.00\nReceiv ed \u20b9 0.00\nBalance \u20b9 19,116.00"
        },
        {
            "page_number": 11,
            "content": "9049337337\nxcellentmadha v1@gmail\n.com\nShop No 1, L unkad\nPlaza, Nr BSNL O\ufb03ce,\nViman Nagar , Pune\nXcellent Xerox And Online Services\nGSTIN: 27AJYP J9896Q1Z G\nState: 27-Mahar ashtr aTax Invoice\nBill To\nS Form India Pvt Ltd\n4 Mar vel Edge O\ufb03ce No.4030, Wing B Viman\nNagar Road Viman Nagar\nContact No.: 8378844537\nGSTIN Number: 27AARCS2471B1Z A\nState: 27-Mahar ashtr aInvoice No.: 226\nDate: 31-07-2024\nPlace of Supply: 27-Mahar ashtr a\n#Item name HSN/ SAC Quantity Price/ unit GST Amount\n1A1 Lamination 48239030 96 \u20b9 190.00 \u20b9 912.00 (5.0%) \u20b9 19,152.00\nTotal 96 \u20b9 912.00\u20b9 19,152.00\nPay To:\nBank Name: Y es Bank, Amanor a Pune\nBank Account No.: 058563400001177\nBank IFSC code: YESB0000585\nAccount Holder 's Name: X cellent X erox And Online Ser vices\nInvoice Amount In Words\nNinteen Thousand One Hundr ed and Fifty Two Rupees only\nTerms And Conditions\nThank y ou for doing business with us.\nFor: X cellent X erox And Online Ser vices\nAuthorized SignatorySub Total \u20b9 18,240.00\nSGST@2.5% \u20b9 456.00\nCGST@2.5% \u20b9 456.00\nTotal \u20b9 19,152.00\nReceiv ed \u20b9 0.00\nBalance \u20b9 19,152.00"
        },
        {
            "page_number": 12,
            "content": "9049337337\nxcellentmadha v1@gmail\n.com\nShop No 1, L unkad\nPlaza, Nr BSNL O\ufb03ce,\nViman Nagar , Pune\nXcellent Xerox And Online Services\nGSTIN: 27AJYP J9896Q1Z G\nState: 27-Mahar ashtr aTax Invoice\nBill To\nS Form India Pvt Ltd\n4 Mar vel Edge O\ufb03ce No.4030, Wing B Viman\nNagar Road Viman Nagar\nContact No.: 8378844537\nGSTIN Number: 27AARCS2471B1Z A\nState: 27-Mahar ashtr aInvoice No.: 227\nDate: 31-07-2024\nPlace of Supply: 27-Mahar ashtr a\n#Item name HSN/ SAC Quantity Price/ unit GST Amount\n1Toilet Roll 48181000 20 \u20b9 28.00 \u20b9 100.80 (18.0%) \u20b9 660.80\nTotal 20 \u20b9 100.80 \u20b9 660.80\nPay To:\nBank Name: Y es Bank, Amanor a Pune\nBank Account No.: 058563400001177\nBank IFSC code: YESB0000585\nAccount Holder 's Name: X cellent X erox And Online Ser vices\nInvoice Amount In Words\nSix Hundr ed and Sixty Rupees and Eighty P aisa only\nTerms And Conditions\nThank y ou for doing business with us.\nFor: X cellent X erox And Online Ser vices\nAuthorized SignatorySub Total \u20b9 560.00\nSGST@9.0% \u20b9 50.40\nCGST@9.0% \u20b9 50.40\nTotal \u20b9 660.80\nReceiv ed \u20b9 0.00\nBalance \u20b9 660.80"
        },
        {
            "page_number": 13,
            "content": "9049337337\nxcellentmadha v1@gmail\n.com\nShop No 1, L unkad\nPlaza, Nr BSNL O\ufb03ce,\nViman Nagar , Pune\nXcellent Xerox And Online Services\nGSTIN: 27AJYP J9896Q1Z G\nState: 27-Mahar ashtr aTax Invoice\nBill To\nBanyan & Buddha Pvt Ltd\n1361/B/1/3/B, H.No. 1306, Singh Commer cial\nComplex, W agholi, Pune 412207.\nContact No.: 8551964455\nGSTIN Number: 27AAHCD5157H1ZI\nState: 27-Mahar ashtr aInvoice No.: 228\nDate: 31-07-2024\nPlace of Supply: 27-Mahar ashtr a\n#Item name HSN/ SAC Quantity Price/ unit GST Amount\n1Permanent Marker 96082000 50 \u20b9 16.95 \u20b9 152.54 (18.0%) \u20b9 1,000.00\n2CD Marker 960820 6 \u20b9 8.93 \u20b9 6.43 (12.0%) \u20b9 60.00\n3Chart Paper 48021010 864 \u20b9 10.71\u20b9 1,110.86\n(12.0%)\u20b9 10,368.00\n4Ball Pen 96083019 100 \u20b9 4.24 \u20b9 76.27 (18.0%) \u20b9 500.00\nTotal 1020 \u20b9 1,346.10\u20b9 11,928.00\nPay To:\nBank Name: Y es Bank, Amanor a Pune\nBank Account No.: 058563400001177\nBank IFSC code: YESB0000585\nAccount Holder 's Name: X cellent X erox And Online Ser vices\nInvoice Amount In Words\nEleven Thousand Nine Hundr ed and Twenty Eight Rupees only\nTerms And Conditions\nThank y ou for doing business with us.Sub Total \u20b9 10,581.90\nSGST@6.0% \u20b9 558.64\nCGST@6.0% \u20b9 558.64\nSGST@9.0% \u20b9 114.41\nCGST@9.0% \u20b9 114.41\nTotal \u20b9 11,928.00\nReceiv ed \u20b9 0.00\nBalance \u20b9 11,928.00"
        },
        {
            "page_number": 14,
            "content": "9049337337\nxcellentmadha v1@gmail\n.com\nShop No 1, L unkad\nPlaza, Nr BSNL O\ufb03ce,\nViman Nagar , Pune\nXcellent Xerox And Online Services\nGSTIN: 27AJYP J9896Q1Z G\nState: 27-Mahar ashtr aTax Invoice\nBill To\nSamarth Printers\nS.No.45/1 sadguru krupa soecity sangharsh\nchowk chandan nagar\nContact No.: 9823721929\nGSTIN Number: 27CD VPB4401L1ZD\nState: 27-Mahar ashtr aInvoice No.: 229\nDate: 31-07-2024\nPlace of Supply: 27-Mahar ashtr a\n#Item name HSN/ SAC Quantity Price/ unit GST Amount\n1Print For the Month of July 24 48211090 1 \u20b9 14,859.00\u20b9 2,674.62\n(18.0%)\u20b9 17,533.62\nTotal 1 \u20b9 2,674.62\u20b9 17,533.62\nPay To:\nBank Name: Y es Bank, Amanor a Pune\nBank Account No.: 058563400001177\nBank IFSC code: YESB0000585\nAccount Holder 's Name: X cellent X erox And Online Ser vices\nInvoice Amount In Words\nSeventeen Thousand Fiv e Hundr ed and Thirty Three Rupees and Sixty Two\nPaisa only\nTerms And Conditions\nThank y ou for doing business with us.\nFor: X cellent X erox And Online Ser vices\nAuthorized SignatorySub Total \u20b9 14,859.00\nSGST@9.0% \u20b9 1,337.31\nCGST@9.0% \u20b9 1,337.31\nTotal \u20b9 17,533.62\nReceiv ed \u20b9 0.00\nBalance \u20b9 17,533.62"
        },
        {
            "page_number": 15,
            "content": "9049337337\nxcellentmadha v1@gmail\n.com\nShop No 1, L unkad\nPlaza, Nr BSNL O\ufb03ce,\nViman Nagar , Pune\nXcellent Xerox And Online Services\nGSTIN: 27AJYP J9896Q1Z G\nState: 27-Mahar ashtr aTax Invoice\nBill To\nBank of India\n8TH ST AR HOUSE C-5, G BL OCK BANDRA\nKURL A COMPLEX BANDRA EAST\nContact No.: 9112182041\nGSTIN Number: 27AAA CB0472C1Z9\nState: 27-Mahar ashtr aInvoice No.: 230\nDate: 31-07-2024\nPlace of Supply: 27-Mahar ashtr a\n#Item name HSN/ SAC Quantity Price/ unit GST Amount\n1Print For the Month of July 24 48211090 1 \u20b9 2,912.00 \u20b9 524.16 (18.0%) \u20b9 3,436.16\nTotal 1 \u20b9 524.16 \u20b9 3,436.16\nPay To:\nBank Name: Y es Bank, Amanor a Pune\nBank Account No.: 058563400001177\nBank IFSC code: YESB0000585\nAccount Holder 's Name: X cellent X erox And Online Ser vices\nInvoice Amount In Words\nThree Thousand F our Hundr ed and Thirty Six Rupees and Sixteen P aisa only\nTerms And Conditions\nThank y ou for doing business with us.\nFor: X cellent X erox And Online Ser vices\nAuthorized SignatorySub Total \u20b9 2,912.00\nSGST@9.0% \u20b9 262.08\nCGST@9.0% \u20b9 262.08\nTotal \u20b9 3,436.16\nReceiv ed \u20b9 0.00\nBalance \u20b9 3,436.16"
        },
        {
            "page_number": 16,
            "content": "9049337337\nxcellentmadha v1@gmail\n.com\nShop No 1, L unkad\nPlaza, Nr BSNL O\ufb03ce,\nViman Nagar , Pune\nXcellent Xerox And Online Services\nGSTIN: 27AJYP J9896Q1Z G\nState: 27-Mahar ashtr aTax Invoice\nBill To\nBank of India\n8TH ST AR HOUSE C-5, G BL OCK BANDRA\nKURL A COMPLEX BANDRA EAST\nContact No.: 9112182041\nGSTIN Number: 27AAA CB0472C1Z9\nState: 27-Mahar ashtr aInvoice No.: 231\nDate: 31-07-2024\nPlace of Supply: 27-Mahar ashtr a\n#Item name HSN/ SAC Quantity Price/ unit GST Amount\n1O\ufb03ce Stationery 48201010 1 \u20b9 330.00 \u20b9 59.40 (18.0%) \u20b9 389.40\nTotal 1 \u20b9 59.40 \u20b9 389.40\nPay To:\nBank Name: Y es Bank, Amanor a Pune\nBank Account No.: 058563400001177\nBank IFSC code: YESB0000585\nAccount Holder 's Name: X cellent X erox And Online Ser vices\nInvoice Amount In Words\nThree Hundr ed and Eighty Nine Rupees and F orty Paisa only\nTerms And Conditions\nThank y ou for doing business with us.\nFor: X cellent X erox And Online Ser vices\nAuthorized SignatorySub Total \u20b9 330.00\nSGST@9.0% \u20b9 29.70\nCGST@9.0% \u20b9 29.70\nTotal \u20b9 389.40\nReceiv ed \u20b9 0.00\nBalance \u20b9 389.40"
        },
        {
            "page_number": 17,
            "content": "9049337337\nxcellentmadha v1@gmail\n.com\nShop No 1, L unkad\nPlaza, Nr BSNL O\ufb03ce,\nViman Nagar , Pune\nXcellent Xerox And Online Services\nGSTIN: 27AJYP J9896Q1Z G\nState: 27-Mahar ashtr aTax Invoice\nBill To\nUnion Bank of India\nUNION BANK BH AVAN 6TH FL OOR VIDH AN\nBHAVAN M ARGS N ARIM AN POIN T\nContact No.: 9908990847\nGSTIN Number: 27AAA CU0564G1ZH\nState: 27-Mahar ashtr aInvoice No.: 232\nDate: 31-07-2024\nPlace of Supply: 27-Mahar ashtr a\n#Item name HSN/ SAC Quantity Price/ unit GST Amount\n1Print For the Month of July 24 48211090 1 \u20b9 5,381.00 \u20b9 968.58 (18.0%) \u20b9 6,349.58\nTotal 1 \u20b9 968.58 \u20b9 6,349.58\nPay To:\nBank Name: Y es Bank, Amanor a Pune\nBank Account No.: 058563400001177\nBank IFSC code: YESB0000585\nAccount Holder 's Name: X cellent X erox And Online Ser vices\nInvoice Amount In Words\nSix Thousand Three Hundr ed and F orty Nine Rupees and Fifty Eight P aisa only\nTerms And Conditions\nThank y ou for doing business with us.\nFor: X cellent X erox And Online Ser vices\nAuthorized SignatorySub Total \u20b9 5,381.00\nSGST@9.0% \u20b9 484.29\nCGST@9.0% \u20b9 484.29\nTotal \u20b9 6,349.58\nReceiv ed \u20b9 0.00\nBalance \u20b9 6,349.58"
        },
        {
            "page_number": 18,
            "content": "9049337337\nxcellentmadha v1@gmail\n.com\nShop No 1, L unkad\nPlaza, Nr BSNL O\ufb03ce,\nViman Nagar , Pune\nXcellent Xerox And Online Services\nGSTIN: 27AJYP J9896Q1Z G\nState: 27-Mahar ashtr aTax Invoice\nBill To\nUnion Bank of India\nUNION BANK BH AVAN 6TH FL OOR VIDH AN\nBHAVAN M ARGS N ARIM AN POIN T\nContact No.: 9908990847\nGSTIN Number: 27AAA CU0564G1ZH\nState: 27-Mahar ashtr aInvoice No.: 233\nDate: 31-07-2024\nPlace of Supply: 27-Mahar ashtr a\n#Item name HSN/ SAC Quantity Price/ unit GST Amount\n1O\ufb03ce Stationery 48201010 1 \u20b9 625.00 \u20b9 112.50 (18.0%) \u20b9 737.50\nTotal 1 \u20b9 112.50 \u20b9 737.50\nPay To:\nBank Name: Y es Bank, Amanor a Pune\nBank Account No.: 058563400001177\nBank IFSC code: YESB0000585\nAccount Holder 's Name: X cellent X erox And Online Ser vices\nInvoice Amount In Words\nSeven Hundr ed and Thirty Se ven Rupees and Fifty P aisa only\nTerms And Conditions\nThank y ou for doing business with us.\nFor: X cellent X erox And Online Ser vices\nAuthorized SignatorySub Total \u20b9 625.00\nSGST@9.0% \u20b9 56.25\nCGST@9.0% \u20b9 56.25\nTotal \u20b9 737.50\nReceiv ed \u20b9 0.00\nBalance \u20b9 737.50"
        },
        {
            "page_number": 19,
            "content": "9049337337\nxcellentmadha v1@gmail\n.com\nShop No 1, L unkad\nPlaza, Nr BSNL O\ufb03ce,\nViman Nagar , Pune\nXcellent Xerox And Online Services\nGSTIN: 27AJYP J9896Q1Z G\nState: 27-Mahar ashtr aTax Invoice\nBill To\nLIC\n3rd Floor , East Wing, Y ogakshema Mumbai\nDivisional O\ufb03ce -IV Jee van Bima Mar g\nMumbai\nContact No.: 8390170770\nGSTIN Number: 27AAA CL0582H1ZM\nState: 27-Mahar ashtr aInvoice No.: 234\nDate: 31-07-2024\nPlace of Supply: 27-Mahar ashtr a\n#Item name HSN/ SAC Quantity Price/ unit GST Amount\n1Print For The Month of June 24 48211090 1 \u20b9 1,667.00 \u20b9 300.06 (18.0%) \u20b9 1,967.06\nTotal 1 \u20b9 300.06 \u20b9 1,967.06\nPay To:\nBank Name: Y es Bank, Amanor a Pune\nBank Account No.: 058563400001177\nBank IFSC code: YESB0000585\nAccount Holder 's Name: X cellent X erox And Online Ser vices\nInvoice Amount In Words\nOne Thousand Nine Hundr ed and Sixty Se ven Rupees and Six P aisa only\nTerms And Conditions\nThank y ou for doing business with us.\nFor: X cellent X erox And Online Ser vices\nAuthorized SignatorySub Total \u20b9 1,667.00\nSGST@9.0% \u20b9 150.03\nCGST@9.0% \u20b9 150.03\nTotal \u20b9 1,967.06\nReceiv ed \u20b9 0.00\nBalance \u20b9 1,967.06"
        },
        {
            "page_number": 20,
            "content": "9049337337\nxcellentmadha v1@gmail\n.com\nShop No 1, L unkad\nPlaza, Nr BSNL O\ufb03ce,\nViman Nagar , Pune\nXcellent Xerox And Online Services\nGSTIN: 27AJYP J9896Q1Z G\nState: 27-Mahar ashtr aTax Invoice\nBill To\nLIC\n3rd Floor , East Wing, Y ogakshema Mumbai\nDivisional O\ufb03ce -IV Jee van Bima Mar g\nMumbai\nContact No.: 8390170770\nGSTIN Number: 27AAA CL0582H1ZM\nState: 27-Mahar ashtr aInvoice No.: 235\nDate: 31-07-2024\nPlace of Supply: 27-Mahar ashtr a\n#Item name HSN/ SAC Quantity Price/ unit GST Amount\n1Print For the Month of July 24 48211090 1 \u20b9 3,274.00 \u20b9 589.32 (18.0%) \u20b9 3,863.32\nTotal 1 \u20b9 589.32 \u20b9 3,863.32\nPay To:\nBank Name: Y es Bank, Amanor a Pune\nBank Account No.: 058563400001177\nBank IFSC code: YESB0000585\nAccount Holder 's Name: X cellent X erox And Online Ser vices\nInvoice Amount In Words\nThree Thousand Eight Hundr ed and Sixty Three Rupees and Thirty Two P aisa\nonly\nTerms And Conditions\nThank y ou for doing business with us.\nFor: X cellent X erox And Online Ser vices\nAuthorized SignatorySub Total \u20b9 3,274.00\nSGST@9.0% \u20b9 294.66\nCGST@9.0% \u20b9 294.66\nTotal \u20b9 3,863.32\nReceiv ed \u20b9 0.00\nBalance \u20b9 3,863.32"
        },
        {
            "page_number": 21,
            "content": "9049337337\nxcellentmadha v1@gmail\n.com\nShop No 1, L unkad\nPlaza, Nr BSNL O\ufb03ce,\nViman Nagar , Pune\nXcellent Xerox And Online Services\nGSTIN: 27AJYP J9896Q1Z G\nState: 27-Mahar ashtr aTax Invoice\nBill To\nUnion Bank of India\nUNION BANK BH AVAN 6TH FL OOR VIDH AN\nBHAVAN M ARGS N ARIM AN POIN T\nContact No.: 9908990847\nGSTIN Number: 27AAA CU0564G1ZH\nState: 27-Mahar ashtr aInvoice No.: 236\nDate: 31-07-2024\nPlace of Supply: 27-Mahar ashtr a\n#Item name HSN/ SAC Quantity Price/ unit GST Amount\n1Print For the Month of July 24 48211090 1 \u20b9 2,038.00 \u20b9 366.84 (18.0%) \u20b9 2,404.84\nTotal 1 \u20b9 366.84 \u20b9 2,404.84\nPay To:\nBank Name: Y es Bank, Amanor a Pune\nBank Account No.: 058563400001177\nBank IFSC code: YESB0000585\nAccount Holder 's Name: X cellent X erox And Online Ser vices\nInvoice Amount In Words\nTwo Thousand F our Hundr ed and F our Rupees and Eighty F our P aisa only\nTerms And Conditions\nThank y ou for doing business with us.\nFor: X cellent X erox And Online Ser vices\nAuthorized SignatorySub Total \u20b9 2,038.00\nSGST@9.0% \u20b9 183.42\nCGST@9.0% \u20b9 183.42\nTotal \u20b9 2,404.84\nReceiv ed \u20b9 0.00\nBalance \u20b9 2,404.84"
        }
    ],
    "OD14417-sales_invoice-Mukesh_Choudhari_compressed.pdf": [
        {
            "page_number": 1,
            "content": "Opulent Duende\nFourth Floor, Ft.405, Eeco Valley-west block, Madhavapuri Hills,Ameenapur, Near GSM Mall,\nChanda Nagar, Hyderabad, Rangareddy, Hyderabad, Telangana, 500050\nMobile: 9885001313 GSTIN:36BARPS5773E1Z2\nEmail:opulentduende@gmail.com\nInvoice No. :OD14417 Invoice Date :15/08/2024\nBILL TO\nMukesh Choudhari\nMobile:9765988736\nSERVICES SAC TAX AMOUN T\nQUEENPRO BOT CREDITS\nLicensing services for the right to use computer software and databases.997331 762.71\n(18%)5,000\nSUB TOTAL \u20b9 762.71 \u20b9 5,000\nBANK DE TAILS\nName: Opulent Duende\nIFSC Code:UTIB0000538\nAccount No:924020007176709\nBank: Axis Bank,CHANDANAGAR\nPAYMEN T QR CODE\nUPI ID:\nopulentduende@axisbank\nTERMS AND CONDI TIONS\nFees once collected is non-refundable.TAXABLE AMOUNT \u20b9 4,237.29\nCGST @9% \u20b9 381.36\nSGST @9% \u20b9 381.36\nTOTAL AMOUN T \u20b9 5,000\nReceived Amount \u20b9 5,000\nBalance \u20b9 0\nTotal A mount (in wor ds)\nFive Thousand RupeesTAX INVOICEORIGINAL FOR RECIPIENT"
        }
    ],
    "FY2407161013708.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n654\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-16\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n32568686PZP169A104\n2024-\n07-16\n2024-07-\n16\n17580728\nWudnKrafts\nWooden\nCalculation\nShelf\nLearning\nand\tE\nVCWKTYWO-\n104\n95036010@12\n1\n799.00\n35.00\n519.35\n255.04\n255.04\n6.00\n15.30\n6.00\n15.30\n285.64\n2\n31982878PBK8EF80BD\n2024-\n07-15\n2024-07-\n16\n17580765\nWudn\nKrafts\nWooden\nStick\tBaby\nRattles\nwith\t12\nGhun\nVCWKTYWO-\n141\n95036010@12\n1\n199.00\n0.00\n199\n97.72\n97.72\n6.00\n5.86\n6.00\n5.86\n109.45\n3\n8132455JOU54F6D09\n2024-\n07-16\n2024-07-\n16\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n4\n12783156SOX30BBF4C\n2024-\n07-16\n2024-07-\n16\n17580778\nWudnKrafts\nWooden\t3D\nCapital\nAlphabets\nand\nNumber\nVCWKTYWO-\n154\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n5\n5326648ZUS92A79C7\n2024-\n07-16\n2024-07-\n16\n17580778\nWudnKrafts\nWooden\t3D\nCapital\nAlphabets\nand\nNumber\nVCWKTYWO-\n154\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n5\n3395.00\n1011.93\n1011.93\n54.12\n54.12\n1120.18\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n12.00\n4\n823.6\n6.00\n49.42\n6.00\n49.42\n98.83\n5.00\n1\n188.33\n2.50\n4.71\n2.50\n4.71\n9.42\nTotal\n5\n1011.93\n54.12\n54.12\n108.25\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n950300\n1\n188.33\n4.71\n4.71\n197.75\n95036010\n4\n823.60\n49.42\n49.42\n922.43\nGrand\tTotal\n5\n1011.93\n54.12\n54.12\n1120.18\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n5\n1011.93\n54.12\n54.12\n1120.18"
        }
    ],
    "EDISON INVOICE 39_compressed.pdf": [
        {
            "page_number": 1,
            "content": "Invoice. No.& Date\nSHP/39/24-2025\n29-Jul-24\nContact Person\nKARTHICK\n9345057059\n1 85369090 200Set 9,800.00\n2 392390 60 Pkd 10,320.00\n3 3923 200 Pkd 17,000.00\n4 94054200 2 Nos 4,230.00\n41,350.00\n7443\n48,793.00\n0\n48,793.00\nAuthorised SignatoryShipping  Address\nEDISON ENERGY INDIA PVT LTD\nSF No :4/1 ,4/2 ,4/3 Kongudipatty Village\nPudukotai  District ,TamilnaduIlluppur TK ,Post  622102\nMC4 connector 1000V ,30 AINVOICE\nSRI HARI ENTERPRISES\n5/252 Plot 80 1st floor\nSudaruli Nagar\n Billing AddressThalambur, Chennai 600130\nGST IN : 33CKBPS7055L1ZSState Name : Tamil Nadu, Code : 33\nE-Mail : Thenkilakku2019@gmail.com\nEDISON ENERGY INDIA PVT LTD\nSuite No :3 ,2 nd Floor ,Sun Plaza\nG.N.Chetty Road ,Chennai 600006\n15 MW  Solar Power Plant \ncable tie - 12inch \ncable tie - 6inch \nHawells Flood light 60 W\nGST 18%Total\nGrand Total\nRound off\nBank Details:                                                                                                               \nName :SRI HARI ENTERPRISE\nBank : Canara bank\nAccount number : 120002029065\nIFSC code : CNRB0016495\nPayment terms :\n1. 100% advance payment along with the purchase order. \n2.  Delivery Time : 10 To 15 days. Supply Items :\nSI.No QTYGST IN : 33AADCE9497R2ZNOld No : 19 ,New No :39\n                    TOTAL AMOUNT                          DESCRIPTION HSN CODE\nFinal Net Amount\nAmount in worlds: Forty Eight  thousand seven  hundred and Ninety  three  only\nfor SriHari Enterprise\n"
        },
        {
            "page_number": 2,
            "content": ""
        },
        {
            "page_number": 3,
            "content": ""
        },
        {
            "page_number": 4,
            "content": ""
        }
    ],
    "03Jul_Invoice_617_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 617\nDate: 03/07/2024\nPlace of Supply: 33-Tamil NaduBill To:\nVARASAI COMMERCIALS\n1St Floor 22 ANNA PILLAI STREET, \nPARRYS, Georgetown,\nCHENNAI-600 001.\nContact No.: 9884927736\nGSTIN Number: 33AALFV6889P1ZZ\nState: 33-Tamil NaduTransporta\u0000on Details:\nVehicle Number: TN 29 CW 7674\nDelivery Date: 04/07/24\nDelivery Loca\u0000on: Chennai\nShip To:\n1St Floor 22 ANNA PILLAI STREET, \n PARRYS, Georgetown,\n CHENNAI-600 001.\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit Amount\n1 Cane Jaggery Ball 1701 310 Bag \u20b9 1,340.00 \u20b9 4,15,400.00\n2 Cane Jaggery Powder 1701 30 Bag \u20b9 1,755.00 \u20b9 52,650.00\nTotal 340 \u20b9 4,68,050.00\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nFour Lakh Sixty Eight Thousand Fi\u0000y Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 4,68,050.00\nTotal \u20b9 4,68,050.00"
        }
    ],
    "INV-105_Urmila Jangam.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-105 Invoice Date: 27 Jan 2024 Due Date: 27 Jan 2024\nCustomer Details:\nUrmila Jangam\nPh: 9552724364\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Acne-UV Gel - spf 50 (50 gm)  620.34\n775.42 (-20%)1 PAC620.34111.66 (18%)732.00\n2keraglo Eva - 30 Tablet   488.14\n610.17 (-20%)1 BTL488.1487.86 (18%)576.00\n3Cetaphil Moisturising Lotion with Avo-\ncado Oil, Vitamin E, B3 & B5 - 250 ml  864.34\n982.20 (-12%)1 BTL864.34155.58 (18%)1,019.92\nTaxable Amount \u20b91,972.81\nCGST\u00a09.0% \u20b9177.55\nSGST\u00a09.0% \u20b9177.55\nRound Off 0.08\nTotal\u20b92,328.00\nTotal Discount \u20b9466.08\nTotal Items / Qty : 3 / 3.000 Total amount (in words): \u00a0INR Two Thousand, Three Hundred And Twenty-Eight Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "10Jul_Invoice_625_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 625\nDate: 10/07/2024\nPlace of Supply: 33-Tamil NaduBill To:\nGrace Mart Private Limited\nNo.21/4B2, Noombal Road,\nPuliambedu,\nChennai - 600077\nContact No.: 9095334777\nGSTIN Number: 33AAKCG0744Q1Z9\nState: 33-Tamil NaduTransporta\u0000on Details:\nVehicle Number: TN 29 BR 6160\nDelivery Date: 11/07/24\nDelivery Loca\u0000on: Chennai\nShip To:\nNo.21/4B2, Noombal Road,\n Puliambedu,\n Chennai - 600077\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit Amount\n1 Cane Jaggery Ball 1701 125 Bag \u20b9 1,515.00 \u20b9 1,89,375.00\n2 Cane Jaggery Powder 1701 89 Bag \u20b9 1,450.00 \u20b9 1,29,050.00\nTotal 214 \u20b9 3,18,425.00\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nThree Lakh Eighteen Thousand Four Hundred Twenty Five Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 3,18,425.00\nTotal \u20b9 3,18,425.00"
        }
    ],
    "INV-124_Ankita Sattva.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-124 Invoice Date: 10 Feb 2024 Due Date: 10 Feb 2024\nCustomer Details:\nAnkita Sattva\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Arachitol Nano (60k) 4*5ml  340.433 BTL1,021.29122.55 (12%)1,143.84\n2Neurobion Forte - 30 tablets   34.753 STRP104.2418.76 (18%)123.00\nTaxable Amount \u20b91,125.52\nCGST\u00a06.0% \u20b961.28\nSGST\u00a06.0% \u20b961.28\nCGST\u00a09.0% \u20b99.38\nSGST\u00a09.0% \u20b99.38\nDiscount - \u20b9\u00a0152.02\nRound Off 0.18\nTotal\u20b91,115.00\nTotal Discount \u20b9152.02\nTotal Items / Qty : 2 / 6.000 Total amount (in words): \u00a0INR One Thousand, One Hundred And Fifteen Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "FY2407201018258.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n658\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-20\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n19381323LWK0FE98FE\n2024-\n07-20\n2024-07-\n20\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n2\n10149231VUQ2DBF201\n2024-\n07-19\n2024-07-\n20\n17580779\nWudnKrafts\nWooden\t3D\nAlphabet\nand\nNumber\nBoard\tCom\nVCWKTYWO-\n155\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n3\n7863570DGK62F03AE\n2024-\n07-19\n2024-07-\n20\n17580742\nWudnKrafts\n120\tPcs\t12\nColor\nWooden\nDomino\nBlocks\tS\nVCWKTYWO-\n118\n95036010@12\n1\n749.00\n40.00\n449.4\n220.69\n220.69\n6.00\n13.24\n6.00\n13.24\n247.17\n4\n23984736RCSFA291B5\n2024-\n07-19\n2024-07-\n20\n17580778\nWudnKrafts\nWooden\t3D\nCapital\nAlphabets\nand\nNumber\nVCWKTYWO-\n154\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n5\n17936262XWGDD50028\n2024-\n07-20\n2024-07-\n20\n17580765\nWudn\nKrafts\nWooden\nStick\tBaby\nRattles\nwith\t12\nGhun\nVCWKTYWO-\n141\n95036010@12\n1\n199.00\n0.00\n199\n97.72\n97.72\n6.00\n5.86\n6.00\n5.86\n109.45\n6\n31736353VKV4D5CD2E\n2024-\n07-19\n2024-07-\n20\n17580780\nWudnKrafts\n3D\tWooden\nCapital\nAlphabet,\nSmall\nAlpha\nVCWKTYWO-\n156\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n7\n9113594OAF0C76C05\n2024-\n07-19\n2024-07-\n20\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n7\n4443.00\n1301.10\n1301.10\n71.47\n71.47\n1444.05\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n5.00\n1\n188.33\n2.50\n4.71\n2.50\n4.71\n9.42\n12.00\n6\n1112.77\n6.00\n66.77\n6.00\n66.77\n133.53\nTotal\n7\n1301.10\n71.47\n71.47\n142.95\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n950300\n1\n188.33\n4.71\n4.71\n197.75\n95036010\n6\n1112.77\n66.77\n66.77\n1246.30\nGrand\tTotal\n7\n1301.10\n71.47\n71.47\n1444.05\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n7\n1301.10\n71.47\n71.47\n1444.05"
        }
    ],
    "18.pdf": [
        {
            "page_number": 1,
            "content": "Campos Technologies\nF25-26, 1st Floor, Alfran Plaza, Opp. Naturals Ice Cream,\nNr. Don Bosco School, M.G Road, Panaji Goa 403001\nPhone no. : 9145511772\nEmail : campostechgoa@gmail.com\nGSTIN : 30BHLPC5928C2ZS\nState: 30-Goa\nInvoice\nBill To\nYOG SERVICES\nSHOP NO G 23 NR MUNCIPAL GARDERN PONDA\nPONDA GOA\nGSTIN : 30BHWPS5251F1Z4\nState: 30-GoaInvoice Details\nInvoice No. : CT/2024-25/18\nDate : 08-07-2024\nPlace of supply: 30-Goa\n# Item name HSN/ SAC Quantity Unit Price/ Unit GST Amount\n1Laptop Rentals\n((HP i5 6TH GEN 8/256 SSD Qty:\n14 Days, Dates: 18th Mar - 31st\nMar) Rate: \u20b9200)997315 5 Pcs \u20b9 2,800.00\u20b9 2,520.00\n(18%)\u20b9 16,520.00\n2Laptop Rentals\n((HP i5 6TH GEN 8/256 SSD Qty:\n14 Days, Dates: 1st May - 14th\nMay) Rate: \u20b9200)997315 25 Pcs \u20b9 2,800.00\u20b9 12,600.00\n(18%)\u20b9 82,600.00\n3Laptop Rentals\n((HP i5 6TH GEN 8/256 SSD Qty: 6\nDays, Dates: 15 May - 20th May)\nRate: \u20b9200)997315 9 Pcs \u20b9 1,200.00\u20b9 1,944.00\n(18%)\u20b9 12,744.00\n4Laptop Rentals\n((HP i5 6TH GEN 8/256 SSD Qty:\n17 Days, Dates: 21st May - 6th\nJune) Rate: \u20b9200)997315 22 Pcs \u20b9 3,400.00\u20b9 13,464.00\n(18%)\u20b9 88,264.00\n5Laptop Rentals\n((HP i5 6TH GEN 8/256 SSD Qty:\n10 Days, Dates: 28th June - 7th\nJuly) Rate: \u20b9250)997315 3 Pcs \u20b9 2,500.00\u20b9 1,350.00\n(18%)\u20b9 8,850.00\n6Laptop Rentals\n((HP i5 6TH GEN 8/256 SSD Qty: 7\nDays, Dates: 1st June - 7th July)\nRate: \u20b9250)997315 2 Pcs \u20b9 1,750.00\u20b9 630.00\n(18%)\u20b9 4,130.00\nTotal\u20b9\n32,508.00\u20b9\n2,13,108.00"
        },
        {
            "page_number": 2,
            "content": "For :Campos Technologies\nAuthorized SignatoryCampos Technologies\nF25-26, 1st Floor, Alfran Plaza, Opp. Naturals Ice Cream,\nNr. Don Bosco School, M.G Road, Panaji Goa 403001\nPhone no. : 9145511772\nEmail : campostechgoa@gmail.com\nGSTIN : 30BHLPC5928C2ZS\nState: 30-Goa\nInvoice Amount In Words\nTwo Lakh Thirteen Thousand One Hundred Eight Rupees only\nTerms and Conditions\nThank you for your business!Sub Total \u20b9 1,80,600.00\nSGST@9% \u20b9 16,254.00\nCGST@9% \u20b9 16,254.00\nTotal \u20b9 2,13,108.00\nReceived \u20b9 0.00\nBalance \u20b9 2,13,108.00\nPayment mode Credit\n"
        }
    ],
    "CITY_MART_SUPERMARKET_Sales_Invoice_AGNI1002.pdf": [
        {
            "page_number": 1,
            "content": "AGNI IN TEGRA TED F ARM\nNear ganapathy temple,Dhoni, P alakkad, K erala, 678009\nMobile: \u00a0 9004693241 GSTIN: \u00a0 32AOFPA0580C1ZT PAN Number: \u00a0AOFPA0580C\nEmail: \u00a0Ankhulvs@gmail.com\nInvoice No. :AGNI1002 Invoice Date :21/05/2024\nBILL TO\nCITY MART SUPERMARKET\n15/653-9,10,11,12,13,14,15, SOOR YAKAM ALAM, CALICUT\nBYE P ASS RO AD, SEKH ARIPURAM, P alakkad, K erala,\nPalakkad, 678005\nMobile: 9037272310\nGSTIN: 32AANFC5809N1Z4\nPAN Number: AANFC5809N\nState: KeralaSHIP TO\nCITY MART SUPERMARKET\n15/653-9,10,11,12,13,14,15, SOOR YAKAM ALAM, CALICUT\nBYE P ASS RO AD, SEKH ARIPURAM, P alakkad, K erala,\nPalakkad, 678005\nITEMS HSN QTY. RATE TAX AMOUN T\nSALTED TAPIOCA CHIPS 19059030 6 KGS 98.21 70.71\n(12%)660\nMASALA TAPIOCA CHIPS 19059030 6 KGS 98.21 70.71\n(12%)660\nSALTED POTATO CHIPS - 100G 21069099 5 KGS 135.59 122.03\n(18%)800\nCHILLI POTATO CHIPS - 100G 21069099 5 KGS 135.59 122.03\n(18%)800\nSUB TOTAL 22 \u20b9 385.5 \u20b9 2,920\nTERMS AND CONDI TIONS\n1. Goods once sold will not be tak en back or ex changed\n2. All disputes ar e subject t o KERAL A jurisdiction onlyTAXABLE AMOUN T \u20b9 2,534.5\nTOTAL AMOUN T \u20b9 2,920\nReceiv ed Amount \u20b9 0\nTotal A mount (in wor ds)\nTwo Thousand Nine Hundr ed Twenty RupeesTAX INVOICEORIGINAL Harvesting Hope Since 2020"
        }
    ],
    "CITYMART_SUPERMARKET_TWO_Sales_Invoice_1002.pdf": [
        {
            "page_number": 1,
            "content": "AGNI IN TEGRA TED F ARM\nNear ganapathy temple,Dhoni, P alakkad, K erala, 678009\nMobile: \u00a0 9004693241 GSTIN: \u00a0 32AOFPA0580C1ZT PAN Number: \u00a0AOFPA0580C\nEmail: \u00a0Ankhulvs@gmail.com\nInvoice No. :1002 Invoice Date :18/05/2024\nBILL TO\nCITYMART SUPERMARKET TWO\n14/1437, H.M.T OWER, Kalmandapam, Chandr anagar ,\nPalakkad, P alakkad, 678007\nMobile: 8281813003\nGSTIN: 32AA QFC4284N1ZY\nPAN Number: AAQFC4284N\nState: KeralaSHIP TO\nCITYMART SUPERMARKET TWO\n14/1437, H.M.T OWER, Kalmandapam, Chandr anagar ,\nPalakkad, P alakkad, 678007\nITEMS HSN QTY. RATE TAX AMOUN T\nSALTED TAPIOCA CHIPS 19059030 10 K GS 98.21 117.86\n(12%)1,100\nSALTED POTATO CHIPS - 100G 21069099 4 KGS 135.59 97.63\n(18%)640\nSUB TOTAL 14 \u20b9 215.48 \u20b9 1,740\nTERMS AND CONDI TIONS\n1. Goods once sold will not be tak en back or ex changed\n2. All disputes ar e subject t o KERAL A jurisdiction onlyTAXABLE AMOUN T \u20b9 1,524.52\nTOTAL AMOUN T \u20b9 1,740\nReceiv ed Amount \u20b9 0\nTotal A mount (in wor ds)\nOne Thousand Se ven Hundr ed F orty RupeesTAX INVOICEORIGINAL Harvesting Hope Since 2020"
        }
    ],
    "CITY_MART_SUPERMARKET_Sales_Invoice_1.pdf": [
        {
            "page_number": 1,
            "content": "AGNI IN TEGRA TED F ARM\nNear ganapathy temple,Dhoni, P alakkad, K erala, 678009\nMobile: \u00a0 9004693241 GSTIN: \u00a0 32AOFPA0580C1ZT PAN Number: \u00a0AOFPA0580C\nEmail: \u00a0Ankhulvs@gmail.com\nInvoice No. :1 Invoice Date :11/05/2024\nBILL TO\nCITY MART SUPERMARKET\n15/653-9,10,11,12,13,14,15, SOOR YAKAM ALAM, CALICUT\nBYE P ASS RO AD, SEKH ARIPURAM, P alakkad, K erala,\nPalakkad, 678005\nMobile: 9037272310\nGSTIN: 32AANFC5809N1Z4\nPAN Number: AANFC5809N\nState: KeralaSHIP TO\nCITY MART SUPERMARKET\n15/653-9,10,11,12,13,14,15, SOOR YAKAM ALAM, CALICUT\nBYE P ASS RO AD, SEKH ARIPURAM, P alakkad, K erala,\nPalakkad, 678005\nITEMS HSN QTY. RATE TAX AMOUN T\nMASALA TAPIOCA CHIPS 19059030 8 KGS 98.21 94.29\n(12%)880\nSALTED TAPIOCA CHIPS 19059030 8 KGS 98.21 94.29\n(12%)880\nSESAME SEED MURUKKU 21069099 4 KGS 152.54 109.83\n(18%)720\nSALTED POTATO CHIPS - 100G 21069099 2 KGS 135.59 48.81\n(18%)320\nRAGI MIXTURE 19042000 3 KGS 178.57 64.29\n(12%)600\nBEETROOT MURUKKU 21069099 3 KGS 169.49 91.53\n(18%)600\nCHILLI POTATO CHIPS - 100G 21069099 2 KGS 135.59 48.81\n(18%)320\nSUB TOTAL 30 \u20b9 551.84 \u20b9 4,320\nTERMS AND CONDI TIONS\n1. Goods once sold will not be tak en back or ex changed\n2. All disputes ar e subject t o KERAL A jurisdiction onlyTAXABLE AMOUN T \u20b9 3,768.16\nTOTAL AMOUN T \u20b9 4,320\nReceiv ed Amount \u20b9 0\nTotal A mount (in wor ds)\nFour Thousand Three Hundr ed Twenty RupeesTAX INVOICEORIGINAL Harvesting Hope Since 2020"
        }
    ],
    "IN1-100CMBSS01_signed_compressed.pdf": [
        {
            "page_number": 1,
            "content": "1Sub Total \n(Tax Inclusive)25,000.00\nCGST9 (9%) 1,906.78\nSGST9 (9%) 1,906.78\nTotal \u20b925,000.00\nPayment Made (-) 25,000.00\nBalance Due \u20b90.00\nAuthorized SignatureTotal In Words\nIndian Rupee Twenty-Five Thousand Only\nNotes\nThanks for your business.\n100 Cubes Home Interiors\nKarnataka \nIndia\nGSTIN 29CITPK3346L2ZG\n9606472571\npatanabdulahadkhan@gmail.com\nTAX INVOICE\n  # : IN1-100CMBSS01\n  Invoice Date : 19/07/2024\n  Terms : Due on Receipt\n Due Date : 19/07/2024 Place Of Supply : Karnataka (29)\nBill To\nMrs. Jayashree Barve\n# Item & DescriptionHSN\n/SAC Qty RateCGST SGST\nAmount % Amt % Amt\n1 Booking fee - Home Interiors 998391 1.00 25,000.00 9% 1,906.78 9% 1,906.78 25,000.00\n"
        }
    ],
    "FY2407211018953.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n659\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-21\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n12846824PIW9F3E8C6\n2024-\n07-21\n2024-07-\n21\n17580778\nWudnKrafts\nWooden\t3D\nCapital\nAlphabets\nand\nNumber\nVCWKTYWO-\n154\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n2\n23946482LXC042C87B\n2024-\n07-20\n2024-07-\n21\n17580769\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n145\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n3\n21250687JJK8F32D8D\n2024-\n07-21\n2024-07-\n21\n17580773\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n149\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n4\n32591145HHDEF03984\n2024-\n07-21\n2024-07-\n21\n17580772\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n148\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n5\n23933862OSKC5A2E6F\n2024-\n07-20\n2024-07-\n21\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n6\n32093594BUJC8A0E62\n2024-\n07-20\n2024-07-\n21\n14546401\nZivilion\nMagic\tSpin\nRound\nPuzzle\nGame\nSimilar\tto\tR\nVCZITYHT-\n32\n950300@5\n1\n999.00\n45.00\n549.45\n287.81\n287.81\n2.50\n7.20\n2.50\n7.20\n302.20\n7\n21017702XRJ50F0944\n2024-\n07-20\n2024-07-\n21\n17580778\nWudnKrafts\nWooden\t3D\nCapital\nAlphabets\nand\nNumber\nVCWKTYWO-\n154\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n8\n8118392QZZB145F1E\n2024-\n07-20\n2024-07-\n21\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n9\n32355096WGZ07925D1\n2024-\n07-21\n2024-07-\n21\n17580770\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n146\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n10\n25489433ERPAFC2C13\n2024-\n07-20\n2024-07-\n21\n17580778\nWudnKrafts\nWooden\t3D\nCapital\nAlphabets\nand\nNumber\nVCWKTYWO-\n154\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n10\n5290.00\n1524.58\n1524.58\n74.81\n74.81\n1674.20\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n12.00\n8\n1048.44\n6.00\n62.91\n6.00\n62.91\n125.81\n5.00\n2\n476.14\n2.50\n11.90\n2.50\n11.90\n23.81\nTotal\n10\n1524.58\n74.81\n74.81\n149.62\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n950300\n2\n476.14\n11.90\n11.90\n499.95\n95036010\n8\n1048.44\n62.91\n62.91\n1174.25\nGrand\tTotal\n10\n1524.58\n74.81\n74.81\n1674.20\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n10\n1524.58\n74.81\n74.81\n1674.20"
        }
    ],
    "OD14415-sales_invoice-Anup_Manasseh_compressed.pdf": [
        {
            "page_number": 1,
            "content": "Opulent Duende\nFourth Floor, Ft.405, Eeco Valley-west block, Madhavapuri Hills,Ameenapur, Near GSM Mall,\nChanda Nagar, Hyderabad, Rangareddy, Hyderabad, Telangana, 500050\nMobile: 9885001313 GSTIN:36BARPS5773E1Z2\nEmail:opulentduende@gmail.com\nInvoice No. :OD14415 Invoice Date :14/08/2024\nBILL TO\nAnup Manasseh\nMobile:9985996969\nSERVICES SAC TAX AMOUN T\nQUEENPRO BOT CREDITS\nLicensing services for the right to use computer software and databases.997331 2,288.14\n(18%)15,000\nSUB TOTAL \u20b9 2,288.14 \u20b9 15,000\nBANK DE TAILS\nName: Opulent Duende\nIFSC Code:UTIB0000538\nAccount No:924020007176709\nBank: Axis Bank,CHANDANAGAR\nPAYMEN T QR CODE\nUPI ID:\nopulentduende@axisbank\nTERMS AND CONDI TIONS\nFees once collected is non-refundable.TAXABLE AMOUNT \u20b9 12,711.86\nCGST @9% \u20b9 1,144.07\nSGST @9% \u20b9 1,144.07\nTOTAL AMOUN T \u20b9 15,000\nReceived Amount \u20b9 15,000\nBalance \u20b9 0\nTotal A mount (in wor ds)\nFifteen Thousand RupeesTAX INVOICEORIGINAL FOR RECIPIENT"
        }
    ],
    "OD14420-sales_invoice-Pankaj_Sharma_compressed.pdf": [
        {
            "page_number": 1,
            "content": "Opulent Duende\nFourth Floor, Ft.405, Eeco Valley-west block, Madhavapuri Hills,Ameenapur, Near GSM Mall,\nChanda Nagar, Hyderabad, Rangareddy, Hyderabad, Telangana, 500050\nMobile: 9885001313 GSTIN:36BARPS5773E1Z2\nEmail:opulentduende@gmail.com\nInvoice No. :OD14420 Invoice Date :16/08/2024\nBILL TO\nPankaj Sharma\nMobile:9999071129\nSERVICES SAC TAX AMOUN T\nQUEENPRO BOT CREDITS\nLicensing services for the right to use computer software and databases.997331 1,525.42\n(18%)10,000\nSUB TOTAL \u20b9 1,525.42 \u20b9 10,000\nBANK DE TAILS\nName: Opulent Duende\nIFSC Code:UTIB0000538\nAccount No:924020007176709\nBank: Axis Bank,CHANDANAGAR\nPAYMEN T QR CODE\nUPI ID:\nopulentduende@axisbank\nTERMS AND CONDI TIONS\nFees once collected is non-refundable.TAXABLE AMOUNT \u20b9 8,474.58\nCGST @9% \u20b9 762.71\nSGST @9% \u20b9 762.71\nTOTAL AMOUN T \u20b9 10,000\nReceived Amount \u20b9 10,000\nBalance \u20b9 0\nTotal A mount (in wor ds)\nTen Thousand RupeesTAX INVOICEORIGINAL FOR RECIPIENT"
        }
    ],
    "FY2407071003911.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n645\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\t560032\nDate\n2024-07-07\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n15851621XNZB5FDD2E\n2024-\n07-07\n2024-07-\n07\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n2\n8340878TCP93FA646\n2024-\n07-07\n2024-07-\n07\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n3\n22455876IYD2D53788\n2024-\n07-07\n2024-07-\n07\n10992045\nZivilion\nPlastic\nGeometric\nShape\nSorting\nBoard\t-\tM\nVCZITYGS-\n16\n950300@12\n1\n699.00\n50.00\n349.5\n171.63\n171.63\n6.00\n10.30\n6.00\n10.30\n192.23\n4\n24671510FYMFB7411C\n2024-\n07-07\n2024-07-\n07\n10992045\nZivilion\nPlastic\nGeometric\nShape\nSorting\nBoard\t-\tM\nVCZITYGS-\n16\n950300@12\n1\n699.00\n50.00\n349.5\n171.63\n171.63\n6.00\n10.30\n6.00\n10.30\n192.23\n5\n13300474ZFCC92A9D8\n2024-\n07-06\n2024-07-\n07\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n6\n15874830HIH57BEE1C\n2024-\n07-06\n2024-07-\n07\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n7\n9308762CLG5090C31\n2024-\n07-07\n2024-07-\n07\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n8\n30004528NNZF82D617\n2024-\n07-07\n2024-07-\n07\n17580762\nWudn\nKrafts\nWooden\nStick\tBaby\nRattles\nwith\t12\nGhun\nVCWKTYWO-\n138\n95036010@12\n1\n199.00\n0.00\n199\n97.72\n97.72\n6.00\n5.86\n6.00\n5.86\n109.45\n9\n19764634IPW50DD3D2\n2024-\n07-07\n2024-07-\n07\n17580772\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n148\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n10\n31549321XJCE87F2B2\n2024-\n07-07\n2024-07-\n07\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n11\n24931571EZO7C57B1E\n2024-\n07-07\n2024-07-\n07\n17580769\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n145\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n12\n4253735GHR7F3CB3D\n2024-\n07-07\n2024-07-\n07\n17580770\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n146\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n13\n32188486PTX4F66F0B\n2024-\n07-06\n2024-07-\n07\n17580780\nWudnKrafts\n3D\tWooden\nCapital\nAlphabet,\nSmall\nAlpha\nVCWKTYWO-\n156\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n14\n12632406YAV9D07CDE\n2024-\n07-06\n2024-07-\n07\n14546400\nZivilion\nGeometric\nShapes\tand\nColour\nSorter\nPuzzle\nVCZITYHT-\n31\n950300@5\n1\n799.00\n60.00\n319.6\n167.41\n167.41\n2.50\n4.19\n2.50\n4.19\n175.78\n15\n12943135AWW9426383\n2024-\n07-07\n2024-07-\n07\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67"
        },
        {
            "page_number": 2,
            "content": "Rattles\nMusical\tT\n16\n32273197OPCB11D527\n2024-\n07-06\n2024-07-\n07\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n17\n21427873YDC7CF5DC5\n2024-\n07-07\n2024-07-\n07\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n18\n32418883USFA8BEC56\n2024-\n07-06\n2024-07-\n07\n17580742\nWudnKrafts\n120\tPcs\t12\nColor\nWooden\nDomino\nBlocks\tS\nVCWKTYWO-\n118\n95036010@12\n1\n749.00\n40.00\n449.4\n220.69\n220.69\n6.00\n13.24\n6.00\n13.24\n247.17\n19\n11218413DVS3F21BF1\n2024-\n07-07\n2024-07-\n07\n17580737\nWudnKrafts\nWooden\nCalendar\nClock\nPuzzle\nWooden\nClo\nVCWKTYWO-\n113\n95036010@12\n1\n699.00\n40.00\n419.4\n205.96\n205.96\n6.00\n12.36\n6.00\n12.36\n230.68\n20\n14631757HUNE2D099D\n2024-\n07-06\n2024-07-\n07\n17580742\nWudnKrafts\n120\tPcs\t12\nColor\nWooden\nDomino\nBlocks\tS\nVCWKTYWO-\n118\n95036010@12\n1\n749.00\n40.00\n449.4\n220.69\n220.69\n6.00\n13.24\n6.00\n13.24\n247.17\n21\n20816998IIL2B8B7F1\n2024-\n07-07\n2024-07-\n07\n17580769\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n145\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n22\n7251453FZTF15E151\n2024-\n07-07\n2024-07-\n07\n17580730\nWudnKrafts\nMontessori\nWooden\nAlphabet\nLetters\nwith\nVCWKTYWO-\n106\n95036010@12\n1\n699.00\n40.00\n419.4\n205.96\n205.96\n6.00\n12.36\n6.00\n12.36\n230.68\n22\n11578.00\n3145.01\n3145.01\n156.48\n156.48\n3457.96\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n12.00\n17\n2224.28\n6.00\n133.46\n6.00\n133.46\n266.91\n5.00\n5\n920.73\n2.50\n23.02\n2.50\n23.02\n46.04\nTotal\n22\n3145.01\n156.48\n156.48\n312.95\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n950300\n7\n1263.99\n43.61\n43.61\n1351.22\n95036010\n15\n1881.02\n112.86\n112.86\n2106.74\nGrand\tTotal\n22\n3145.01\n156.48\n156.48\n3457.96\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n22\n3145.01\n156.48\n156.48\n3457.96"
        }
    ],
    "05Jul_Invoice_618_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 618\nDate: 05/07/2024\nPlace of Supply: 29-KarnatakaBill To:\nH.N.Anjanayache\u0000y\n466/13 BAZAR ROAD BANGRPET\nGSTIN Number: 29AADFH2874G1Z8\nState: 29-KarnatakaTransporta\u0000on Details:\nVehicle Number: TN 76 Z 6768\nDelivery Date: 06/07/24\nDelivery Loca\u0000on: Bangarpet\nShip To:\n466/13 BAZAR ROAD BANGARPET\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit GST Amount\n1 HTO Jaggery Powder 1701 25 Box \u20b9 572.00 \u20b9 715.00 (5%) \u20b9 15,015.00\nTotal 25 \u20b9 715.00 \u20b9 15,015.00\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nFi\u0000een Thousand Fi\u0000een Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 14,300.00\nIGST@5% \u20b9 715.00\nTotal \u20b9 15,015.00"
        }
    ],
    "INV-143_Prashant.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-143 Invoice Date: 28 Mar 2024 Due Date: 28 Mar 2024\nCustomer Details:\nPrashant\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Arachitol 317.86 72,225.00267.00 (12%)2,492.00\n2PRP  2,169.49\n2,542.37 \n(-14.67%)24,338.98781.02 (18%)5,120.00\nTaxable Amount \u20b96,563.98\nCGST\u00a06.0% \u20b9133.50\nSGST\u00a06.0% \u20b9133.50\nCGST\u00a09.0% \u20b9390.51\nSGST\u00a09.0% \u20b9390.51\nTotal\u20b97,612.00\nTotal Discount \u20b9880.00\nTotal Items / Qty : 2 / 9.000 Total amount (in words): \u00a0INR Seven Thousand, Six Hundred And Twelve Rupees Only .\nAmount Payable: \u20b9152.00\nAmount Paid: \u20b97,460.00\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "CITYMART_SUPERMARKET_TWO_Sales_Invoice_AGNI1003.pdf": [
        {
            "page_number": 1,
            "content": "AGNI IN TEGRA TED F ARM\nNear ganapathy temple,Dhoni, P alakkad, K erala, 678009\nMobile: \u00a0 9004693241 GSTIN: \u00a0 32AOFPA0580C1ZT PAN Number: \u00a0AOFPA0580C\nEmail: \u00a0Ankhulvs@gmail.com\nInvoice No. :AGNI1003 Invoice Date :21/05/2024\nBILL TO\nCITYMART SUPERMARKET TWO\n14/1437, H.M.T OWER, Kalmandapam, Chandr anagar ,\nPalakkad, P alakkad, 678007\nMobile: 8281813003\nGSTIN: 32AA QFC4284N1ZY\nPAN Number: AAQFC4284N\nState: KeralaSHIP TO\nCITYMART SUPERMARKET TWO\n14/1437, H.M.T OWER, Kalmandapam, Chandr anagar ,\nPalakkad, P alakkad, 678007\nITEMS HSN QTY. RATE TAX AMOUN T\nSALTED POTATO CHIPS - 100G 21069099 4 KGS 135.59 97.63\n(18%)640\nCHILLI POTATO CHIPS - 100G 21069099 3 KGS 135.59 73.22\n(18%)480\nSUB TOTAL 7 \u20b9 170.85 \u20b9 1,120\nTERMS AND CONDI TIONS\n1. Goods once sold will not be tak en back or ex changed\n2. All disputes ar e subject t o KERAL A jurisdiction onlyTAXABLE AMOUN T \u20b9 949.15\nTOTAL AMOUN T \u20b9 1,120\nReceiv ed Amount \u20b9 0\nTotal A mount (in wor ds)\nOne Thousand One Hundr ed Twenty RupeesTAX INVOICEORIGINAL Harvesting Hope Since 2020"
        }
    ],
    "FY2407181015996.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n656\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-18\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n23386916NTOACE2FF9\n2024-\n07-18\n2024-07-\n18\n17580784\nWudnKrafts\nSheesham\n(Rosewood)\nCooking\nUtensils\tNo\nVCWKHKCB-\n160\n44190010@12\n1\n499.00\n45.00\n274.45\n134.77\n134.77\n6.00\n8.09\n6.00\n8.09\n150.94\n2\n15162165PZG1E46027\n2024-\n07-17\n2024-07-\n18\n17580769\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n145\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n3\n16897357AFZ3AD1492\n2024-\n07-17\n2024-07-\n18\n17580780\nWudnKrafts\n3D\tWooden\nCapital\nAlphabet,\nSmall\nAlpha\nVCWKTYWO-\n156\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n4\n17411274FMJA7D052E\n2024-\n07-17\n2024-07-\n18\n17580784\nWudnKrafts\nSheesham\n(Rosewood)\nCooking\nUtensils\tNo\nVCWKHKCB-\n160\n44190010@12\n1\n499.00\n45.00\n274.45\n134.77\n134.77\n6.00\n8.09\n6.00\n8.09\n150.94\n5\n5314718QYB3DD2E42\n2024-\n07-17\n2024-07-\n18\n17580769\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n145\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n6\n32604592FNB6F8C39B\n2024-\n07-18\n2024-07-\n18\n17580778\nWudnKrafts\nWooden\t3D\nCapital\nAlphabets\nand\nNumber\nVCWKTYWO-\n154\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n7\n32596978LDU1BF889C\n2024-\n07-17\n2024-07-\n18\n17580772\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n148\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n8\n20186259BHQ04A9A1B\n2024-\n07-17\n2024-07-\n18\n10992045\nZivilion\nPlastic\nGeometric\nShape\nSorting\nBoard\t-\tM\nVCZITYGS-\n16\n950300@12\n1\n699.00\n50.00\n349.5\n171.63\n171.63\n6.00\n10.30\n6.00\n10.30\n192.23\n9\n32603396NTQ0888B70\n2024-\n07-17\n2024-07-\n18\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n10\n5314718QYB3DD2E42\n2024-\n07-17\n2024-07-\n18\n17580765\nWudn\nKrafts\nWooden\nStick\tBaby\nRattles\nwith\t12\nGhun\nVCWKTYWO-\n141\n95036010@12\n1\n199.00\n0.00\n199\n97.72\n97.72\n6.00\n5.86\n6.00\n5.86\n109.45\n11\n14874560ZTZB801739\n2024-\n07-18\n2024-07-\n18\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n12\n32603396NTQ0888B70\n2024-\n07-17\n2024-07-\n18\n17580770\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n146\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n13\n12139267FNX12D467C\n2024-\n07-17\n2024-07-\n18\n17580742\nWudnKrafts\n120\tPcs\t12\nColor\nWooden\nDomino\nBlocks\tS\nVCWKTYWO-\n118\n95036010@12\n1\n749.00\n40.00\n449.4\n220.69\n220.69\n6.00\n13.24\n6.00\n13.24\n247.17\n14\n2485624TZJE15C087\n2024-\n07-17\n2024-07-\n18\n17580747\nWudnKrafts\nColourful\nWooden\nBaby\tRattle\nMusical\tTo\nVCWKTYWO-\n123\n95036010@12\n1\n399.00\n40.00\n239.4\n117.56\n117.56\n6.00\n7.05\n6.00\n7.05\n131.67\n14\n6036.00\n1778.26\n1778.26\n106.70\n106.70\n1991.65\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax"
        },
        {
            "page_number": 2,
            "content": "12.00\n14\n1778.26\n6.00\n106.70\n6.00\n106.70\n213.39\nTotal\n14\n1778.26\n106.70\n106.70\n213.39\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n44190010\n2\n269.54\n16.17\n16.17\n301.88\n950300\n1\n171.63\n10.30\n10.30\n192.23\n95036010\n11\n1337.09\n80.23\n80.23\n1497.54\nGrand\tTotal\n14\n1778.26\n106.70\n106.70\n1991.65\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n14\n1778.26\n106.70\n106.70\n1991.65"
        }
    ],
    "FY2407311030207.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n669\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-31\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n16874242NXN59C84E2\n2024-\n07-31\n2024-07-\n31\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n2\n7804360DTAFFE9E0F\n2024-\n07-30\n2024-07-\n31\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n3\n16646143CJHE28C902\n2024-\n07-30\n2024-07-\n31\n17580779\nWudnKrafts\nWooden\t3D\nAlphabet\nand\nNumber\nBoard\tCom\nVCWKTYWO-\n155\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n4\n24420231VCJC1C7396\n2024-\n07-30\n2024-07-\n31\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n5\n3791954BJJ5EBB9B6\n2024-\n07-31\n2024-07-\n31\n14546403\nZivilion\nShape\nStacker\nSquare\nSmart\nActivity\nFun\ta\nVCZITYHT-\n34\n950300@5\n1\n799.00\n55.00\n359.55\n188.34\n188.34\n2.50\n4.71\n2.50\n4.71\n197.76\n6\n12317932EZR171E9BB\n2024-\n07-31\n2024-07-\n31\n17580780\nWudnKrafts\n3D\tWooden\nCapital\nAlphabet,\nSmall\nAlpha\nVCWKTYWO-\n156\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n7\n4400279AOQ45293D8\n2024-\n07-31\n2024-07-\n31\n17580778\nWudnKrafts\nWooden\t3D\nCapital\nAlphabets\nand\nNumber\nVCWKTYWO-\n154\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n8\n4414783FDN0E2A135\n2024-\n07-30\n2024-07-\n31\n17580741\nWudnKrafts\n100\tpcs\t12\nColor\nWooden\nDominos\nConstru\nVCWKTYWO-\n117\n95036010@12\n1\n699.00\n40.00\n419.4\n205.96\n205.96\n6.00\n12.36\n6.00\n12.36\n230.68\n9\n31066761TJMF5A4AE6\n2024-\n07-30\n2024-07-\n31\n14546399\nZivilion\nShape\tand\nColor\nSorting\nCube\nGeometric\nSh\nVCZITYHT-\n30\n950300@5\n1\n699.00\n55.00\n314.55\n164.76\n164.76\n2.50\n4.12\n2.50\n4.12\n173.00\n10\n23388985TQEC5F6E8B\n2024-\n07-31\n2024-07-\n31\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n10\n6290.00\n1717.95\n1717.95\n84.13\n84.13\n1886.20\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n5.00\n3\n541.43\n2.50\n13.54\n2.50\n13.54\n27.07\n12.00\n7\n1176.52\n6.00\n70.59\n6.00\n70.59\n141.18\nTotal\n10\n1717.95\n84.13\n84.13\n168.25\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n950300\n3\n541.43\n13.54\n13.54\n568.50\n95036010\n7\n1176.52\n70.59\n70.59\n1317.70\nGrand\tTotal\n10\n1717.95\n84.13\n84.13\n1886.20\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n10\n1717.95\n84.13\n84.13\n1886.20"
        }
    ],
    "INV-104_Joseph Wincet.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-104 Invoice Date: 27 Jan 2024 Due Date: 27 Jan 2024\nCustomer Details:\nJoseph Wincet\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Cutacapil Stem - 60 ml  1,029.15\n1,169.49 (-12%)1 BTL1,029.15185.25 (18%)1,214.40\n2Follihair AMPM Tablet Gluten Free - 20 \ntablets  326.64\n371.19 (-12%)6 STRP1,959.86352.78 (18%)2,312.64\n3Dermaq Basic Hair Test Men - B12, V-D \n& TST  849.00\n899.00 (-5.56%)1849.000.00 (0%)849.00\n4Arachitol Nano (60k) 4*5ml  340.433 BTL1,021.29122.55 (12%)1,143.84\n5Dermatologist Consultation  350.00 1350.000.00 (0%)350.00\n6Vitamin b12 test  349.00\n399.00 (-12.53%)1349.000.00 (0%)349.00\nTaxable Amount \u20b95,558.30\nCGST\u00a06.0% \u20b961.28\nSGST\u00a06.0% \u20b961.28\nCGST\u00a09.0% \u20b9269.01\nSGST\u00a09.0% \u20b9269.01\nRound Off 0.12\nTotal\u20b96,219.00\nTotal Discount \u20b9580.96\nTotal Items / Qty : 6 / 13.000 Total amount (in words): \u00a0INR Six Thousand, Two Hundred And Nineteen Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "22.pdf": [
        {
            "page_number": 1,
            "content": "For :Campos Technologies\nAuthorized SignatoryCampos Technologies\nF25-26, 1st Floor, Alfran Plaza, Opp. Naturals Ice Cream,\nNr. Don Bosco School, M.G Road, Panaji Goa 403001\nPhone no. : 9145511772\nEmail : campostechgoa@gmail.com\nGSTIN : 30BHLPC5928C2ZS\nState: 30-Goa\nInvoice\nBill To\nGomantak Private Limited\n11121 GOMANTAK BHAVAN D.B.BANDOKAR ST. INEZ\nGSTIN : 30AAACG6717D1Z7\nState: 30-GoaInvoice Details\nInvoice No. : CT/2024-25/22\nDate : 24-07-2024\nPlace of supply: 30-Goa\n# Item name HSN/ SAC Quantity Unit Price/ Unit GST Amount\n1Lenovo Keyboard LN 320-\n1484716040 1 Pcs \u20b9 1,200.00\u20b9 216.00\n(18%)\u20b9 1,416.00\n\u00a0\n\u00a0\n\u00a0\n\u00a0\nTotal 1 \u20b9 216.00 \u20b9 1,416.00\nInvoice Amount In Words\nOne Thousand Four Hundred Sixteen Rupees only\nTerms and Conditions\nThank you for your business!Sub Total \u20b9 1,200.00\nSGST@9% \u20b9 108.00\nCGST@9% \u20b9 108.00\nTotal \u20b9 1,416.00\nReceived \u20b9 1,416.00\nBalance \u20b9 0.00\nPayment mode Kotak CTech Current"
        }
    ],
    "17Jul_Invoice_631_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 631\nDate: 17/07/2024\nPlace of Supply: 29-KarnatakaBill To:\nSri Mahalakshmi Commercials\nNO.72/2 2ND MAIN ROAD NEW THARAGUPET\nContact No.: 08026705904, 08048903677\nGSTIN Number: 29ADHFS9241H1ZM\nState: 29-KarnatakaTransporta\u0000on Details:\nDelivery Date: 17/07/24\nDelivery Loca\u0000on: Bangalore\nShip To:\nNO.72/2 2ND MAIN ROAD NEW THARAGUPET\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit GST Amount\n1 HTO Jaggery Powder 1701 15 Box \u20b9 590.00 \u20b9 442.50 (5%) \u20b9 9,292.50\nTotal 15 \u20b9 442.50 \u20b9 9,292.50\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nNine Thousand Two Hundred Ninety Three Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 8,850.00\nIGST@5% \u20b9 442.50\nRound o\ufb00 \u20b9 0.50\nTotal \u20b9 9,293.00"
        }
    ],
    "14Jul_Invoice_628_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 628\nDate: 14/07/2024\nPlace of Supply: 29-KarnatakaBill To:\nH.N.Anjanayache\u0000y\n466/13 BAZAR ROAD BANGRPET\nGSTIN Number: 29AADFH2874G1Z8\nState: 29-KarnatakaTransporta\u0000on Details:\nVehicle Number: TN 76 Z 6768\nDelivery Date: 15/07/24\nDelivery Loca\u0000on: Bangarpet\nShip To:\n466/13 BAZAR ROAD BANGARPET\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit GST Amount\n1 HTO Jaggery Powder 1701 25 Box \u20b9 572.00 \u20b9 715.00 (5%) \u20b9 15,015.00\nTotal 25 \u20b9 715.00 \u20b9 15,015.00\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nFi\u0000een Thousand Fi\u0000een Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 14,300.00\nIGST@5% \u20b9 715.00\nTotal \u20b9 15,015.00"
        }
    ],
    "10Jul_Invoice_622_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 622\nDate: 10/07/2024\nPlace of Supply: 29-KarnatakaBill To:\nSri Mahalakshmi Enterprises\nNO.10/41 APMC YARD 6TH MAIN ROAD YESHWANTHPUR\nContact No.: 08026705904, 08048903677\nGSTIN Number: 29AGWPA9333Q1ZE\nState: 29-KarnatakaTransporta\u0000on Details:\nDelivery Date: 10/07/24\nDelivery Loca\u0000on: Yeshwanthpur\nShip To:\nNO.10/41 APMC YARD 6TH MAIN ROAD YESHWANTHPUR\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit GST Amount\n1 HTO Jaggery Powder 1701 26 Box \u20b9 590.00 \u20b9 767.00 (5%) \u20b9 16,107.00\nTotal 26 \u20b9 767.00 \u20b9 16,107.00\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nSixteen Thousand One Hundred Seven Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 15,340.00\nIGST@5% \u20b9 767.00\nTotal \u20b9 16,107.00"
        }
    ],
    "INV-115_Akhil Abhay.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-115 Invoice Date: 29 Jan 2024 Due Date: 29 Jan 2024\nCustomer Details:\nAkhil Abhay\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Arachitol Nano (60k) 4*5ml  299.58\n340.43 (-12%)3 BTL898.73107.85 (12%)1,006.58\n2ME-12 OD (15 tabs)  100.18\n113.84 (-12%)6 STRP601.0772.13 (12%)673.20\n3Keraglo Men Tablet - 30 tablets  477.97\n597.46 (-20%)3 BTL1,433.90258.10 (18%)1,692.00\nTaxable Amount \u20b92,933.70\nCGST\u00a06.0% \u20b989.99\nSGST\u00a06.0% \u20b989.99\nCGST\u00a09.0% \u20b9129.05\nSGST\u00a09.0% \u20b9129.05\nRound Off 0.22\nTotal\u20b93,372.00\nTotal Discount \u20b9652.06\nTotal Items / Qty : 3 / 12.000 Total amount (in words): \u00a0INR Three Thousand, Three Hundred And Seventy-Two Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "FY2407041001148.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n642\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-04\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n23080058IMI162B475\n2024-\n07-03\n2024-07-\n04\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n2\n11327746UDR9D0F0A9\n2024-\n07-03\n2024-07-\n04\n14546401\nZivilion\nMagic\tSpin\nRound\nPuzzle\nGame\nSimilar\tto\tR\nVCZITYHT-\n32\n950300@5\n1\n999.00\n45.00\n549.45\n287.81\n287.81\n2.50\n7.20\n2.50\n7.20\n302.20\n3\n12194152CJH03F135D\n2024-\n07-04\n2024-07-\n04\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n4\n4741939OLIBBAA70D\n2024-\n07-04\n2024-07-\n04\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n5\n11303614INN7AF2A1A\n2024-\n07-04\n2024-07-\n04\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n6\n24290954QTA0D863AE\n2024-\n07-03\n2024-07-\n04\n17580772\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n148\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n7\n24732908KFD0B4F8E9\n2024-\n07-04\n2024-07-\n04\n17580778\nWudnKrafts\nWooden\t3D\nCapital\nAlphabets\nand\nNumber\nVCWKTYWO-\n154\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n8\n11313238FCY624BE4B\n2024-\n07-04\n2024-07-\n04\n17580779\nWudnKrafts\nWooden\t3D\nAlphabet\nand\nNumber\nBoard\tCom\nVCWKTYWO-\n155\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n9\n16673238FMM29277FB\n2024-\n07-03\n2024-07-\n04\n17580769\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n145\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n10\n14385008YWZE516F40\n2024-\n07-03\n2024-07-\n04\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n10\n5990.00\n1626.88\n1626.88\n67.76\n67.76\n1762.41\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n12.00\n6\n774.08\n6.00\n46.44\n6.00\n46.44\n92.89\n5.00\n4\n852.8\n2.50\n21.32\n2.50\n21.32\n42.64\nTotal\n10\n1626.88\n67.76\n67.76\n135.53\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n950300\n4\n852.80\n21.32\n21.32\n895.44\n95036010\n6\n774.08\n46.44\n46.44\n866.97\nGrand\tTotal\n10\n1626.88\n67.76\n67.76\n1762.41\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n10\n1626.88\n67.76\n67.76\n1762.41"
        }
    ],
    "FY2407061003473.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n644\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-06\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n6235597WXMF6B3C13\n2024-\n07-05\n2024-07-\n06\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n2\n32409454HTQC3D3222\n2024-\n07-06\n2024-07-\n06\n14546403\nZivilion\nShape\nStacker\nSquare\nSmart\nActivity\nFun\ta\nVCZITYHT-\n34\n950300@5\n1\n799.00\n55.00\n359.55\n188.34\n188.34\n2.50\n4.71\n2.50\n4.71\n197.76\n3\n20452317YNU919F667\n2024-\n07-05\n2024-07-\n06\n17580764\nWudn\nKrafts\nWooden\nStick\tBaby\nRattles\nwith\t12\nGhun\nVCWKTYWO-\n140\n95036010@12\n1\n199.00\n0.00\n199\n97.72\n97.72\n6.00\n5.86\n6.00\n5.86\n109.45\n4\n32407249LYBE5163FB\n2024-\n07-05\n2024-07-\n06\n17580741\nWudnKrafts\n100\tpcs\t12\nColor\nWooden\nDominos\nConstru\nVCWKTYWO-\n117\n95036010@12\n1\n699.00\n40.00\n419.4\n205.96\n205.96\n6.00\n12.36\n6.00\n12.36\n230.68\n5\n458531KMGFD38E0F\n2024-\n07-05\n2024-07-\n06\n17580762\nWudn\nKrafts\nWooden\nStick\tBaby\nRattles\nwith\t12\nGhun\nVCWKTYWO-\n138\n95036010@12\n1\n199.00\n0.00\n199\n97.72\n97.72\n6.00\n5.86\n6.00\n5.86\n109.45\n6\n32403477SIW36F0ADF\n2024-\n07-05\n2024-07-\n06\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n7\n20454140BMCF237267\n2024-\n07-06\n2024-07-\n06\n10992045\nZivilion\nPlastic\nGeometric\nShape\nSorting\nBoard\t-\tM\nVCZITYGS-\n16\n950300@12\n1\n699.00\n50.00\n349.5\n171.63\n171.63\n6.00\n10.30\n6.00\n10.30\n192.23\n8\n11043362ZFZA24EA89\n2024-\n07-06\n2024-07-\n06\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n9\n10318884XUCFFFFF7A\n2024-\n07-05\n2024-07-\n06\n10992045\nZivilion\nPlastic\nGeometric\nShape\nSorting\nBoard\t-\tM\nVCZITYGS-\n16\n950300@12\n1\n699.00\n50.00\n349.5\n171.63\n171.63\n6.00\n10.30\n6.00\n10.30\n192.23\n10\n32410089QGS3733303\n2024-\n07-06\n2024-07-\n06\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n11\n8699568KIEFAA3021\n2024-\n07-05\n2024-07-\n06\n17580772\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n148\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n11\n5689.00\n1549.38\n1549.38\n73.19\n73.19\n1695.76\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n12.00\n8\n984.38\n6.00\n59.06\n6.00\n59.06\n118.13\n5.00\n3\n565\n2.50\n14.13\n2.50\n14.13\n28.25\nTotal\n11\n1549.38\n73.19\n73.19\n146.38\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n950300\n5\n908.26\n34.72\n34.72\n977.70\n95036010\n6\n641.12\n38.47\n38.47\n718.05"
        },
        {
            "page_number": 2,
            "content": "Grand\tTotal\n11\n1549.38\n73.19\n73.19\n1695.76\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n11\n1549.38\n73.19\n73.19\n1695.76"
        }
    ],
    "OD14422-sales_invoice-Sachin_Shilwant_compressed.pdf": [
        {
            "page_number": 1,
            "content": "Opulent Duende\nFourth Floor, Ft.405, Eeco Valley-west block, Madhavapuri Hills,Ameenapur, Near GSM Mall,\nChanda Nagar, Hyderabad, Rangareddy, Hyderabad, Telangana, 500050\nMobile: 9885001313 GSTIN:36BARPS5773E1Z2\nEmail:opulentduende@gmail.com\nInvoice No. :OD14422 Invoice Date :21/08/2024\nBILL TO\nSachin Shilwant\nMobile:9581855196\nSERVICES SAC TAX AMOUN T\nQUEENPRO BOT CREDITS\nLicensing services for the right to use computer software and databases.997331 2,288.14\n(18%)15,000\nSUB TOTAL \u20b9 2,288.14 \u20b9 15,000\nBANK DE TAILS\nName: Opulent Duende\nIFSC Code:UTIB0000538\nAccount No:924020007176709\nBank: Axis Bank,CHANDANAGAR\nPAYMEN T QR CODE\nUPI ID:\nopulentduende@axisbank\nTERMS AND CONDI TIONS\nFees once collected is non-refundable.TAXABLE AMOUNT \u20b9 12,711.86\nCGST @9% \u20b9 1,144.07\nSGST @9% \u20b9 1,144.07\nTOTAL AMOUN T \u20b9 15,000\nReceived Amount \u20b9 15,000\nBalance \u20b9 0\nTotal A mount (in wor ds)\nFifteen Thousand RupeesTAX INVOICEORIGINAL FOR RECIPIENT"
        }
    ],
    "16Jul_Invoice_629_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 629\nDate: 16/07/2024\nPlace of Supply: 33-Tamil NaduBill To:\nLAKSHMI SAI COMMERCIALS\n22 ANNAPILLAI STREET,\nParrys, Georgetown,\nCHENNAI-600 001\nContact No.: 9884927736\nGSTIN Number: 33AJXPK0990M1ZI\nState: 33-Tamil NaduTransporta\u0000on Details:\nVehicle Number: TN 29 BR 6160\nDelivery Date: 17/07/24\nDelivery Loca\u0000on: Chennai\nShip To:\n22 ANNAPILLAI STREET,\n Parrys, Georgetown,\n CHENNAI-600 001\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit Amount\n1 Cane Jaggery Ball 1701 160 Bag \u20b9 1,360.00 \u20b9 2,17,600.00\n2 Cane Jaggery Powder 1701 30 Bag \u20b9 1,450.00 \u20b9 43,500.00\n3 Cane Jaggery Powder 1701 8 Bag \u20b9 1,500.00 \u20b9 12,000.00\n4 Palm Jaggery 40 Kg \u20b9 240.00 \u20b9 9,600.00\n5 Cane Jaggery Ball 1701 47 Bag \u20b9 1,500.00 \u20b9 70,500.00\nTotal 285 \u20b9 3,53,200.00\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nThree Lakh Fi\u0000y Three Thousand Two Hundred Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 3,53,200.00\nTotal \u20b9 3,53,200.00"
        }
    ],
    "INV-140_Ankit.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-140 Invoice Date: 06 Mar 2024 Due Date: 06 Mar 2024\nCustomer Details:\nAnkit\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Keraglo Men Tablet - 30 tablets  477.97\n597.46 (-20%)1 BTL477.9786.03 (18%)564.00\n2Kera-FM 5% Topical Solution  521.39\n651.74 (-20%)1 BTL521.3962.57 (12%)583.96\nTaxable Amount \u20b9999.36\nCGST\u00a06.0% \u20b931.28\nSGST\u00a06.0% \u20b931.28\nCGST\u00a09.0% \u20b943.02\nSGST\u00a09.0% \u20b943.02\nRound Off 0.04\nTotal\u20b91,148.00\nTotal Discount \u20b9286.99\nTotal Items / Qty : 2 / 2.000 Total amount (in words): \u00a0INR One Thousand, One Hundred And Forty-Eight Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "30Jul_Invoice_644_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 644\nDate: 30/07/2024\nPlace of Supply: 33-Tamil NaduBill To:\nSamy Traders\n77 G-1,CHIDAMBARAM FLATS MANDAPAM LINK ROAD,\nKRISHNA NAGAR, Ullagaram, Chennai-600091\nContact No.: 9150844735\nGSTIN Number: 33CHDPT1108G1ZP\nState: 33-Tamil NaduTransporta\u0000on Details:\nVehicle Number: TN 29 BR 6160\nDelivery Date: 31/07/24\nDelivery Loca\u0000on: Chennai\nShip To:\n77 G-1,CHIDAMBARAM FLATS MANDAPAM LINK ROAD,\nKRISHNA NAGAR, Ullagaram, Chennai-600091\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit Amount\n1 Cane Jaggery Ball 1701 60 Bag \u20b9 1,620.00 \u20b9 97,200.00\n2 Cane Jaggery Powder 1701 10 Bag \u20b9 1,470.00 \u20b9 14,700.00\nTotal 70 \u20b9 1,11,900.00\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nOne Lakh Eleven Thousand Nine Hundred Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 1,11,900.00\nTotal \u20b9 1,11,900.00"
        }
    ],
    "FY2407221020118.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n660\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-22\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n18573976KZFC0837D3\n2024-\n07-22\n2024-07-\n22\n17580737\nWudnKrafts\nWooden\nCalendar\nClock\nPuzzle\nWooden\nClo\nVCWKTYWO-\n113\n95036010@12\n1\n699.00\n40.00\n419.4\n205.96\n205.96\n6.00\n12.36\n6.00\n12.36\n230.68\n2\n13172123AOO6B4E58E\n2024-\n07-22\n2024-07-\n22\n17580780\nWudnKrafts\n3D\tWooden\nCapital\nAlphabet,\nSmall\nAlpha\nVCWKTYWO-\n156\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n3\n23042654PKO40DFD37\n2024-\n07-22\n2024-07-\n22\n17580778\nWudnKrafts\nWooden\t3D\nCapital\nAlphabets\nand\nNumber\nVCWKTYWO-\n154\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n4\n32662353LYTD2A2B7A\n2024-\n07-21\n2024-07-\n22\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n4\n2596.00\n764.90\n764.90\n45.89\n45.89\n856.69\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n12.00\n4\n764.9\n6.00\n45.89\n6.00\n45.89\n91.79\nTotal\n4\n764.90\n45.89\n45.89\n91.79\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n95036010\n4\n764.90\n45.89\n45.89\n856.69\nGrand\tTotal\n4\n764.90\n45.89\n45.89\n856.69\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n4\n764.90\n45.89\n45.89\n856.69"
        }
    ],
    "22Jul_invoice_635_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 635\nDate: 22/07/2024\nPlace of Supply: 29-KarnatakaBill To:\nSri Mahalakshmi Enterprises\nNO.10/41 APMC YARD 6TH MAIN ROAD YESHWANTHPUR\nContact No.: 08026705904, 08048903677\nGSTIN Number: 29AGWPA9333Q1ZE\nState: 29-KarnatakaTransporta\u0000on Details:\nDelivery Date: 22/07/24\nDelivery Loca\u0000on: Yeshwanthpur\nShip To:\nNO.10/41 APMC YARD 6TH MAIN ROAD YESHWANTHPUR\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit GST Amount\n1 HTO Jaggery Powder 1701 40 Box \u20b9 590.00\u20b9 1,180.00 (5%) \u20b9 24,780.00\nTotal 40 \u20b9 1,180.00 \u20b9 24,780.00\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nTwenty Four Thousand Seven Hundred Eighty Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 23,600.00\nIGST@5% \u20b9 1,180.00\nTotal \u20b9 24,780.00"
        }
    ],
    "INV-107_Prashant.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-107 Invoice Date: 29 Jan 2024 Due Date: 29 Jan 2024\nCustomer Details:\nPrashant\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Bioderma Pigmentbio C-concentrate  2,232.33\n2,626.27 (-15%)1 PCS2,232.33401.82 (18%)2,634.15\nTaxable Amount \u20b92,232.33\nCGST\u00a09.0% \u20b9200.91\nSGST\u00a09.0% \u20b9200.91\nRound Off -0.15\nTotal\u20b92,634.00\nTotal Discount \u20b9464.85\nTotal Items / Qty : 1 / 1.000 Total amount (in words): \u00a0INR Two Thousand, Six Hundred And Thirty-Four Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "23Jul_Invoice_638_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 638\nDate: 23/07/2024\nPlace of Supply: 33-Tamil NaduBill To:\nLAKSHMI SAI COMMERCIALS\n22 ANNAPILLAI STREET,\nParrys, Georgetown,\nCHENNAI-600 001\nContact No.: 9884927736\nGSTIN Number: 33AJXPK0990M1ZI\nState: 33-Tamil NaduTransporta\u0000on Details:\nVehicle Number: Tn 29 BR 6160\nDelivery Date: 24/07/24\nDelivery Loca\u0000on: Chennai\nShip To:\n22 ANNAPILLAI STREET,\n Parrys, Georgetown,\n CHENNAI-600 001\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit Amount\n1 Cane Jaggery Ball 1701 52 Bag \u20b9 1,620.00 \u20b9 84,240.00\nTotal 52 \u20b9 84,240.00\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nEighty Four Thousand Two Hundred Forty Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 84,240.00\nTotal \u20b9 84,240.00"
        }
    ],
    "INV-121_Jitesh Soni.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-121 Invoice Date: 29 Jan 2024 Due Date: 29 Jan 2024\nCustomer Details:\nJitesh Soni\nPh: 9351333360Billing Address:\nWakad\nPune, MAHARASHTRA, 411057\nPlace of Supply: \n27-MAHARASHTRA \n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Acne-UV Gel - spf 50 (50 gm)  581.57\n775.42 (-25%)1 PAC581.57104.68 (18%)686.25\n2Arachitol Nano (60k) 4*5ml  289.36\n340.43 (-15%)1 BTL289.3634.72 (12%)324.09\nTaxable Amount \u20b9870.93\nIGST\u00a012.0% \u20b934.72\nIGST\u00a018.0% \u20b9104.68\nRound Off -0.34\nTotal\u20b91,010.00\nTotal Discount \u20b9285.94\nTotal Items / Qty : 2 / 2.000 Total amount (in words): \u00a0INR One Thousand And Ten Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "FY2407051002243.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n643\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-05\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n32389272BGG00B0339\n2024-\n07-04\n2024-07-\n05\n17580737\nWudnKrafts\nWooden\nCalendar\nClock\nPuzzle\nWooden\nClo\nVCWKTYWO-\n113\n95036010@12\n1\n699.00\n40.00\n419.4\n205.96\n205.96\n6.00\n12.36\n6.00\n12.36\n230.68\n2\n11066858LMT9F3A026\n2024-\n07-04\n2024-07-\n05\n17580769\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n145\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n3\n32381509HNFCF4C024\n2024-\n07-04\n2024-07-\n05\n17580780\nWudnKrafts\n3D\tWooden\nCapital\nAlphabet,\nSmall\nAlpha\nVCWKTYWO-\n156\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n4\n24770042PYHC4FAD16\n2024-\n07-04\n2024-07-\n05\n17580773\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n149\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n5\n25460188GFGC70C010\n2024-\n07-05\n2024-07-\n05\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n6\n32014627UJKD492F2A\n2024-\n07-04\n2024-07-\n05\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n7\n6878254PXH11F2996\n2024-\n07-05\n2024-07-\n05\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n8\n32377715KTC2826740\n2024-\n07-04\n2024-07-\n05\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n9\n4783155CUI2077DD5\n2024-\n07-04\n2024-07-\n05\n17580728\nWudnKrafts\nWooden\nCalculation\nShelf\nLearning\nand\tE\nVCWKTYWO-\n104\n95036010@12\n1\n799.00\n35.00\n519.35\n255.04\n255.04\n6.00\n15.30\n6.00\n15.30\n285.64\n10\n22191891EYH0ADBDE4\n2024-\n07-04\n2024-07-\n05\n17580778\nWudnKrafts\nWooden\t3D\nCapital\nAlphabets\nand\nNumber\nVCWKTYWO-\n154\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n11\n13971021ZYAB912D91\n2024-\n07-04\n2024-07-\n05\n17580780\nWudnKrafts\n3D\tWooden\nCapital\nAlphabet,\nSmall\nAlpha\nVCWKTYWO-\n156\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n11\n6489.00\n1847.16\n1847.16\n97.65\n97.65\n2042.45\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n12.00\n9\n1470.5\n6.00\n88.23\n6.00\n88.23\n176.46\n5.00\n2\n376.66\n2.50\n9.42\n2.50\n9.42\n18.83\nTotal\n11\n1847.16\n97.65\n97.65\n195.29\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n950300\n2\n376.66\n9.42\n9.42\n395.49\n95036010\n9\n1470.50\n88.23\n88.23\n1646.96\nGrand\tTotal\n11\n1847.16\n97.65\n97.65\n2042.45\nQuantity\nBasic\nCGST\nSGST\nTotal"
        },
        {
            "page_number": 2,
            "content": "Net\tAmount\n11\n1847.16\n97.65\n97.65\n2042.45"
        }
    ],
    "FY2407151011564.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n653\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-15\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n31077214QGM89A16C4\n2024-\n07-15\n2024-07-\n15\n17580772\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n148\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n2\n17401020VWH5EE4B9E\n2024-\n07-15\n2024-07-\n15\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n3\n10214931SQF36EB76A\n2024-\n07-14\n2024-07-\n15\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n4\n16515014OHMC5BA844\n2024-\n07-14\n2024-07-\n15\n17580778\nWudnKrafts\nWooden\t3D\nCapital\nAlphabets\nand\nNumber\nVCWKTYWO-\n154\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n5\n31018076YEY1D97292\n2024-\n07-14\n2024-07-\n15\n17580773\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n149\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n6\n31018076YEY1D97292\n2024-\n07-14\n2024-07-\n15\n17580747\nWudnKrafts\nColourful\nWooden\nBaby\tRattle\nMusical\tTo\nVCWKTYWO-\n123\n95036010@12\n1\n399.00\n40.00\n239.4\n117.56\n117.56\n6.00\n7.05\n6.00\n7.05\n131.67\n7\n13362618PMU38A5D6E\n2024-\n07-15\n2024-07-\n15\n17580737\nWudnKrafts\nWooden\nCalendar\nClock\nPuzzle\nWooden\nClo\nVCWKTYWO-\n113\n95036010@12\n1\n699.00\n40.00\n419.4\n205.96\n205.96\n6.00\n12.36\n6.00\n12.36\n230.68\n7\n3893.00\n1062.64\n1062.64\n50.58\n50.58\n1163.79\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n12.00\n5\n685.98\n6.00\n41.16\n6.00\n41.16\n82.32\n5.00\n2\n376.66\n2.50\n9.42\n2.50\n9.42\n18.83\nTotal\n7\n1062.64\n50.58\n50.58\n101.15\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n950300\n2\n376.66\n9.42\n9.42\n395.49\n95036010\n5\n685.98\n41.16\n41.16\n768.30\nGrand\tTotal\n7\n1062.64\n50.58\n50.58\n1163.79\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n7\n1062.64\n50.58\n50.58\n1163.79"
        }
    ],
    "INV-102_Kasturi Kalwar.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-102 Invoice Date: 24 Jan 2024 Due Date: 24 Jan 2024\nCustomer Details:\nKasturi Kalwar\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Acne-UV Gel - spf 50 (50 gm)  620.34\n775.42 (-20%)1 PAC620.34111.66 (18%)732.00\n2Cetaphil gentle cleansing lotion - 250 \nML 443.01\n521.19 (-15%)2886.02159.48 (18%)1,045.50\n3Depiwhite Advanced Cream - 15 ml  479.03\n563.56 (-15%)1 TUB479.0386.22 (18%)565.25\n4Neurobion Forte - 30 tablets   31.27\n34.75 (-10%)3 STRP93.8116.89 (18%)110.70\n5Benzac AC 2.5% 82.34\n96.87 (-15%)1 PAC82.349.88 (12%)92.22\nTaxable Amount \u20b92,161.54\nCGST\u00a06.0% \u20b94.94\nSGST\u00a06.0% \u20b94.94\nCGST\u00a09.0% \u20b9187.13\nSGST\u00a09.0% \u20b9187.13\nRound Off 0.33\nTotal\u20b92,546.00\nTotal Discount \u20b9495.82\nTotal Items / Qty : 5 / 8.000 Total amount (in words): \u00a0INR Two Thousand, Five Hundred And Forty-Six Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "FY240703999891.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n641\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-03\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n10809013ZJJ0EA7177\n2024-\n07-03\n2024-07-\n03\n17580769\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n145\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n2\n31505696SOQ3FB7752\n2024-\n07-03\n2024-07-\n03\n17580730\nWudnKrafts\nMontessori\nWooden\nAlphabet\nLetters\nwith\nVCWKTYWO-\n106\n95036010@12\n1\n699.00\n40.00\n419.4\n205.96\n205.96\n6.00\n12.36\n6.00\n12.36\n230.68\n3\n11151979NDLDF830F7\n2024-\n07-02\n2024-07-\n03\n17580778\nWudnKrafts\nWooden\t3D\nCapital\nAlphabets\nand\nNumber\nVCWKTYWO-\n154\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n4\n32358390CVLB91A46E\n2024-\n07-02\n2024-07-\n03\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n4\n1996.00\n593.00\n593.00\n35.58\n35.58\n664.16\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n12.00\n4\n593\n6.00\n35.58\n6.00\n35.58\n71.16\nTotal\n4\n593.00\n35.58\n35.58\n71.16\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n95036010\n4\n593.00\n35.58\n35.58\n664.16\nGrand\tTotal\n4\n593.00\n35.58\n35.58\n664.16\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n4\n593.00\n35.58\n35.58\n664.16"
        }
    ],
    "FY2407141011327.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n652\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-14\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n11309493IVGC2936FC\n2024-\n07-14\n2024-07-\n14\n17580769\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n145\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n2\n32534983HYYF5CB525\n2024-\n07-13\n2024-07-\n14\n10992045\nZivilion\nPlastic\nGeometric\nShape\nSorting\nBoard\t-\tM\nVCZITYGS-\n16\n950300@12\n1\n699.00\n50.00\n349.5\n171.63\n171.63\n6.00\n10.30\n6.00\n10.30\n192.23\n3\n8608281NRJ1090FFE\n2024-\n07-13\n2024-07-\n14\n10992045\nZivilion\nPlastic\nGeometric\nShape\nSorting\nBoard\t-\tM\nVCZITYGS-\n16\n950300@12\n1\n699.00\n50.00\n349.5\n171.63\n171.63\n6.00\n10.30\n6.00\n10.30\n192.23\n4\n19450288HHX83E996D\n2024-\n07-14\n2024-07-\n14\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n5\n32003362IKC1F0A9B9\n2024-\n07-13\n2024-07-\n14\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n6\n32537044OIL2BAC9CA\n2024-\n07-13\n2024-07-\n14\n17580769\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n145\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n7\n32537044OIL2BAC9CA\n2024-\n07-13\n2024-07-\n14\n17580773\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n149\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n8\n20711471NCW0E42411\n2024-\n07-13\n2024-07-\n14\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n9\n12729009LCB0F14785\n2024-\n07-14\n2024-07-\n14\n17580784\nWudnKrafts\nSheesham\n(Rosewood)\nCooking\nUtensils\tNo\nVCWKHKCB-\n160\n44190010@12\n1\n499.00\n45.00\n274.45\n134.77\n134.77\n6.00\n8.09\n6.00\n8.09\n150.94\n10\n30098005NHN5284D6F\n2024-\n07-13\n2024-07-\n14\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n11\n5605564SXI3869B92\n2024-\n07-14\n2024-07-\n14\n17580737\nWudnKrafts\nWooden\nCalendar\nClock\nPuzzle\nWooden\nClo\nVCWKTYWO-\n113\n95036010@12\n1\n699.00\n40.00\n419.4\n205.96\n205.96\n6.00\n12.36\n6.00\n12.36\n230.68\n11\n4889.00\n1327.18\n1327.18\n73.04\n73.04\n1473.26\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n12.00\n10\n1138.85\n6.00\n68.33\n6.00\n68.33\n136.66\n5.00\n1\n188.33\n2.50\n4.71\n2.50\n4.71\n9.42\nTotal\n11\n1327.18\n73.04\n73.04\n146.08\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n44190010\n1\n134.77\n8.09\n8.09\n150.94\n950300\n3\n531.59\n25.30\n25.30\n582.20\n95036010\n7\n660.82\n39.65\n39.65\n740.12\nGrand\tTotal\n11\n1327.18\n73.04\n73.04\n1473.26\nQuantity\nBasic\nCGST\nSGST\nTotal"
        },
        {
            "page_number": 2,
            "content": "Net\tAmount\n11\n1327.18\n73.04\n73.04\n1473.26"
        }
    ],
    "28Jul_Invoice_643_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 643\nDate: 28/07/2024\nPlace of Supply: 33-Tamil NaduBill To:\nLAKSHMI SAI COMMERCIALS\n22 ANNAPILLAI STREET,\nParrys, Georgetown,\nCHENNAI-600 001\nContact No.: 9884927736\nGSTIN Number: 33AJXPK0990M1ZI\nState: 33-Tamil NaduTransporta\u0000on Details:\nVehicle Number: TN 29 BR 6160\nDelivery Date: 29/07/24\nDelivery Loca\u0000on: Chennai\nShip To:\n22 ANNAPILLAI STREET,\n Parrys, Georgetown,\n CHENNAI-600 001\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit Amount\n1 Cane Jaggery Ball 1701 160 Bag \u20b9 1,380.00 \u20b9 2,20,800.00\n2 Cane Jaggery Ball 1701 70 Bag \u20b9 1,590.00 \u20b9 1,11,300.00\n3 Cane Jaggery Powder 1701 10 Bag \u20b9 1,450.00 \u20b9 14,500.00\n4 Cane Jaggery Powder 1701 10 Bag \u20b9 1,710.00 \u20b9 17,100.00\nTotal 250 \u20b9 3,63,700.00\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nThree Lakh Sixty Three Thousand Seven Hundred Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 3,63,700.00\nTotal \u20b9 3,63,700.00"
        }
    ],
    "INV-116_Shefali.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-116 Invoice Date: 29 Jan 2024 Due Date: 29 Jan 2024\nCustomer Details:\nShefali\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Arachitol Nano (60k) 4*5ml  299.58\n340.43 (-12%)3 BTL898.73107.85 (12%)1,006.58\n2ME-12 OD (15 tabs)  100.18\n113.84 (-12%)6 STRP601.0772.13 (12%)673.20\n3Livogen-Z Captabs 15's  84.48\n96.00 (-12%)4 STRP337.920.00 (0%)337.92\nTaxable Amount \u20b91,837.72\nCGST\u00a06.0% \u20b989.99\nSGST\u00a06.0% \u20b989.99\nRound Off 0.30\nTotal\u20b92,018.00\nTotal Discount \u20b9275.14\nTotal Items / Qty : 3 / 13.000 Total amount (in words): \u00a0INR Two Thousand And Eighteen Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "INV-804.pdf": [
        {
            "page_number": 1,
            "content": "1\n POWERED BY\nSub Total 6,400.00\nCGST9 (9%) 576.00\nSGST9 (9%) 576.00\nTotal \u20b97,552.00\nBalance Due \u20b97,552.00\nAuthorized SignatureTotal In Words\nIndian Rupee Seven Thousand Five Hundred Fifty-Two Only\nNotes\nThanks for your business.\nTerms & Conditions\nGoods Once sold cannot be taken back.\nSRI BHAVANI RUBBER & PLASTICS\nNo, 1, Thilagar Nagar 1st Street, Ashok Sylicate Campus,\nKaladipet, Thiruvottiyur,\nChennai Tamil Nadu 600019\nIndia\nGSTIN 33AFIPM5717R1ZV\nsribhavanirubber.plastics@gmail.com TAX INVOICE\n  # : INV-804\n  Invoice Date : 01/07/2024\n  Terms : Net 30\n  Due Date : 31/07/2024\n P.O.# : 224735 Place Of Supply : Tamil Nadu (33)\nBill To Ship To\nSRF LIMITED\nK-1, Sipcot Industrial Area,\nGummudipondi\n601201 Tamilnadu\nIndia\nGSTIN 33AAACS0206P1ZKK-1, Sipcot Industrial Area,\nGummudipondi\n601201 Tamilnadu\nIndia\nGSTIN 33AAACS0206P1ZK\n# Item & DescriptionHSN\n/SAC Qty RateCGST SGST\nAmount % Amt % Amt\n1 55NS00006648\nSilicon Rubber O RIng for \nChuckring for Regular Winder \n(Size: 73.5 * 1.8 C/S)4016932\n0200.00\nNos32.00 9% 576.00 9% 576.00 6,400.00"
        }
    ],
    "FY2407081004010.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n646\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-08\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n7421484RBU9A2CBEA\n2024-\n07-08\n2024-07-\n08\n17580737\nWudnKrafts\nWooden\nCalendar\nClock\nPuzzle\nWooden\nClo\nVCWKTYWO-\n113\n95036010@12\n1\n699.00\n40.00\n419.4\n205.96\n205.96\n6.00\n12.36\n6.00\n12.36\n230.68\n2\n32446585ENVF7CEA8E\n2024-\n07-08\n2024-07-\n08\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n3\n15567674PED411B7A7\n2024-\n07-08\n2024-07-\n08\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n3\n1797.00\n482.39\n482.39\n22.35\n22.35\n527.09\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n12.00\n2\n294.06\n6.00\n17.64\n6.00\n17.64\n35.29\n5.00\n1\n188.33\n2.50\n4.71\n2.50\n4.71\n9.42\nTotal\n3\n482.39\n22.35\n22.35\n44.70\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n950300\n1\n188.33\n4.71\n4.71\n197.75\n95036010\n2\n294.06\n17.64\n17.64\n329.35\nGrand\tTotal\n3\n482.39\n22.35\n22.35\n527.09\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n3\n482.39\n22.35\n22.35\n527.09"
        }
    ],
    "FY2407121009791.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n650\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-12\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n8124343UAR95514AE\n2024-\n07-11\n2024-07-\n12\n17580737\nWudnKrafts\nWooden\nCalendar\nClock\nPuzzle\nWooden\nClo\nVCWKTYWO-\n113\n95036010@12\n1\n699.00\n40.00\n419.4\n205.96\n205.96\n6.00\n12.36\n6.00\n12.36\n230.68\n2\n24821860GETD71D234\n2024-\n07-11\n2024-07-\n12\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n3\n21391968ASX23FC185\n2024-\n07-11\n2024-07-\n12\n17580778\nWudnKrafts\nWooden\t3D\nCapital\nAlphabets\nand\nNumber\nVCWKTYWO-\n154\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n4\n12129125MPGF6BC2F0\n2024-\n07-11\n2024-07-\n12\n17580780\nWudnKrafts\n3D\tWooden\nCapital\nAlphabet,\nSmall\nAlpha\nVCWKTYWO-\n156\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n5\n5934468FBI20DEEA9\n2024-\n07-11\n2024-07-\n12\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n6\n31975233IVJ6D811B9\n2024-\n07-12\n2024-07-\n12\n17580770\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n146\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n7\n18696412RYW399947E\n2024-\n07-11\n2024-07-\n12\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n7\n3393.00\n1004.62\n1004.62\n60.28\n60.28\n1125.17\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n12.00\n7\n1004.62\n6.00\n60.28\n6.00\n60.28\n120.55\nTotal\n7\n1004.62\n60.28\n60.28\n120.55\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n95036010\n7\n1004.62\n60.28\n60.28\n1125.17\nGrand\tTotal\n7\n1004.62\n60.28\n60.28\n1125.17\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n7\n1004.62\n60.28\n60.28\n1125.17"
        }
    ],
    "INV-101_Abhikaran Jalonha.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-101 Invoice Date: 24 Jan 2024 Due Date: 24 Jan 2024\nCustomer Details:\nAbhikaran Jalonha\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Solasafe sunscreen gel spf 50  450.64\n600.85 (-25%)1450.6481.11 (18%)531.75\n2Dermatologist Consultation  350.00 1350.000.00 (0%)350.00\n3Ahaglow Advanced Skin Rejuvenating \nFace Wash Gel 556.34\n632.20 (-12%)1 PAC556.34100.14 (18%)656.48\n4Acutret 10 capsules  146.79\n183.48 (-20%)1 STRP146.7917.61 (12%)164.40\n5Biluma cream - 15 gm  462.37\n525.42 (-12%)1 PAC462.3783.23 (18%)545.60\n6Triluma Cream - 15 gm  535.04\n629.46 (-15%)1 TUB535.0464.21 (12%)599.25\n7Isotroin 10 MG - 10 Capsule  156.36\n177.68 (-12%)2 STRP312.7137.53 (12%)350.24\n8Cetaphil DAM Advance Ultra-Hydrati-\nng Lotion Face - 100 gm  386.31\n438.98 (-12%)1 TUB386.3169.53 (18%)455.84\nTaxable Amount \u20b93,200.20\nCGST\u00a06.0% \u20b959.67\nSGST\u00a06.0% \u20b959.67\nCGST\u00a09.0% \u20b9167.01\nSGST\u00a09.0% \u20b9167.01\nRound Off 0.44\nTotal\u20b93,654.00\nTotal Discount \u20b9597.94\nTotal Items / Qty : 8 / 9.000 Total amount (in words): \u00a0INR Three Thousand, Six Hundred And Fifty-Four Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "21Jul_Invoice_633_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 633\nDate: 21/07/2024\nPlace of Supply: 33-Tamil NaduBill To:\nLAKSHMI SAI COMMERCIALS\n22 ANNAPILLAI STREET,\nParrys, Georgetown,\nCHENNAI-600 001\nContact No.: 9884927736\nGSTIN Number: 33AJXPK0990M1ZI\nState: 33-Tamil NaduTransporta\u0000on Details:\nVehicle Number: TN 29 BR 6160\nDelivery Date: 22/07/24\nDelivery Loca\u0000on: Chennai\nShip To:\n22 ANNAPILLAI STREET,\n Parrys, Georgetown,\n CHENNAI-600 001\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit Amount\n1 Cane Jaggery Ball 1701 130 Bag \u20b9 1,360.00 \u20b9 1,76,800.00\n2 Cane Jaggery Ball 1701 103 Bag \u20b9 1,560.00 \u20b9 1,60,680.00\n3 Cane Jaggery Ball 1701 10 Bag \u20b9 2,600.00 \u20b9 26,000.00\n4 Cane Jaggery Powder 1701 10 Bag \u20b9 1,450.00 \u20b9 14,500.00\nTotal 253 \u20b9 3,77,980.00\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nThree Lakh Seventy Seven Thousand Nine Hundred Eighty Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 3,77,980.00\nTotal \u20b9 3,77,980.00"
        }
    ],
    "FY2407261024901.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n664\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-26\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n6040124GRV312D0A8\n2024-\n07-26\n2024-07-\n26\n10992045\nZivilion\nPlastic\nGeometric\nShape\nSorting\nBoard\t-\tM\nVCZITYGS-\n16\n950300@12\n1\n699.00\n50.00\n349.5\n171.63\n171.63\n6.00\n10.30\n6.00\n10.30\n192.23\n2\n6040124GRV312D0A8\n2024-\n07-26\n2024-07-\n26\n14546404\nZivilion\nAlphabets\nNumbers\nShapes\tand\nColours\nEduc\nVCZITYHT-\n35\n950300@5\n1\n999.00\n55.00\n449.55\n235.48\n235.48\n2.50\n5.89\n2.50\n5.89\n247.25\n3\n5036672LET6143F4E\n2024-\n07-25\n2024-07-\n26\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n4\n22156692JSH3F31951\n2024-\n07-26\n2024-07-\n26\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n5\n10149231HVQD0CE799\n2024-\n07-25\n2024-07-\n26\n17580779\nWudnKrafts\nWooden\t3D\nAlphabet\nand\nNumber\nBoard\tCom\nVCWKTYWO-\n155\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n6\n25719824KUJ0E1557B\n2024-\n07-26\n2024-07-\n26\n17580747\nWudnKrafts\nColourful\nWooden\nBaby\tRattle\nMusical\tTo\nVCWKTYWO-\n123\n95036010@12\n1\n399.00\n40.00\n239.4\n117.56\n117.56\n6.00\n7.05\n6.00\n7.05\n131.67\n7\n10489335FKYC534EB2\n2024-\n07-25\n2024-07-\n26\n17580778\nWudnKrafts\nWooden\t3D\nCapital\nAlphabets\nand\nNumber\nVCWKTYWO-\n154\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n8\n30718819VQE4E35A2B\n2024-\n07-26\n2024-07-\n26\n17580772\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n148\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n8\n4992.00\n1335.46\n1335.46\n65.29\n65.29\n1466.05\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n12.00\n6\n911.65\n6.00\n54.70\n6.00\n54.70\n109.40\n5.00\n2\n423.81\n2.50\n10.60\n2.50\n10.60\n21.19\nTotal\n8\n1335.46\n65.29\n65.29\n130.59\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n950300\n3\n595.44\n20.89\n20.89\n637.23\n95036010\n5\n740.02\n44.40\n44.40\n828.82\nGrand\tTotal\n8\n1335.46\n65.29\n65.29\n1466.05\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n8\n1335.46\n65.29\n65.29\n1466.05"
        }
    ],
    "INV-128_Atia Latif.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-128 Invoice Date: 23 Feb 2024 Due Date: 23 Feb 2024\nCustomer Details:\nAtia Latif\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Hydrafacial 2,076.27\n2,542.37 \n(-18.33%)12,076.27373.73 (18%)2,450.00\nTaxable Amount \u20b92,076.27\nCGST\u00a09.0% \u20b9186.86\nSGST\u00a09.0% \u20b9186.86\nTotal\u20b92,450.00\nTotal Discount \u20b9550.00\nTotal Items / Qty : 1 / 1.000 Total amount (in words): \u00a0INR Two Thousand, Four Hundred And Fifty Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "CITYMART_SUPERMARKET_TWO_Sales_Invoice_AGNI1005.pdf": [
        {
            "page_number": 1,
            "content": "AGNI IN TEGRA TED F ARM\nNear ganapathy temple,Dhoni, P alakkad, K erala, 678009\nMobile: \u00a0 9004693241 GSTIN: \u00a0 32AOFPA0580C1ZT PAN Number: \u00a0AOFPA0580C\nEmail: \u00a0Ankhulvs@gmail.com\nInvoice No. :AGNI1005 Invoice Date :31/05/2024\nBILL TO\nCITYMART SUPERMARKET TWO\n14/1437, H.M.T OWER, Kalmandapam, Chandr anagar ,\nPalakkad, P alakkad, 678007\nMobile: 8281813003\nGSTIN: 32AA QFC4284N1ZY\nPAN Number: AAQFC4284N\nState: KeralaSHIP TO\nCITYMART SUPERMARKET TWO\n14/1437, H.M.T OWER, Kalmandapam, Chandr anagar ,\nPalakkad, P alakkad, 678007\nITEMS HSN QTY. RATE TAX AMOUN T\nCHILLI POTATO CHIPS - 100G 21069099 6 KGS 135.59 146.44\n(18%)960\nSUB TOTAL 6 \u20b9 146.44 \u20b9 960\nTERMS AND CONDI TIONS\n1. Goods once sold will not be tak en back or ex changed\n2. All disputes ar e subject t o KERAL A jurisdiction onlyTAXABLE AMOUN T \u20b9 813.56\nTOTAL AMOUN T \u20b9 960\nReceiv ed Amount \u20b9 0\nTotal A mount (in wor ds)\nNine Hundr ed Sixty RupeesTAX INVOICEORIGINAL Harvesting Hope Since 2020"
        }
    ],
    "INV-146_Abhikaran Jalonha.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-146 Invoice Date: 29 Mar 2024 Due Date: 29 Mar 2024\nCustomer Details:\nAbhikaran Jalonha\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Cetaphil DAM Advance Ultra-Hydrati-\nng Lotion Face - 100 gm  395.08\n438.98 (-10%)1 TUB395.0871.12 (18%)466.20\n2Ahaglow Advanced Skin Rejuvenating \nFace Wash Gel 556.34\n632.20 (-12%)1 PAC556.34100.14 (18%)656.48\n3Acutret 20 mg - 10 capsules  206.43\n258.04 (-20%)6 STRP1,238.57148.63 (12%)1,387.20\n4 Faireye under eye cream  695.80\n790.68 (-12%)1695.80125.24 (18%)821.04\n5Biluma cream - 15 gm  462.37\n525.42 (-12%)1 PAC462.3783.23 (18%)545.60\nTaxable Amount \u20b93,348.16\nCGST\u00a06.0% \u20b974.31\nSGST\u00a06.0% \u20b974.31\nCGST\u00a09.0% \u20b9189.86\nSGST\u00a09.0% \u20b9189.86\nRound Off 0.48\nTotal\u20b93,877.00\nTotal Discount \u20b9674.48\nTotal Items / Qty : 5 / 10.000 Total amount (in words): \u00a0INR Three Thousand, Eight Hundred And Seventy-Seven Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "INV-110_Kushagra Tiwari.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-110 Invoice Date: 29 Jan 2024 Due Date: 29 Jan 2024\nCustomer Details:\nKushagra Tiwari\nPh: 6393569079\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Arachitol Nano (60k) 4*5ml  299.58\n340.43 (-12%)3 BTL898.73107.85 (12%)1,006.58\n2ME-12 OD (15 tabs)  100.18\n113.84 (-12%)6 STRP601.0772.13 (12%)673.20\nTaxable Amount \u20b91,499.80\nCGST\u00a06.0% \u20b989.99\nSGST\u00a06.0% \u20b989.99\nRound Off 0.22\nTotal\u20b91,680.00\nTotal Discount \u20b9229.06\nTotal Items / Qty : 2 / 9.000 Total amount (in words): \u00a0INR One Thousand, Six Hundred And Eighty Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "23Jul_Invoice_639_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 639\nDate: 23/07/2024\nPlace of Supply: 33-Tamil NaduBill To:\nSamy Traders\n77 G-1,CHIDAMBARAM FLATS MANDAPAM LINK ROAD,\nKRISHNA NAGAR, Ullagaram, Chennai-600091\nContact No.: 9150844735\nGSTIN Number: 33CHDPT1108G1ZP\nState: 33-Tamil NaduTransporta\u0000on Details:\nVehicle Number: Tn 29 BR 6160\nDelivery Date: 24/07/24\nDelivery Loca\u0000on: Chennai\nShip To:\n77 G-1,CHIDAMBARAM FLATS MANDAPAM LINK ROAD,\nKRISHNA NAGAR, Ullagaram, Chennai-600091\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit Amount\n1 Cane Jaggery Ball 1701 53 Bag \u20b9 1,635.00 \u20b9 86,655.00\nTotal 53 \u20b9 86,655.00\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nEighty Six Thousand Six Hundred Fi\u0000y Five Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 86,655.00\nTotal \u20b9 86,655.00"
        }
    ],
    "INV-138_Agrani Kandele.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-138 Invoice Date: 06 Mar 2024 Due Date: 06 Mar 2024\nCustomer Details:\nAgrani Kandele\nPh: 8120482988Shipping Address:\nvadandana beauty parlour\nMurawara\nKatni, MADHYA PRADESH, 483501\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Cetaphil Gentle Cleansing Lotion - Oily \nSkin 125 ml 415.68\n461.86 (-10%)1 BTL415.6874.82 (18%)490.50\n2Solasafe sunscreen gel spf 50  480.68\n600.85 (-20%)1480.6886.52 (18%)567.20\n3IPCA Acne OC Moisturizer Cream - 75 \ngm 378.98\n473.73 (-20%)1378.9868.22 (18%)447.20\nTaxable Amount \u20b91,275.34\nCGST\u00a09.0% \u20b9114.78\nSGST\u00a09.0% \u20b9114.78\nRound Off 0.10\nTotal\u20b91,505.00\nTotal Discount \u20b9308.10\nTotal Items / Qty : 3 / 3.000 Total amount (in words): \u00a0INR One Thousand, Five Hundred And Five Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "26Jul_Invoice_642_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 642\nDate: 26/07/2024\nPlace of Supply: 29-KarnatakaBill To:\nSri Mahalakshmi Commercials\nNO.72/2 2ND MAIN ROAD NEW THARAGUPET\nContact No.: 08026705904, 08048903677\nGSTIN Number: 29ADHFS9241H1ZM\nState: 29-KarnatakaTransporta\u0000on Details:\nDelivery Date: 26/07/24\nDelivery Loca\u0000on: Bangalore\nShip To:\nNO.72/2 2ND MAIN ROAD NEW THARAGUPET\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit GST Amount\n1 HTO Jaggery Powder 1701 15 Box \u20b9 590.00 \u20b9 442.50 (5%) \u20b9 9,292.50\nTotal 15 \u20b9 442.50 \u20b9 9,292.50\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nNine Thousand Two Hundred Ninety Three Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 8,850.00\nIGST@5% \u20b9 442.50\nRound o\ufb00 \u20b9 0.50\nTotal \u20b9 9,293.00"
        }
    ],
    "FY2407171014844.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n655\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-17\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n20280534YSC6929B04\n2024-\n07-16\n2024-07-\n17\n10992045\nZivilion\nPlastic\nGeometric\nShape\nSorting\nBoard\t-\tM\nVCZITYGS-\n16\n950300@12\n1\n699.00\n50.00\n349.5\n171.63\n171.63\n6.00\n10.30\n6.00\n10.30\n192.23\n2\n18891259FPX9A8FC87\n2024-\n07-16\n2024-07-\n17\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n3\n32569109ION81BFB60\n2024-\n07-17\n2024-07-\n17\n17580780\nWudnKrafts\n3D\tWooden\nCapital\nAlphabet,\nSmall\nAlpha\nVCWKTYWO-\n156\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n4\n22721283TNAD70E37E\n2024-\n07-16\n2024-07-\n17\n17580778\nWudnKrafts\nWooden\t3D\nCapital\nAlphabets\nand\nNumber\nVCWKTYWO-\n154\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n5\n25364700ZYC1F5578D\n2024-\n07-17\n2024-07-\n17\n17580772\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n148\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n6\n10332839GLBFEF123E\n2024-\n07-16\n2024-07-\n17\n17580769\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n145\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n7\n7426279WON2C9F654\n2024-\n07-17\n2024-07-\n17\n17580737\nWudnKrafts\nWooden\nCalendar\nClock\nPuzzle\nWooden\nClo\nVCWKTYWO-\n113\n95036010@12\n1\n699.00\n40.00\n419.4\n205.96\n205.96\n6.00\n12.36\n6.00\n12.36\n230.68\n8\n21633944DEN1331729\n2024-\n07-16\n2024-07-\n17\n17580778\nWudnKrafts\nWooden\t3D\nCapital\nAlphabets\nand\nNumber\nVCWKTYWO-\n154\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n9\n7951380UBXFD83C24\n2024-\n07-16\n2024-07-\n17\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n10\n24805704JRID50AA6C\n2024-\n07-17\n2024-07-\n17\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n11\n31543566SPT6EF23E6\n2024-\n07-16\n2024-07-\n17\n17580765\nWudn\nKrafts\nWooden\nStick\tBaby\nRattles\nwith\t12\nGhun\nVCWKTYWO-\n141\n95036010@12\n1\n199.00\n0.00\n199\n97.72\n97.72\n6.00\n5.86\n6.00\n5.86\n109.45\n12\n17940431JNWB7E94B4\n2024-\n07-16\n2024-07-\n17\n17580778\nWudnKrafts\nWooden\t3D\nCapital\nAlphabets\nand\nNumber\nVCWKTYWO-\n154\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n13\n11531993THGABB4BC5\n2024-\n07-16\n2024-07-\n17\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n13\n7387.00\n2096.89\n2096.89\n112.63\n112.63\n2322.15\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n12.00\n11\n1720.23\n6.00\n103.21\n6.00\n103.21\n206.43"
        },
        {
            "page_number": 2,
            "content": "5.00\n2\n376.66\n2.50\n9.42\n2.50\n9.42\n18.83\nTotal\n13\n2096.89\n112.63\n112.63\n225.26\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n950300\n3\n548.29\n19.71\n19.71\n587.72\n95036010\n10\n1548.60\n92.92\n92.92\n1734.43\nGrand\tTotal\n13\n2096.89\n112.63\n112.63\n2322.15\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n13\n2096.89\n112.63\n112.63\n2322.15"
        }
    ],
    "08Jul_Invoice_620_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 620\nDate: 08/07/2024\nPlace of Supply: 33-Tamil NaduBill To:\nSamy Traders\n77 G-1,CHIDAMBARAM FLATS MANDAPAM LINK ROAD,\nKRISHNA NAGAR, Ullagaram, Chennai-600091\nContact No.: 9150844735\nGSTIN Number: 33CHDPT1108G1ZP\nState: 33-Tamil NaduTransporta\u0000on Details:\nVehicle Number: TN 29 BR 6160\nDelivery Date: 09/07/24\nDelivery Loca\u0000on: Chennai\nShip To:\n77 G-1,CHIDAMBARAM FLATS MANDAPAM LINK ROAD,\nKRISHNA NAGAR, Ullagaram, Chennai-600091\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit Amount\n1 Cane Jaggery Ball 1701 90 Bag \u20b9 1,515.00 \u20b9 1,36,350.00\n2 Cane Jaggery Powder 1701 2 Bag \u20b9 1,710.00 \u20b9 3,420.00\nTotal 92 \u20b9 1,39,770.00\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nOne Lakh Thirty Nine Thousand Seven Hundred Seventy Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 1,39,770.00\nTotal \u20b9 1,39,770.00"
        }
    ],
    "FY240701997637.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n639\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-01\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n25397805RVTD08976A\n2024-\n07-01\n2024-07-\n01\n17580780\nWudnKrafts\n3D\tWooden\nCapital\nAlphabet,\nSmall\nAlpha\nVCWKTYWO-\n156\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n2\n32221529NVX915B342\n2024-\n07-01\n2024-07-\n01\n17580773\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n149\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n3\n21403099JYL890B261\n2024-\n07-01\n2024-07-\n01\n17580778\nWudnKrafts\nWooden\t3D\nCapital\nAlphabets\nand\nNumber\nVCWKTYWO-\n154\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n4\n12310385MSC9ACB27A\n2024-\n07-01\n2024-07-\n01\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n4\n2096.00\n622.46\n622.46\n37.35\n37.35\n697.16\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n12.00\n4\n622.46\n6.00\n37.35\n6.00\n37.35\n74.70\nTotal\n4\n622.46\n37.35\n37.35\n74.70\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n95036010\n4\n622.46\n37.35\n37.35\n697.16\nGrand\tTotal\n4\n622.46\n37.35\n37.35\n697.16\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n4\n622.46\n37.35\n37.35\n697.16"
        }
    ],
    "INV-149_Karishma Bande.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-149 Invoice Date: 22 Mar 2024 Due Date: 01 Apr 2024\nCustomer Details:\nKarishma Bande\nPh: 8956804380\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Excela Moisturiser - 50 gm  370.64\n421.19 (-12%)1 BTL370.6466.72 (18%)437.36\nTaxable Amount \u20b9370.64\nCGST\u00a09.0% \u20b933.36\nSGST\u00a09.0% \u20b933.36\nRound Off -0.36\nTotal \u20b9437.00\nTotal Discount \u20b959.64\nTotal Items / Qty : 1 / 1.000 Total amount (in words): \u00a0INR Four Hundred And Thirty-Seven Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "INV-113_Raghvendra.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-113 Invoice Date: 29 Jan 2024 Due Date: 29 Jan 2024\nCustomer Details:\nRaghvendra\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Arachitol Nano (60k) 4*5ml  299.58\n340.43 (-12%)3 BTL898.73107.85 (12%)1,006.58\nTaxable Amount \u20b9898.73\nCGST\u00a06.0% \u20b953.92\nSGST\u00a06.0% \u20b953.92\nRound Off 0.42\nTotal\u20b91,007.00\nTotal Discount \u20b9137.26\nTotal Items / Qty : 1 / 3.000 Total amount (in words): \u00a0INR One Thousand And Seven Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "INV-809.pdf": [
        {
            "page_number": 1,
            "content": "1\n POWERED BY\nSub Total 30,250.00\nCGST9 (9%) 2,722.50\nSGST9 (9%) 2,722.50\nTotal \u20b935,695.00\nBalance Due \u20b935,695.00\nAuthorized SignatureTotal In Words\nIndian Rupee Thirty-Five Thousand Six Hundred Ninety-Five Only\nNotes\nThanks for your business.\nTerms & Conditions\nGoods Once sold cannot be taken back.\nSRI BHAVANI RUBBER & PLASTICS\nNo, 1, Thilagar Nagar 1st Street, Ashok Sylicate Campus,\nKaladipet, Thiruvottiyur,\nChennai Tamil Nadu 600019\nIndia\nGSTIN 33AFIPM5717R1ZV\nsribhavanirubber.plastics@gmail.com TAX INVOICE\n  # : INV-809\n  Invoice Date : 31/07/2024\n  Terms : Due on Receipt\n Due Date : 31/07/2024 Place Of Supply : Tamil Nadu (33)\nBill To Ship To\nExcel Engineering Works\nNew no:1003/2,Old no:326,\nT.H.Road,Old Washermanpet,\nChennai\n600021 Tamil Nadu\nIndia\nGSTIN 33AFPPR8494F1ZSNew no:1003/2,Old no:326,\nT.H.Road,Old Washermanpet,\nChennai\n600021 Tamil Nadu\nIndia\nGSTIN 33AFPPR8494F1ZS\n# Item & DescriptionHSN\n/SAC Qty RateCGST SGST\nAmount % Amt % Amt\n1 BUSHM0015\nBrass Metal Bush, Size: 68 mm \nID * 89 mm OD * 114 mm \nheight 4016939\n03.00\nNos2,500.00 9% 675.00 9% 675.00 7,500.00\n2 BUSHM0014\nBrass Metal Bush, Size: 56 mm \nID * 75 mm OD * 135 mm \nheight 4016939\n013.00\nNos1,750.00 9% 2,047.50 9% 2,047.50 22,750.00"
        }
    ],
    "INV-99_Indraja.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-99 Invoice Date: 16 Jan 2024 Due Date: 16 Jan 2024\nCustomer Details:\nIndraja\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Dermatologist Consultation  300.00\n350.00 (-14.29%)1300.000.00 (0%)300.00\n2Acutret 20 mg - 10 capsules  211.59\n258.04 (-18%)5 STRP1,057.95126.95 (12%)1,184.90\n3Isotroin 20 mg - 10 capsules  271.23\n319.09 (-15%)1 STRP271.2332.55 (12%)303.77\n4Biluma cream - 15 gm  462.37\n525.42 (-12%)1 PAC462.3783.23 (18%)545.60\nTaxable Amount \u20b92,091.55\nCGST\u00a06.0% \u20b979.75\nSGST\u00a06.0% \u20b979.75\nCGST\u00a09.0% \u20b941.61\nSGST\u00a09.0% \u20b941.61\nRound Off -0.27\nTotal\u20b92,334.00\nTotal Discount \u20b9438.11\nTotal Items / Qty : 4 / 8.000 Total amount (in words): \u00a0INR Two Thousand, Three Hundred And Thirty-Four Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "INV-73_Avik Mallick.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-73 Invoice Date: 15 Feb 2024 Due Date: 26 Oct 2023\nCustomer Details:\nAvik Mallick\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1DermaQ Comprehensive Hair Health \nDiagnostic Test 1,521.60\n3,804.00 (-60%)11,521.600.00 (0%)1,521.60\n2Hair Nutrition Plan 499.17\n1,499.00 \n(-66.7%)1499.170.00 (0%)499.17\n38x Shampoo 350.43\n398.21 (-12%)1 BTL350.4342.05 (12%)392.48\n4LONITAB 2.5 mg 157.14\n178.57 (-12%)9 STRP1,414.29169.71 (12%)1,584.00\n5Anaboom AD Lotion - 50 ml  241.63\n274.58 (-12%)1 BTL241.6343.49 (18%)285.12\n6Uprise D-3 60K Capsules - 12  324.93\n369.24 (-12%)1 STRP324.9338.99 (12%)363.92\n7ME-12 OD (15 tabs)  100.18\n113.84 (-12%)6 STRP601.0772.13 (12%)673.20\n8tab dutamax 0.5 mg 102.14\n116.07 (-12%)2 STRP204.2924.51 (12%)228.80\n9Uprise-D3 60K - 8 Capsule   216.62\n246.16 (-12%)1 STRP216.6225.99 (12%)242.62\nTaxable Amount \u20b95,374.02\nCGST\u00a06.0% \u20b9186.70\nSGST\u00a06.0% \u20b9186.70\nCGST\u00a09.0% \u20b921.75\nSGST\u00a09.0% \u20b921.75\nRound Off 0.09\nTotal\u20b95,791.00\nTotal Discount \u20b93,796.34\nTotal Items / Qty : 9 / 23.000 Total amount (in words): \u00a0INR Five Thousand, Seven Hundred And Ninety-One Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "INV-133_Sheetal Kapur.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-133 Invoice Date: 01 Mar 2024 Due Date: 01 Mar 2024\nCustomer Details:\nSheetal Kapur\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Glogeous facewash 785.17\n923.73 (-15%)1 BTL785.17141.33 (18%)926.50\n2Bristaa Intense Cream  749.95\n872.03 (-14%)1 BTL749.95134.99 (18%)884.94\n3Anaboom AD Lotion - 50 ml  233.39\n274.58 (-15%)1 BTL233.3942.01 (18%)275.40\n4Anaboom AD Shampoo  290.10\n329.66 (-12%)1 BTL290.1052.22 (18%)342.32\n5Acutret 5 mg 81.18\n101.47 (-20%)3 STRP243.5429.22 (12%)272.76\nTaxable Amount \u20b92,302.15\nCGST\u00a06.0% \u20b914.61\nSGST\u00a06.0% \u20b914.61\nCGST\u00a09.0% \u20b9185.27\nSGST\u00a09.0% \u20b9185.27\nRound Off 0.08\nTotal\u20b92,702.00\nTotal Discount \u20b9471.03\nTotal Items / Qty : 5 / 7.000 Total amount (in words): \u00a0INR Two Thousand, Seven Hundred And Two Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "INV-136_Rishabh Ramola.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-136 Invoice Date: 15 Feb 2024 Due Date: 04 Mar 2024\nCustomer Details:\nRishabh Ramola\nPh: 7023511429\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Solasafe sunscreen gel spf 50  480.68\n600.85 (-20%)2961.36173.04 (18%)1,134.40\nTaxable Amount \u20b9961.36\nCGST\u00a09.0% \u20b986.52\nSGST\u00a09.0% \u20b986.52\nRound Off -0.40\nTotal\u20b91,134.00\nTotal Discount \u20b9283.60\nTotal Items / Qty : 1 / 2.000 Total amount (in words): \u00a0INR One Thousand, One Hundred And Thirty-Four Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "INV-144_Atia Latif.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-144 Invoice Date: 28 Mar 2024 Due Date: 28 Mar 2024\nCustomer Details:\nAtia Latif\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Bioderma Pigmentbio C-concentrate  2,363.64\n2,626.27 (-10%)1 PCS2,363.64425.46 (18%)2,789.10\n2Arachitol Nano (60k) 4*5ml  299.58\n340.43 (-12%)2 BTL599.1571.90 (12%)671.05\n3Solasafe sunscreen gel spf 50  450.64\n600.85 (-25%)31,351.91243.34 (18%)1,595.25\n4Lab test - full body checkup  2,000.00 36,000.000.00 (0%)6,000.00\n5GFC PRP 11,600.00 111,600.001,392.00 (12%)12,992.00\nTaxable Amount \u20b921,914.71\nCGST\u00a06.0% \u20b9731.95\nSGST\u00a06.0% \u20b9731.95\nCGST\u00a09.0% \u20b9334.40\nSGST\u00a09.0% \u20b9334.40\nRound Off -0.40\nTotal\u20b924,047.00\nTotal Discount \u20b9933.16\nTotal Items / Qty : 5 / 10.000 Total amount (in words): \u00a0INR Twenty-Four Thousand And Forty-Seven Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "INV-808.pdf": [
        {
            "page_number": 1,
            "content": "1\n POWERED BY\nSub Total 28,940.00\nCGST9 (9%) 2,604.60\nSGST9 (9%) 2,604.60\nTotal \u20b934,149.20\nBalance Due \u20b934,149.20\nAuthorized SignatureTotal In Words\nIndian Rupee Thirty-Four Thousand One Hundred Forty-Nine and \nTwenty Paise Only\nNotes\nThanks for your business.\nTerms & Conditions\nGoods Once sold cannot be taken back.\nSRI BHAVANI RUBBER & PLASTICS\nNo, 1, Thilagar Nagar 1st Street, Ashok Sylicate Campus,\nKaladipet, Thiruvottiyur,\nChennai Tamil Nadu 600019\nIndia\nGSTIN 33AFIPM5717R1ZV\nsribhavanirubber.plastics@gmail.com TAX INVOICE\n  # : INV-808\n  Invoice Date : 21/07/2024\n  Terms : Due on Receipt\n Due Date : 21/07/2024 Place Of Supply : Tamil Nadu (33)\nBill To Ship To\nRUBBERAGE INDUSTRIES\nFirst Floor, No 41,\nSembudoss Street,\nChennai\n600001 Tamil Nadu\nIndia\nGSTIN 33AAAFR0628M1ZBFirst Floor, No 41,\nSembudoss Street,\nChennai\n600001 Tamil Nadu\nIndia\nGSTIN 33AAAFR0628M1ZB\n# Item & DescriptionHSN\n/SAC Qty RateCGST SGST\nAmount % Amt % Amt\n1 OILSEALD0019\nOil Seal Die, Size: 89mm * \n117.5mm * 16mm thick8480490\n01.00\nNos1,800.00 9% 162.00 9% 162.00 1,800.00\n2 OILSEALM0024\nOil Seal metal Ring, Size: \n89mm * 117.5mm * 16mm \nthick8480490\n04.00\nNos60.00 9% 21.60 9% 21.60 240.00\n3 MOUNTPADD001\nMounting pad as per drawing, \nSize: 554 * 150 *50 thick8480490\n01.00\nNos11,500.00 9% 1,035.00 9% 1,035.00 11,500.00\n4 MOUNTPADM001\nMounting Metal Peace, Size: \n554 * 150 *50 thick8480490\n020.00\nNos600.00 9% 1,080.00 9% 1,080.00 12,000.00\n5 SQRBLACK001\nSquare Black Die, Size: 65 mm \n* 45 mm * 20 mm Thick8480490\n01.00\nNos700.00 9% 63.00 9% 63.00 700.00\n6 FLATRINGD0008\nFlat ring, size: 21 * 15 * 4 Thick8480490\n01.00\nNos200.00 9% 18.00 9% 18.00 200.00\n7 RECTANBEEDINGD001\nRectangular Beeding as per \nDRG, Size: 50mm * 26mm * \n9mm Thick8480490\n01.00\nNos750.00 9% 67.50 9% 67.50 750.00\n8 DUMMYCUPD001\nDummy Cup, Size: 145mm * \n140mm * 12mm Thick 8480490\n01.00\nNos1,500.00 9% 135.00 9% 135.00 1,500.00\n9 DUMMYCUPM001\nDummy Metal Cup, Size: \n145mm * 140mm * 12mm \nThick 8480490\n05.00\nNos50.00 9% 22.50 9% 22.50 250.00"
        }
    ],
    "INV-134_Sheetal Kapur.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-134 Invoice Date: 01 Mar 2024 Due Date: 01 Mar 2024\nCustomer Details:\nSheetal Kapur\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1ATODERM Moisturiser  723.77\n761.86 (-5%)1 BTL723.77130.28 (18%)854.05\nTaxable Amount \u20b9723.77\nCGST\u00a09.0% \u20b965.14\nSGST\u00a09.0% \u20b965.14\nRound Off -0.05\nTotal \u20b9854.00\nTotal Discount \u20b944.95\nTotal Items / Qty : 1 / 1.000 Total amount (in words): \u00a0INR Eight Hundred And Fifty-Four Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "26Jul_Invoice_641_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 641\nDate: 26/07/2024\nPlace of Supply: 29-KarnatakaBill To:\nSri Mahalakshmi Enterprises\nNO.10/41 APMC YARD 6TH MAIN ROAD YESHWANTHPUR\nContact No.: 08026705904, 08048903677\nGSTIN Number: 29AGWPA9333Q1ZE\nState: 29-KarnatakaTransporta\u0000on Details:\nDelivery Date: 26/07/24\nDelivery Loca\u0000on: Yeshwanthpur\nShip To:\nNO.10/41 APMC YARD 6TH MAIN ROAD YESHWANTHPUR\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit GST Amount\n1 HTO Jaggery Powder 1701 35 Box \u20b9 590.00\u20b9 1,032.50 (5%) \u20b9 21,682.50\nTotal 35 \u20b9 1,032.50 \u20b9 21,682.50\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nTwenty One Thousand Six Hundred Eighty Three Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 20,650.00\nIGST@5% \u20b9 1,032.50\nRound o\ufb00 \u20b9 0.50\nTotal \u20b9 21,683.00"
        }
    ],
    "FY2407241022519.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n662\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-24\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n4209301TGMAF18208\n2024-\n07-24\n2024-07-\n24\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n2\n20310392HIVF138DEE\n2024-\n07-23\n2024-07-\n24\n14546399\nZivilion\nShape\tand\nColor\nSorting\nCube\nGeometric\nSh\nVCZITYHT-\n30\n950300@5\n1\n699.00\n55.00\n314.55\n164.76\n164.76\n2.50\n4.12\n2.50\n4.12\n173.00\n3\n15801450IJA825AB80\n2024-\n07-24\n2024-07-\n24\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n4\n15170353WZZE5F7EFA\n2024-\n07-23\n2024-07-\n24\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n5\n32690422SOS210F297\n2024-\n07-23\n2024-07-\n24\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n6\n23803575GFI0B8EFE8\n2024-\n07-23\n2024-07-\n24\n17580778\nWudnKrafts\nWooden\t3D\nCapital\nAlphabets\nand\nNumber\nVCWKTYWO-\n154\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n7\n22585034WTJFAF35F0\n2024-\n07-23\n2024-07-\n24\n17580779\nWudnKrafts\nWooden\t3D\nAlphabet\nand\nNumber\nBoard\tCom\nVCWKTYWO-\n155\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n8\n25581531YJYA5279BC\n2024-\n07-23\n2024-07-\n24\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n9\n32372149NTX11A2389\n2024-\n07-24\n2024-07-\n24\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n10\n10406214MQRAF06A41\n2024-\n07-23\n2024-07-\n24\n17580773\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n149\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n10\n5790.00\n1528.41\n1528.41\n66.16\n66.16\n1660.74\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n5.00\n4\n729.75\n2.50\n18.24\n2.50\n18.24\n36.49\n12.00\n6\n798.66\n6.00\n47.92\n6.00\n47.92\n95.84\nTotal\n10\n1528.41\n66.16\n66.16\n132.33\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n950300\n4\n729.75\n18.24\n18.24\n766.24\n95036010\n6\n798.66\n47.92\n47.92\n894.50\nGrand\tTotal\n10\n1528.41\n66.16\n66.16\n1660.74\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n10\n1528.41\n66.16\n66.16\n1660.74"
        }
    ],
    "OD14425-sales_invoice-Joyce_Supriya_Sakani_compressed.pdf": [
        {
            "page_number": 1,
            "content": "Opulent Duende\nFourth Floor, Ft.405, Eeco Valley-west block, Madhavapuri Hills,Ameenapur, Near GSM Mall,\nChanda Nagar, Hyderabad, Rangareddy, Hyderabad, Telangana, 500050\nMobile: 9885001313 GSTIN:36BARPS5773E1Z2\nEmail:opulentduende@gmail.com\nInvoice No. :OD14425 Invoice Date :25/08/2024\nBILL TO\nJoyce Supriya Sakani\nMobile:9885134663\nSERVICES SAC TAX AMOUN T\nQUEENPRO BOT CREDITS\nLicensing services for the right to use computer software and databases.997331 1,525.42\n(18%)10,000\nSUB TOTAL \u20b9 1,525.42 \u20b9 10,000\nBANK DE TAILS\nName: Opulent Duende\nIFSC Code:UTIB0000538\nAccount No:924020007176709\nBank: Axis Bank,CHANDANAGAR\nPAYMEN T QR CODE\nUPI ID:\nopulentduende@axisbank\nTERMS AND CONDI TIONS\nFees once collected is non-refundable.TAXABLE AMOUNT \u20b9 8,474.58\nCGST @9% \u20b9 762.71\nSGST @9% \u20b9 762.71\nTOTAL AMOUN T \u20b9 10,000\nReceived Amount \u20b9 10,000\nBalance \u20b9 0\nTotal A mount (in wor ds)\nTen Thousand RupeesTAX INVOICEORIGINAL FOR RECIPIENT"
        }
    ],
    "INV-123_Asit.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-123 Invoice Date: 08 Feb 2024 Due Date: 08 Feb 2024\nCustomer Details:\nAsit\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Arachitol Nano (60k) 4*5ml  299.58\n340.43 (-12%)3 BTL898.73107.85 (12%)1,006.58\n2Neurobion Forte - 30 tablets   30.58\n34.75 (-12%)3 STRP91.7316.51 (18%)108.24\nTaxable Amount \u20b9990.46\nCGST\u00a06.0% \u20b953.92\nSGST\u00a06.0% \u20b953.92\nCGST\u00a09.0% \u20b98.26\nSGST\u00a09.0% \u20b98.26\nRound Off 0.18\nTotal\u20b91,115.00\nTotal Discount \u20b9152.02\nTotal Items / Qty : 2 / 6.000 Total amount (in words): \u00a0INR One Thousand, One Hundred And Fifteen Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "12Jul_Invoice_626_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 626\nDate: 12/07/2024\nPlace of Supply: 29-KarnatakaBill To:\nSri Mahalakshmi Enterprises\nNO.10/41 APMC YARD 6TH MAIN ROAD YESHWANTHPUR\nContact No.: 08026705904, 08048903677\nGSTIN Number: 29AGWPA9333Q1ZE\nState: 29-KarnatakaTransporta\u0000on Details:\nDelivery Date: 12/07/24\nDelivery Loca\u0000on: Yeshwanthpur\nShip To:\nNO.10/41 APMC YARD 6TH MAIN ROAD YESHWANTHPUR\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit GST Amount\n1 Cane Jaggery Ball 1701 35 Bag \u20b9 1,390.00 \u20b9 0.00 (0%) \u20b9 48,650.00\n2 HTO Jaggery Powder 1701 21 Box \u20b9 590.00 \u20b9 619.50 (5%) \u20b9 13,009.50\nTotal 56 \u20b9 619.50 \u20b9 61,659.50\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nSixty One Thousand Six Hundred Sixty Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 61,040.00\nIGST@5% \u20b9 619.50\nRound o\ufb00 \u20b9 0.50\nTotal \u20b9 61,660.00"
        }
    ],
    "OD14416-sales_invoice-Ravinder_Reddy_Yedulla_compressed.pdf": [
        {
            "page_number": 1,
            "content": "Opulent Duende\nFourth Floor, Ft.405, Eeco Valley-west block, Madhavapuri Hills,Ameenapur, Near GSM Mall,\nChanda Nagar, Hyderabad, Rangareddy, Hyderabad, Telangana, 500050\nMobile: 9885001313 GSTIN:36BARPS5773E1Z2\nEmail:opulentduende@gmail.com\nInvoice No. :OD14416 Invoice Date :15/08/2024\nBILL TO\nRavinder Reddy Yedulla\nMobile:9949515251\nSERVICES SAC TAX AMOUN T\nQUEENPRO BOT CREDITS\nLicensing services for the right to use computer software and databases.997331 1,525.42\n(18%)10,000\nSUB TOTAL \u20b9 1,525.42 \u20b9 10,000\nBANK DE TAILS\nName: Opulent Duende\nIFSC Code:UTIB0000538\nAccount No:924020007176709\nBank: Axis Bank,CHANDANAGAR\nPAYMEN T QR CODE\nUPI ID:\nopulentduende@axisbank\nTERMS AND CONDI TIONS\nFees once collected is non-refundable.TAXABLE AMOUNT \u20b9 8,474.58\nCGST @9% \u20b9 762.71\nSGST @9% \u20b9 762.71\nTOTAL AMOUN T \u20b9 10,000\nReceived Amount \u20b9 10,000\nBalance \u20b9 0\nTotal A mount (in wor ds)\nTen Thousand RupeesTAX INVOICEORIGINAL FOR RECIPIENT"
        }
    ],
    "Tax Invoice (6)_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nInvoice No :  LP1757/24062 Invoice Date :2024-07-13\nBill From : Bill To :\nName : 100 Cubes  HOME INTERIOR DESIGNS\nE-COMMERCE PRIV ATE LIMITED\nAddress : 117, St Johns church road, Fraser town,  \nBangalore  Address : 283/58/7, 3 Floor ,  Aurbis Business  \nPark, Outer ring road, Devarabisanahalli,  \nBellandur , BENGAL URU, Bengaluru (Bangalore),  \nUrban, Karnataka, 560103.  \nPincode : 560005  \nPAN : CITPK3346L  PAN : AADCH4222R\nGSTIN : 29CITPK3346L2ZG  GSTIN : 29AADCH4222R3Z3  \nState Name : Karnataka  State Name : Karnataka  \nState Code : 29 State Code : 29 \nSl.\nNo.Description of Services HSN/SAC Rate Amount\n1 Designer Services as a Liveprenuer from 16-30 June-\n2024 998599  11,648 \nCGST  Output\nSGST  Output 9%\n9% 1,048\n 1,048 \nIGST  Output 18% 0 \nGST Total   2,097 \n    \nGross Total 13,745  \nAmount Chargeable (in words) E. & O.E - Indian Rupees Thirteen Thousand Seven Hundred And Forty Four Only  \nDocument Ref: G7CTE-YKBJQ-TF55C-ELWBW Page 1 of 2"
        },
        {
            "page_number": 2,
            "content": "Declaration:\nThe supplier declares that the registration number mentioned on this invoice is valid registration under the GST  \nlaw applicable to the State of issue and the value mentioned on this invoice for the supply (including for supply  \nnot having consideration) shall be appropriately accounted for in the books and included in the turnover while  \nfilling the return under the GST  laws for ef fecting such supplies.\nContact Person Mobile No8660089169  \nContact Person Email IDpatanabdulahadkhan.lp@livspace.com  \nBank Account Number12630200000797  \nIFSC CodeBARB0FRASER  \nBeneficiary Name100 Cubes  \n**Note : TDS will be deducted on Taxable Income V alue at applicable rate\nDocument Ref: G7CTE-YKBJQ-TF55C-ELWBW Page 2 of 2"
        },
        {
            "page_number": 3,
            "content": "Signature Certificate\nReference number: G7CTE-YKBJQ-TF55C-ELWBW\nDocument completed by all parties on:\n13 Jul 2024 12:41:04 UTC\nPage 1 of 1Signer Timestamp Signature\nPatan Abdul Ahad Khan\nEmail: patanabdulahadkhan.lp@livspace.com\nRecipient Verification:Sent: 13 Jul 2024 06:23:38 UTC\nViewed: 13 Jul 2024 12:40:14 UTC\nSigned: 13 Jul 2024 12:41:04 UTC\n\u2714Email verified 13 Jul 2024 12:40:14 UTC\nIP address: 223.231.155.227\nLocation: Bengaluru, India\nSigned with PandaDoc\nPandaDoc is a document workflow and certified eSignature\nsolution trusted by 50,000+ companies worldwide."
        }
    ],
    "OD14414-sales_invoice-Ravinder_Reddy_Yedulla_compressed.pdf": [
        {
            "page_number": 1,
            "content": "Opulent Duende\nFourth Floor, Ft.405, Eeco Valley-west block, Madhavapuri Hills,Ameenapur, Near GSM Mall,\nChanda Nagar, Hyderabad, Rangareddy, Hyderabad, Telangana, 500050\nMobile: 9885001313 GSTIN:36BARPS5773E1Z2\nEmail:opulentduende@gmail.com\nInvoice No. :OD14414 Invoice Date :12/08/2024\nBILL TO\nRavinder Reddy Yedulla\nMobile:9949515251\nSERVICES SAC TAX AMOUN T\nQUEENPRO BOT CREDITS\nLicensing services for the right to use computer software and databases.997331 1,452.2\n(18%)9,520\nSUB TOTAL \u20b9 1,452.2 \u20b9 9,520\nBANK DE TAILS\nName: Opulent Duende\nIFSC Code:UTIB0000538\nAccount No:924020007176709\nBank: Axis Bank,CHANDANAGAR\nPAYMEN T QR CODE\nUPI ID:\nopulentduende@axisbank\nTERMS AND CONDI TIONS\nFees once collected is non-refundable.TAXABLE AMOUNT \u20b9 8,067.8\nCGST @9% \u20b9 726.1\nSGST @9% \u20b9 726.1\nTOTAL AMOUN T \u20b9 9,520\nReceived Amount \u20b9 9,520\nBalance \u20b9 0\nTotal A mount (in wor ds)\nNine Thousand Five Hundred Twenty RupeesTAX INVOICEORIGINAL FOR RECIPIENT"
        }
    ],
    "15.pdf": [
        {
            "page_number": 1,
            "content": "Campos Technologies\nF25-26, 1st Floor, Alfran Plaza, Opp. Naturals Ice Cream,\nNr. Don Bosco School, M.G Road, Panaji Goa 403001\nPhone no. : 9145511772\nEmail : campostechgoa@gmail.com\nGSTIN : 30BHLPC5928C2ZS\nState: 30-Goa\nInvoice\nBill To\nHIMTIK DESIGN\nH NO 3/6 CHAMUNDA RESIDENCY CARANZALEM\nCARANZALEM\nGSTIN : 30AAJFH3702R1ZB\nState: 30-GoaInvoice Details\nInvoice No. : CT/2024-25/15\nDate : 03-07-2024\nPlace of supply: 30-Goa\n# Item nameHSN/\nSACMRP Quantity UnitPrice/\nUnitGST Amount\n1Benq GP100 LED\nProjector\n(Warranty - 2 Years)8528\u20b9\n1,25,000.001 Pcs\u20b9\n67,203.39\u20b9\n12,096.61\n(18%)\u20b9\n79,300.00\n2Motorized Projector\nScreen\n(1 Year Warranty. Aspect Ratio\n16:9, 100\" inch)9011\u20b9\n21,999.001 Pcs\u20b9\n10,000.00\u20b9 1,800.00\n(18%)\u20b9\n11,800.00\n\u00a0\n\u00a0\n\u00a0\nTotal 2\u20b9\n13,896.61\u20b9\n91,100.00"
        },
        {
            "page_number": 2,
            "content": "For :Campos Technologies\nAuthorized SignatoryCampos Technologies\nF25-26, 1st Floor, Alfran Plaza, Opp. Naturals Ice Cream,\nNr. Don Bosco School, M.G Road, Panaji Goa 403001\nPhone no. : 9145511772\nEmail : campostechgoa@gmail.com\nGSTIN : 30BHLPC5928C2ZS\nState: 30-Goa\nInvoice Amount In Words\nNinety One Thousand One Hundred Rupees only\nTerms and Conditions\nThank you for your business!Sub Total \u20b9 77,203.39\nSGST@9% \u20b9 6,948.31\nCGST@9% \u20b9 6,948.31\nTotal \u20b9 91,100.00\nReceived \u20b9 91,100.00\nBalance \u20b9 0.00\nPayment mode Kotak CTech Current\nYou Saved \u20b9 55,899.00"
        }
    ],
    "INV-100_Agrani Kandele.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-100 Invoice Date: 24 Jan 2024 Due Date: 24 Jan 2024\nCustomer Details:\nAgrani Kandele\nPh: 8120482988Shipping Address:\nvadandana beauty parlour\nMurawara\nKatni, MADHYA PRADESH, 483501\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Acne UV Gel - 30 SPF  531.99\n709.32 (-25%)1 UNT531.9995.76 (18%)627.75\n2Clindac-A mist spray 392.07\n445.54 (-12%)1 BTL392.0747.05 (12%)439.12\n3AKNAYBAR soap 184.20\n209.32 (-12%)1 BOX184.2033.16 (18%)217.36\nTaxable Amount \u20b91,108.27\nCGST\u00a06.0% \u20b923.52\nSGST\u00a06.0% \u20b923.52\nCGST\u00a09.0% \u20b964.46\nSGST\u00a09.0% \u20b964.46\nRound Off -0.23\nTotal\u20b91,284.00\nTotal Discount \u20b9298.77\nTotal Items / Qty : 3 / 3.000 Total amount (in words): \u00a0INR One Thousand, Two Hundred And Eighty-Four Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "OD14418-sales_invoice-Surendra_Munemmagari_compressed.pdf": [
        {
            "page_number": 1,
            "content": "Opulent Duende\nFourth Floor, Ft.405, Eeco Valley-west block, Madhavapuri Hills,Ameenapur, Near GSM Mall,\nChanda Nagar, Hyderabad, Rangareddy, Hyderabad, Telangana, 500050\nMobile: 9885001313 GSTIN:36BARPS5773E1Z2\nEmail:opulentduende@gmail.com\nInvoice No. :OD14418 Invoice Date :15/08/2024\nBILL TO\nSurendra Munemmagari\nMobile:9704799937\nSERVICES SAC TAX AMOUN T\nQUEENPRO BOT CREDITS\nLicensing services for the right to use computer software and databases.997331 3,203.39\n(18%)21,000\nSUB TOTAL \u20b9 3,203.39 \u20b9 21,000\nBANK DE TAILS\nName: Opulent Duende\nIFSC Code:UTIB0000538\nAccount No:924020007176709\nBank: Axis Bank,CHANDANAGAR\nPAYMEN T QR CODE\nUPI ID:\nopulentduende@axisbank\nTERMS AND CONDI TIONS\nFees once collected is non-refundable.TAXABLE AMOUNT \u20b9 17,796.61\nCGST @9% \u20b9 1,601.69\nSGST @9% \u20b9 1,601.69\nTOTAL AMOUN T \u20b9 21,000\nReceived Amount \u20b9 21,000\nBalance \u20b9 0\nTotal A mount (in wor ds)\nTwenty One Thousand RupeesTAX INVOICEORIGINAL FOR RECIPIENT"
        }
    ],
    "10Jul_Invoice_624_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 624\nDate: 10/07/2024\nPlace of Supply: 29-KarnatakaBill To:\nSri Mahalakshmi Enterprises\nNO.10/41 APMC YARD 6TH MAIN ROAD YESHWANTHPUR\nContact No.: 08026705904, 08048903677\nGSTIN Number: 29AGWPA9333Q1ZE\nState: 29-KarnatakaTransporta\u0000on Details:\nDelivery Date: 10/07/24\nDelivery Loca\u0000on: Yeshwanthpur\nShip To:\nNO.10/41 APMC YARD 6TH MAIN ROAD YESHWANTHPUR\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit Amount\n1 Cane Jaggery Ball 1701 25 Bag \u20b9 1,390.00 \u20b9 34,750.00\nTotal 25 \u20b9 34,750.00\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nThirty Four Thousand Seven Hundred Fi\u0000y Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 34,750.00\nTotal \u20b9 34,750.00"
        }
    ],
    "IN3-100C24052504_signed_compressed.pdf": [
        {
            "page_number": 1,
            "content": "1Sub Total \n(Tax Inclusive)4,78,358.00\nIGST18 (18%) 72,969.86\nTotal \u20b94,78,358.00\nPayment Made (-) 4,78,358.00\nBalance Due \u20b90.00\nAuthorized SignatureTotal In Words\nIndian Rupee Four Lakh Seventy-Eight Thousand Three Hundred \nFifty-Eight Only\nNotes\nThanks for your Patronage.\n100 Cubes Home Interiors\nKarnataka \nIndia\nGSTIN 29CITPK3346L2ZG\n9606472571\npatanabdulahadkhan@gmail.com\nTAX INVOICE\n  # : IN3-100C24052504\n  Invoice Date : 25/07/2024\n  Terms : Due on Receipt\n Due Date : 25/07/2024 Place Of Supply : Chhattisgarh (22)\nBill To Ship To\nXPERIENCE SERVICE\nSMART NU TOWN, PIRDA ROAD, PIRDA 2, RAIPUR\nCHATTISGARH\n492005 Chhattisgarh\nIndia\nGSTIN 22FAAPS6430M1ZJGSTIN 22FAAPS6430M1ZJ\n# Item & Description HSN/SAC Qty RateIGST\nAmount % Amt\n1 50% Payment stage- Home Interiors 998391 1.00 4,78,358.00 18% 72,969.86 4,78,358.00\n"
        }
    ],
    "OD14427-sales_invoice-Anurag_Sikka_compressed.pdf": [
        {
            "page_number": 1,
            "content": "Opulent Duende\nFourth Floor, Ft.405, Eeco Valley-west block, Madhavapuri Hills,Ameenapur, Near GSM Mall,\nChanda Nagar, Hyderabad, Rangareddy, Hyderabad, Telangana, 500050\nMobile: 9885001313 GSTIN:36BARPS5773E1Z2\nEmail:opulentduende@gmail.com\nInvoice No. :OD14427 Invoice Date :26/08/2024\nBILL TO\nAnurag Sikka\nMobile:9643891842\nSERVICES SAC TAX AMOUN T\nQUEENPRO BOT CREDITS\nLicensing services for the right to use computer software and databases.997331 762.71\n(18%)5,000\nSUB TOTAL \u20b9 762.71 \u20b9 5,000\nBANK DE TAILS\nName: Opulent Duende\nIFSC Code:UTIB0000538\nAccount No:924020007176709\nBank: Axis Bank,CHANDANAGAR\nPAYMEN T QR CODE\nUPI ID:\nopulentduende@axisbank\nTERMS AND CONDI TIONS\nFees once collected is non-refundable.TAXABLE AMOUNT \u20b9 4,237.29\nCGST @9% \u20b9 381.36\nSGST @9% \u20b9 381.36\nTOTAL AMOUN T \u20b9 5,000\nReceived Amount \u20b9 5,000\nBalance \u20b9 0\nTotal A mount (in wor ds)\nFive Thousand RupeesTAX INVOICEORIGINAL FOR RECIPIENT"
        }
    ],
    "IN5-100C24020601 (1)_compressed.pdf": [
        {
            "page_number": 1,
            "content": "1Sub Total \n(Tax Inclusive)2,96,414.00\nCGST9 (9%) 22,607.85\nSGST9 (9%) 22,607.85\nTotal \u20b92,96,414.00\nPayment Made (-) 2,96,414.00\nBalance Due \u20b90.00\nAuthorized SignatureTotal In Words\nIndian Rupee Two Lakh Ninety-Six Thousand Four Hundred Fourteen \nOnly\nNotes\nThanks for your Patronage.\n100 Cubes Home Interiors\nKarnataka \nIndia\nGSTIN 29CITPK3346L2ZG\n9606472571\npatanabdulahadkhan@gmail.com\nTAX INVOICE\n  # : IN5-100C24020601\n  Invoice Date : 21/07/2024\n  Terms : Due on Receipt\n Due Date : 21/07/2024 Place Of Supply : Karnataka (29)\nBill To\nMr. Somenath Ghosh\n# Item & DescriptionHSN\n/SAC Qty RateCGST SGST\nAmount % Amt % Amt\n1 90% Stage payment - Home\nInteriors998391 1.00 2,96,414.0\n09% 22,607.85 9% 22,607.85 2,96,414.00"
        }
    ],
    "24Jul_Invoice_640_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 640\nDate: 24/07/2024\nPlace of Supply: 33-Tamil NaduBill To:\nVARASAI COMMERCIALS\n1St Floor 22 ANNA PILLAI STREET, \nPARRYS, Georgetown,\nCHENNAI-600 001.\nContact No.: 9884927736\nGSTIN Number: 33AALFV6889P1ZZ\nState: 33-Tamil NaduTransporta\u0000on Details:\nVehicle Number: TN 29 BR 6160\nDelivery Date: 25/07/24\nDelivery Loca\u0000on: Chennai\nShip To:\n1St Floor 22 ANNA PILLAI STREET, \n PARRYS, Georgetown,\n CHENNAI-600 001.\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit Amount\n1 Cane Jaggery Ball 1701 100 Bag \u20b9 1,390.00 \u20b9 1,39,000.00\n2 Cane Jaggery Ball 1701 75 Bag \u20b9 2,700.00 \u20b9 2,02,500.00\nTotal 175 \u20b9 3,41,500.00\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nThree Lakh Forty One Thousand Five Hundred Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 3,41,500.00\nTotal \u20b9 3,41,500.00"
        }
    ],
    "FY2407271026065.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n665\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-27\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n10880911IDCEA2765F\n2024-\n07-27\n2024-07-\n27\n17580778\nWudnKrafts\nWooden\t3D\nCapital\nAlphabets\nand\nNumber\nVCWKTYWO-\n154\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n2\n9227571KRHC15CFEB\n2024-\n07-26\n2024-07-\n27\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n3\n11541222IQJ9A99D2A\n2024-\n07-27\n2024-07-\n27\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n4\n32744451USWC56A016\n2024-\n07-27\n2024-07-\n27\n17580778\nWudnKrafts\nWooden\t3D\nCapital\nAlphabets\nand\nNumber\nVCWKTYWO-\n154\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n4\n2696.00\n747.27\n747.27\n38.24\n38.24\n823.76\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n12.00\n3\n558.94\n6.00\n33.54\n6.00\n33.54\n67.07\n5.00\n1\n188.33\n2.50\n4.71\n2.50\n4.71\n9.42\nTotal\n4\n747.27\n38.24\n38.24\n76.49\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n950300\n1\n188.33\n4.71\n4.71\n197.75\n95036010\n3\n558.94\n33.54\n33.54\n626.01\nGrand\tTotal\n4\n747.27\n38.24\n38.24\n823.76\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n4\n747.27\n38.24\n38.24\n823.76"
        }
    ],
    "FY2407101007378.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n648\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-10\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\tCode\nHSN\tCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n30010854CAN679B1C5\n2024-\n07-09\n2024-07-\n10\n17580773\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n149\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n2\n14831932FANF675C34\n2024-\n07-09\n2024-07-\n10\n17580741\nWudnKrafts\n100\tpcs\t12\nColor\nWooden\nDominos\nConstru\nVCWKTYWO-\n117\n95036010@12\n1\n699.00\n40.00\n419.4\n205.96\n205.96\n6.00\n12.36\n6.00\n12.36\n230.68\n3\n25337673SICAEC24E1\n2024-\n07-09\n2024-07-\n10\n17580747\nWudnKrafts\nColourful\nWooden\nBaby\tRattle\nMusical\tTo\nVCWKTYWO-\n123\n95036010@12\n1\n399.00\n40.00\n239.4\n117.56\n117.56\n6.00\n7.05\n6.00\n7.05\n131.67\n4\n20177692PCX56A5444\n2024-\n07-10\n2024-07-\n10\n17580730\nWudnKrafts\nMontessori\nWooden\nAlphabet\nLetters\nwith\nVCWKTYWO-\n106\n95036010@12\n1\n699.00\n40.00\n419.4\n205.96\n205.96\n6.00\n12.36\n6.00\n12.36\n230.68\n5\n12668851JMI8DE9945\n2024-\n07-10\n2024-07-\n10\n17580778\nWudnKrafts\nWooden\t3D\nCapital\nAlphabets\nand\nNumber\nVCWKTYWO-\n154\n95036010@12\n1\n799.00\n40.00\n479.4\n235.42\n235.42\n6.00\n14.13\n6.00\n14.13\n263.67\n6\n11821443MJSDA116B4\n2024-\n07-09\n2024-07-\n10\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\tGame\nfor\tK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n7\n31443750WQJ16CE13B\n2024-\n07-09\n2024-07-\n10\n17580748\nWudnKrafts\nColourful\nWooden\nBaby\nRattles\nMusical\tT\nVCWKTYWO-\n124\n95036010@12\n1\n299.00\n40.00\n179.4\n88.10\n88.10\n6.00\n5.29\n6.00\n5.29\n98.67\n8\n13906646XXUA751E07\n2024-\n07-09\n2024-07-\n10\n17580769\nWudnKrafts\nWooden\nTeddy\tFace\nStick\tBaby\nRattles\twi\nVCWKTYWO-\n145\n95036010@12\n1\n199.00\n35.00\n129.35\n63.52\n63.52\n6.00\n3.81\n6.00\n3.81\n71.14\n9\n3759384DUD8150B27\n2024-\n07-09\n2024-07-\n10\n17580728\nWudnKrafts\nWooden\nCalculation\nShelf\nLearning\nand\tE\nVCWKTYWO-\n104\n95036010@12\n1\n799.00\n35.00\n519.35\n255.04\n255.04\n6.00\n15.30\n6.00\n15.30\n285.64\n9\n4891.00\n1423.41\n1423.41\n78.81\n78.81\n1581.04\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n12.00\n8\n1235.08\n6.00\n74.10\n6.00\n74.10\n148.21\n5.00\n1\n188.33\n2.50\n4.71\n2.50\n4.71\n9.42\nTotal\n9\n1423.41\n78.81\n78.81\n157.63\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n950300\n1\n188.33\n4.71\n4.71\n197.75\n95036010\n8\n1235.08\n74.10\n74.10\n1383.29\nGrand\tTotal\n9\n1423.41\n78.81\n78.81\n1581.04\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n9\n1423.41\n78.81\n78.81\n1581.04"
        }
    ],
    "02Jul_Invoice_614_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 614\nDate: 02/07/2024\nPlace of Supply: 33-Tamil NaduBill To:\nGrace Mart Private Limited\nNo.21/4B2, Noombal Road,\nPuliambedu,\nChennai - 600077\nContact No.: 9095334777\nGSTIN Number: 33AAKCG0744Q1Z9\nState: 33-Tamil NaduTransporta\u0000on Details:\nVehicle Number: TN 29 BR 6160\nDelivery Date: 03/07/24\nDelivery Loca\u0000on: Chennai\nShip To:\nNo.21/4B2, Noombal Road,\n Puliambedu,\n Chennai - 600077\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit Amount\n1 Cane Jaggery Ball 1701 110 Bag \u20b9 1,515.00 \u20b9 1,66,650.00\n2 Cane Jaggery Powder 1701 100 Bag \u20b9 1,450.00 \u20b9 1,45,000.00\nTotal 210 \u20b9 3,11,650.00\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nThree Lakh Eleven Thousand Six Hundred Fi\u0000y Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 3,11,650.00\nTotal \u20b9 3,11,650.00"
        }
    ],
    "23Jul_Invoice_637_compressed.pdf": [
        {
            "page_number": 1,
            "content": "TAX INVOICE\nKPP Hometown Organics\nAddress:\n6/549 Puthur Maariyamman Kovil st,\nBye Pass 4 Cross Road Palacode,\nDharmapuri, Tamilnadu - 636808Phone:\n9994218657\nEmail:\norganicsfromhometown@gmail.comGSTIN:\n33BDNPN7360K1Z3\nState:\n33 - Tamil Nadu\nInvoice No.: 637\nDate: 23/07/2024\nPlace of Supply: 33-Tamil NaduBill To:\nGrace Mart Private Limited\nNo.21/4B2, Noombal Road,\nPuliambedu,\nChennai - 600077\nContact No.: 9095334777\nGSTIN Number: 33AAKCG0744Q1Z9\nState: 33-Tamil NaduTransporta\u0000on Details:\nVehicle Number: TN 76 Z 6768\nDelivery Date: 24/07/24\nDelivery Loca\u0000on: Chennai\nShip To:\nNo.21/4B2, Noombal Road,\n Puliambedu,\n Chennai - 600077\n# Item name HSN/ SAC Quan\u0000ty Unit Price/ Unit Amount\n1 Cane Jaggery Ball 1701 100 Bag \u20b9 1,560.00 \u20b9 1,56,000.00\n2 Cane Jaggery Powder 1701 70 Bag \u20b9 1,450.00 \u20b9 1,01,500.00\nTotal 170 \u20b9 2,57,500.00\nPay To:\nBank Name : AXIS BANK, DHARMAPURI\nBank Account No. : 921020007484915\nBank IFSC code : UTIB0000889\nAccount holder's name : KPP Hometown Organics\n \nInvoice Amount In Words\nTwo Lakh Fi\u0000y Seven Thousand Five Hundred Rupees only\nTerms And Condi\u0000ons\n-Price di\ufb00erences between invoice issued for the same products may be due to market \ufb02uctua\u0000on and/or\nchanges in other costs of manufacturing or on raw material acquisi\u0000on.\n -Products are properly packed and secured in such a manner to reach their des\u0000na\u0000on in\n good state under the normal condi\u0000ons of transport.\n -The Customer should examine products immediately a\u0000er delivery and the Company will not be liable for\nany mis-delivery, shortage, defect or damage unless the Company receive details in wri\u0000ng within the date\nof the arrival of goods.\n -In the event of damage or loss in transit, the Company will alter or replace products free of charge. The\nCustomer shall promptly no\u0000fy the Company wri\u0000en no\u0000\ufb01ca\u0000on of such damage or loss with a proof of\nphotographs within the date of delivery.\n -Payments are to be made to the Company without any deduc\u0000on or discount other than as stated in\nthese Terms or in the relevant invoice or statement.\n -All Invoices and Contracts placed with the Company are subject to these Terms and Condi\u0000ons. The\nCompany may at any \u0000me vary or alter these Terms and Condi\u0000ons.\n \nFor : KPP Hometown Organics\nAuthorized SignatorySub Total \u20b9 2,57,500.00\nTotal \u20b9 2,57,500.00"
        }
    ],
    "INV-150_Bhusan Naresh.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-150 Invoice Date: 22 Mar 2024 Due Date: 01 Apr 2024\nCustomer Details:\nBhusan Naresh\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Ahaglow Advanced Skin Rejuvenating \nFace Wash Gel - 100 gm  394.51\n448.31 (-12%)1 PAC394.5171.01 (18%)465.52\nTaxable Amount \u20b9394.51\nCGST\u00a09.0% \u20b935.51\nSGST\u00a09.0% \u20b935.51\nRound Off 0.48\nTotal \u20b9466.00\nTotal Discount \u20b963.48\nTotal Items / Qty : 1 / 1.000 Total amount (in words): \u00a0INR Four Hundred And Sixty-Six Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "INV-106_Kamakshi Thakkar.pdf": [
        {
            "page_number": 1,
            "content": "TA X  I N V O I C E O R I G I N A L  F O R  R E C I P I E N T \nUNCUE DERMACARE PRIVATE LIMITED\nGSTIN 23AADCU2395N1ZY     \nC/o KARUNA GUPTA KURELE, 1st Floor\nS.P Bungalow Ke Pichhe, Shoagpur Shahdol, Shahdol\nShahdol, MADHYA PRADESH, 484001\nMobile +91 8585960963   Email ruhi@dermaq.in\nInvoice #: INV-106 Invoice Date: 27 Jan 2024 Due Date: 27 Jan 2024\nCustomer Details:\nKamakshi Thakkar\nPlace of Supply: \n23-MADHYA PRADESH\n#Item Rate / Item QtyTaxable ValueTax AmountAmount\n1Comprehensive Hair & Health Test - S-\nilver  1,474.00\n1,699.00 \n(-13.24%)22,948.000.00 (0%)2,948.00\n2Arachitol Nano (60k) 4*5ml  299.58\n340.43 (-12%)3 BTL898.73107.85 (12%)1,006.58\n3Neurobion Forte - 30 tablets   30.58\n34.75 (-12%)3 STRP91.7316.51 (18%)108.24\nTaxable Amount \u20b93,938.46\nCGST\u00a06.0% \u20b953.92\nSGST\u00a06.0% \u20b953.92\nCGST\u00a09.0% \u20b98.26\nSGST\u00a09.0% \u20b98.26\nRound Off 0.18\nTotal\u20b94,063.00\nTotal Discount \u20b9602.02\nTotal Items / Qty : 3 / 8.000 Total amount (in words): \u00a0INR Four Thousand And Sixty-Three Rupees Only .\nAmount Paid\nPay using UPI: \nBank Details:\nBank:Kotak Mahindra Bank\nAccount #: 1146860541\nIFSC Code: kkbk0000725\nBranch: PUNE - CHINCHWAD\nUnCue Dermacare Pvt LtdFor UNCUE DERMACARE PRIVATE LIMITED  \nAuthorized Signatory \nSwipe | Simple Invoicing, Billing and Payments | Visit getswipe.in\nPage 1 / 1 This is a computer generated document and requires no signature.Powered By\n"
        }
    ],
    "19.pdf": [
        {
            "page_number": 1,
            "content": "Campos Technologies\nF25-26, 1st Floor, Alfran Plaza, Opp. Naturals Ice Cream,\nNr. Don Bosco School, M.G Road, Panaji Goa 403001\nPhone no. : 9145511772\nEmail : campostechgoa@gmail.com\nGSTIN : 30BHLPC5928C2ZS\nState: 30-Goa\nInvoice\nBill To\nMajestic Dockyard Private Limited\n221/1, Zorinto,\nSancoale, South Goa, 403710\nGSTIN : 30AAHCP6483C2ZL\nState: 30-GoaInvoice Details\nInvoice No. : CT/2024-25/19\nDate : 10-07-2024\nPlace of supply: 30-Goa\n# Item nameHSN/\nSACMRP Quantity UnitPrice/\nUnitGST Amount\n1Dell 7450 Latitude\nLaptop Renewed\n(Core i7 5th Gen, Refurbished\nLaptop, 8 GB RAM, 256 GB\nSSD, Windows 10 Pro, 1\nMonth Replacement\nWarranty, 1 Year Service\nWarranty)\nSerial No.: 668CH9284713020\u20b9\n1,09,000.001 Pcs\u20b9\n20,000.00\u20b9 3,600.00\n(18%)\u20b9\n23,600.00\n\u00a0\n\u00a0\n\u00a0\n\u00a0\nTotal 1\u20b9\n3,600.00\u20b9\n23,600.00\nInvoice Amount In Words\nTwenty Three Thousand Six Hundred Rupees only\nTerms and Conditions\nThank you for your business!Sub Total \u20b9 20,000.00\nSGST@9% \u20b9 1,800.00\nCGST@9% \u20b9 1,800.00\nTotal \u20b9 23,600.00\nReceived \u20b9 0.00\nBalance \u20b9 23,600.00\nPayment mode Credit\nYou Saved \u20b9 85,400.00"
        },
        {
            "page_number": 2,
            "content": "For :Campos Technologies\nAuthorized SignatoryCampos Technologies\nF25-26, 1st Floor, Alfran Plaza, Opp. Naturals Ice Cream,\nNr. Don Bosco School, M.G Road, Panaji Goa 403001\nPhone no. : 9145511772\nEmail : campostechgoa@gmail.com\nGSTIN : 30BHLPC5928C2ZS\nState: 30-Goa\n"
        }
    ],
    "FY2407251023733.pdf": [
        {
            "page_number": 1,
            "content": "Tax\tInvoice\nCompany\nVenture\tCommerce\nInvoice\tNo\n663\nAddress\nNo.\t74,\t2nd\tCross,\tKanaka\tNagar,\tHebbal,\tBengaluru\t-\t560032,\tBengaluru,\tKarnataka,\tIndia,\n560032\nDate\n2024-07-25\nMail\tid\ncare@venturecommerce.in\nContact\tDetails\n+91-7483788177\nIRN\tNo.:\nGst\tNo.\n29CCJPJ4032J1ZG\nCIN\tNo.\nQR\tCode:\nTo\nName\nDigital\tAge\tRetail\tPvt.\tLtd.\nPlace\nKarnataka\nAddress\nSy.No.\t80/6,\tChokkahalli\tVillage,\tHoskote\tTaluk,\tBanglore\t-\t562114\nGST\tNo.\n29AADCD8136E1ZN\nSrNo\nOrder\tId's\nOrder\nDate\nShipping\nDate\nProduct\nID\nProduct\nName\nVendor\nStyle\nCode\nHSN\nCode\nQty\nMRP\nVendor\nDiscount\nVendor\nSP\nBase\nCost\nGross\nAmount\nCGST\n%\nCGST\nAmount\nSGST\n%\nSGST\nAmount\nTotal\n1\n15902744QMA8A157EF\n2024-\n07-25\n2024-07-\n25\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\nGame\tfor\nK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n2\n20363197LAIF964D2D\n2024-\n07-24\n2024-07-\n25\n14546402\nZivilion\nMontessori\nSlide\nPuzzles\nBrain\nGame\tfor\nK\nVCZITYHT-\n33\n950300@5\n1\n799.00\n55.00\n359.55\n188.33\n188.33\n2.50\n4.71\n2.50\n4.71\n197.75\n3\n17212632GWC63D3E3D\n2024-\n07-24\n2024-07-\n25\n14546399\nZivilion\nShape\tand\nColor\nSorting\nCube\nGeometric\nSh\nVCZITYHT-\n30\n950300@5\n1\n699.00\n55.00\n314.55\n164.76\n164.76\n2.50\n4.12\n2.50\n4.12\n173.00\n3\n2297.00\n541.42\n541.42\n13.54\n13.54\n568.49\nTAX-Wise\tSummary\nTAX%\nQuantity\nBasic\nCGST%\nCGST\nSGST%\nSGST\nTotal\tTax\n5.00\n3\n541.42\n2.50\n13.54\n2.50\n13.54\n27.07\nTotal\n3\n541.42\n13.54\n13.54\n27.07\nHSN-Wise\tSummary\nHSN\tCode\nQuantity\nBasic\nCGST\tAmount\nSGST\tAmount\nTotal\tAmount\n950300\n3\n541.42\n13.54\n13.54\n568.49\nGrand\tTotal\n3\n541.42\n13.54\n13.54\n568.49\nQuantity\nBasic\nCGST\nSGST\nTotal\nNet\tAmount\n3\n541.42\n13.54\n13.54\n568.49"
        }
    ],
    "IN1-100CMBTS01_signed_compressed.pdf": [
        {
            "page_number": 1,
            "content": "1Sub Total \n(Tax Inclusive)25,000.00\nCGST9 (9%) 1,906.78\nSGST9 (9%) 1,906.78\nTotal \u20b925,000.00\nPayment Made (-) 25,000.00\nBalance Due \u20b90.00\nAuthorized SignatureTotal In Words\nIndian Rupee Twenty-Five Thousand Only\nNotes\nThanks for your business.\n100 Cubes Home Interiors\nKarnataka \nIndia\nGSTIN 29CITPK3346L2ZG\n9606472571\npatanabdulahadkhan@gmail.com\nTAX INVOICE\n  # : IN1-100CMBTS01\n  Invoice Date : 18/07/2024\n  Terms : Due on Receipt\n Due Date : 18/07/2024 Place Of Supply : Karnataka (29)\nBill To\nMr. Rajesh YS\n# Item & DescriptionHSN\n/SAC Qty RateCGST SGST\nAmount % Amt % Amt\n1 Booking fee - Home Interiors 998391 1.00 25,000.00 9% 1,906.78 9% 1,906.78 25,000.00\n"
        }
    ]
}

# Fields extraction function
def extract_invoice_data(content):
    fields = {
        "invoice_number": None,
        "invoice_date": None,
        "taxable_value": None,
        "cgst_amount": None,
        "sgst_amount": None,
        "total_amount": None,
        "place_of_supply": None,
        "place_of_origin": None,
        "gstin_supplier": None,
        "gstin_recipient": None,
    }

    # Regex patterns to extract data
    fields["invoice_number"] = re.search(r'Invoice No:\s*(\S+)', content)
    fields["invoice_date"] = re.search(r'Date:\s*(\S+)', content)
    fields["taxable_value"] = re.search(r'Taxable Value:\s*₹\s*([\d,.]+)', content)
    fields["cgst_amount"] = re.search(r'CGST:\s*₹\s*([\d,.]+)', content)
    fields["sgst_amount"] = re.search(r'SGST:\s*₹\s*([\d,.]+)', content)
    fields["total_amount"] = re.search(r'Total:\s*₹\s*([\d,.]+)', content)
    fields["place_of_supply"] = re.search(r'Place of Supply:\s*(\S+)', content)
    fields["place_of_origin"] = re.search(r'Place of Origin:\s*(\S+)', content)
    fields["gstin_supplier"] = re.search(r'GSTIN Supplier:\s*(\S+)', content)
    fields["gstin_recipient"] = re.search(r'GSTIN Recipient:\s*(\S+)', content)

    # Clean up and convert matches to strings
    for key, value in fields.items():
        if value:
            fields[key] = value.group(1)

    return fields

# Process each PDF's pages and extract fields
extracted_data = []
for pdf_file, pages in data.items():
    combined_content = " ".join([page["content"] for page in pages])  # Combine all pages' content
    extracted_data.append(extract_invoice_data(combined_content))

# Convert extracted data to a pandas DataFrame
df = pd.DataFrame(extracted_data)

# Save to Excel
df.to_excel("/content/extracted_invoices.xlsx", index=False)

print("Extraction complete. Data saved to Excel.")
```
